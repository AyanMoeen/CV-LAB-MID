## 1.1 Class-Based Collection Manager (add / remove / view / total + error handling)

### Source Task / Concept
Original lab task: `GroceryManager` with `add_item`, `remove_item`, `view_list`, `calculate_total` and error handling.

**Generalized problem:** Write a class that stores named records, supports add/remove/list, computes an aggregate, and guards against invalid operations.

### If the exam question says...
- create a class to manage a list/cart/inventory/library/bank/…
- implement add / remove / view / total methods
- prevent removing an item that does not exist

### Values you may need to change
- Class name and the stored fields (`quantity`, `price`) → rename to match the question (e.g. `stock`, `cost`, `balance`)
- The aggregate in `calculate_total` (sum, average, max…)

### Possible exam variations
- Inventory, library books, bank accounts, shopping cart, order tracker — same structure
- Extra method: `update_item`, `most_expensive()`, `search(name)`

### If it does not work immediately
- `KeyError` → you accessed a key without checking `if item in self.items`
- Wrong totals → check that you multiply quantity × price per item

```python
class ItemManager:
    """Generic manager. self.items = {name: {"quantity": q, "price": p}}"""

    def __init__(self):
        self.items = {}

    def add_item(self, item, quantity, price):
        if quantity <= 0 or price < 0:
            print("Error: quantity must be > 0 and price must be >= 0"); return
        if item in self.items:                                   # already there -> update
            self.items[item]["quantity"] += quantity
            self.items[item]["price"] = price
        else:
            self.items[item] = {"quantity": quantity, "price": price}
        print(f"Added {quantity} x {item} @ {price}")

    def remove_item(self, item):
        if item not in self.items:                               # error handling
            print(f"Error: '{item}' does not exist"); return
        del self.items[item]
        print(f"Removed {item}")

    def view_list(self):
        if not self.items:
            print("(list is empty)"); return
        print(f"{'Item':<15}{'Qty':>5}{'Price':>10}{'Subtotal':>12}")
        for name, d in self.items.items():
            print(f"{name:<15}{d['quantity']:>5}{d['price']:>10.2f}{d['quantity'] * d['price']:>12.2f}")

    def calculate_total(self):
        return sum(d["quantity"] * d["price"] for d in self.items.values())


# ---------- DEMO / USAGE (change values) ----------
m = ItemManager()
m.add_item("Apple", 5, 0.5)
m.add_item("Milk", 2, 1.8)
m.add_item("Apple", 3, 0.6)          # updates existing item
m.view_list()
m.remove_item("Bread")               # -> error message, no crash
m.remove_item("Milk")
m.view_list()
print("TOTAL =", m.calculate_total())
```

## 1.2 Nested-Dictionary Record System (best by average, search by field)

### Source Task / Concept
Original lab task: `{ID: {'Name','Major','Grades':[...]}}`, find the student with the highest average, search by major.

**Generalized problem:** Process a dictionary-of-dictionaries: compute per-record statistics, find the maximum, and filter by a field value.

### If the exam question says...
- nested dictionary / record system
- return the name with the highest average / total / max
- search or print all records matching a field

### Values you may need to change
- `records` contents
- `list_key` (`'Grades'`), `name_key` (`'Name'`), search `field` (`'Major'`)

### Possible exam variations
- Employees by department with salaries; products by category with prices; lowest average; rank all by average

### If it does not work immediately
- Empty grades list → guard against divide-by-zero (already done)
- Case mismatch in search → function lower-cases both sides

```python
students = {
    101: {"Name": "Ali",   "Major": "CS",   "Grades": [85, 90, 78]},
    102: {"Name": "Sara",  "Major": "EE",   "Grades": [92, 88, 95]},
    103: {"Name": "Omar",  "Major": "CS",   "Grades": [70, 65, 80]},
    104: {"Name": "Hina",  "Major": "Math", "Grades": []},
}

def average(values):
    return sum(values) / len(values) if values else 0.0

def top_by_average(records, list_key="Grades", name_key="Name"):
    """Return (name, average) of the record whose list has the highest average."""
    best_name, best_avg = None, float("-inf")
    for rid, info in records.items():
        avg = average(info[list_key])
        if avg > best_avg:
            best_name, best_avg = info[name_key], avg
    return best_name, best_avg

def search_by_field(records, field, value):
    """Print + return every record whose `field` equals `value` (case-insensitive)."""
    found = [(rid, info["Name"]) for rid, info in records.items()
             if str(info[field]).lower() == str(value).lower()]
    if not found: print(f"No records with {field} = {value}")
    for rid, name in found: print(f"ID {rid}: {name}  ({field} = {value})")
    return found

def rank_by_average(records, list_key="Grades", name_key="Name"):
    return sorted(((info[name_key], average(info[list_key])) for info in records.values()),
                  key=lambda t: t[1], reverse=True)

# ---------- USAGE ----------
print("Top:", top_by_average(students))
search_by_field(students, "Major", "cs")
print("Ranking:", rank_by_average(students))
```

## 1.4 Drawing Shapes on a Blank Canvas

### Source Task / Concept
Original lab task: 800×800 black canvas, 5 concentric alternating-color circles, bounding box around the outer circle, print the center.

**Generalized problem:** Create a canvas with NumPy and draw/annotate geometric primitives (circle, rectangle, line, polygon, text) at computed positions.

### If the exam question says...

### Values you may need to change

### Possible exam variations
- Checkerboard, bullseye, flag, grid of circles, draw box + label around a region

### If it does not work immediately
- Colors look wrong → they are **BGR**: red = (0,0,255), blue = (255,0,0)
- Nothing drawn → coordinates are (x, y), not (row, col)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
SIZE    = 800                                   # canvas is SIZE x SIZE
N_RINGS = 5
COLORS  = [(0, 0, 255), (255, 255, 255)]        # BGR, alternated ring by ring (outermost first)
BG      = (0, 0, 0)
BOX_COLOR, BOX_THICK = (0, 255, 0), 4
# ======================================================================

canvas = np.zeros((SIZE, SIZE, 3), np.uint8)
canvas[:] = BG
center = (SIZE // 2, SIZE // 2)                 # (x, y)
max_r  = SIZE // 2 - 20
step   = max_r // N_RINGS

for i in range(N_RINGS):                        # draw the BIGGEST first, then smaller on top
    r = max_r - i * step
    cv2.circle(canvas, center, r, COLORS[i % len(COLORS)], -1)

# bounding box around the outermost circle
top_left     = (center[0] - max_r, center[1] - max_r)
bottom_right = (center[0] + max_r, center[1] + max_r)
cv2.rectangle(canvas, top_left, bottom_right, BOX_COLOR, BOX_THICK)

# other primitives you may be asked for (uncomment)
# cv2.line(canvas, (0, 0), (SIZE - 1, SIZE - 1), (255, 255, 0), 3)
# cv2.polylines(canvas, [np.array([[100,100],[300,120],[250,300]], np.int32)], True, (0,255,255), 3)
# cv2.putText(canvas, "Label", (50, 60), cv2.FONT_HERSHEY_SIMPLEX, 2, (255,255,255), 3)

print("Center (x, y) used for drawing:", center)
print("Geometric center of the pixel grid:", ((SIZE - 1) / 2, (SIZE - 1) / 2))
print("Box corners:", top_left, bottom_right)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)); plt.title("Concentric target + bounding box"); plt.axis("off"); plt.show()
```

## 1.5 Blur an Image and Extract / Compare a Center ROI

### Source Task / Concept
Original lab task: 25×25 Gaussian blur, crop the exact-center 300×300 ROI from original and blurred, plot side by side.

**Generalized problem:** Apply a smoothing filter and use NumPy slicing to crop a region of interest (center or given coordinates) for comparison.

### If the exam question says...
- blur / smooth / denoise the image
- extract / crop / ROI of size H×W from the center (or from (x1,y1) to (x2,y2))
- compare original and filtered region side by side

### Values you may need to change
- `FILTER` (gaussian / median / box / bilateral)
- `KSIZE` (must be ODD)
- `ROI_H`, `ROI_W`
- For a non-center ROI set `USE_CENTER=False` and give `X1,Y1`

### Possible exam variations
- Median filter for salt-and-pepper noise; box blur; bilateral to keep edges; crop top-left / given rectangle

### If it does not work immediately
- Error on kernel → size must be odd (e.g. 5, 25)
- ROI empty → ROI bigger than image or x/y swapped (slicing is `img[y1:y2, x1:x2]`)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.jpg"
FILTER     = "gaussian"          # "gaussian" | "median" | "box" | "bilateral"
KSIZE      = 25                  # odd number
ROI_H, ROI_W = 300, 300
USE_CENTER = True
X1, Y1     = 0, 0                # top-left of ROI if USE_CENTER = False
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

img = load_img(IMAGE_PATH)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
h, w = img.shape[:2]

if FILTER == "gaussian":    blurred = cv2.GaussianBlur(img, (KSIZE, KSIZE), 0)
elif FILTER == "median":    blurred = cv2.medianBlur(img, KSIZE)
elif FILTER == "box":       blurred = cv2.blur(img, (KSIZE, KSIZE))
else:                       blurred = cv2.bilateralFilter(img, 9, 75, 75)

# ---- ROI by NumPy slicing: img[y1:y2, x1:x2] ----
rh, rw = min(ROI_H, h), min(ROI_W, w)
if USE_CENTER:
    y1, x1 = (h - rh) // 2, (w - rw) // 2
else:
    y1, x1 = Y1, X1
y2, x2 = y1 + rh, x1 + rw
roi_orig, roi_blur = img[y1:y2, x1:x2], blurred[y1:y2, x1:x2]
print(f"ROI rows {y1}:{y2}, cols {x1}:{x2}  -> shape {roi_orig.shape}")

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(roi_orig); ax[0].set_title("Original ROI", fontsize=14, color="navy")
ax[1].imshow(roi_blur); ax[1].set_title(f"{FILTER.title()} blur ROI (k={KSIZE})", fontsize=14, color="darkred")
for a in ax: a.axis("off")
plt.tight_layout(); plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH    = "exam_image.jpg"
BAND_FRACTION = 0.20                 # covers this fraction of the height
POSITION      = "bottom"             # "bottom" | "top"
OVERLAY_COLOR = (0, 0, 255)          # RGB (image is converted to RGB below) -> blue
ALPHA         = 0.5                  # overlay opacity
TEXT          = "My Caption"
FONT          = cv2.FONT_HERSHEY_SIMPLEX
FONT_SCALE, THICKNESS = 2.0, 3
TEXT_COLOR    = (255, 255, 255)
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

img = cv2.cvtColor(load_img(IMAGE_PATH), cv2.COLOR_BGR2RGB)
h, w = img.shape[:2]
band = int(h * BAND_FRACTION)
y1, y2 = (h - band, h) if POSITION == "bottom" else (0, band)

overlay = img.copy()
cv2.rectangle(overlay, (0, y1), (w, y2), OVERLAY_COLOR, -1)             # solid rectangle on a COPY
blended = cv2.addWeighted(overlay, ALPHA, img, 1 - ALPHA, 0)            # out = a*overlay + (1-a)*img

(tw, th), base = cv2.getTextSize(TEXT, FONT, FONT_SCALE, THICKNESS)
org = ((w - tw) // 2, y1 + (band + th) // 2)                             # centered in the band
cv2.putText(blended, TEXT, org, FONT, FONT_SCALE, TEXT_COLOR, THICKNESS, cv2.LINE_AA)

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].imshow(img);     ax[0].set_title("Original"); ax[0].axis("off")
ax[1].imshow(blended); ax[1].set_title(f"Alpha-blended caption (alpha={ALPHA})"); ax[1].axis("off")
plt.tight_layout(); plt.show()
```

### Source Task / Concept

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE1_PATH = "image1.jpg"           # shape is cut from this one
IMAGE2_PATH = "image2.jpg"           # used for the background
SIZE   = (500, 500)                  # (width, height) both images are resized to
SHAPE  = "circle"                    # "circle" | "polygon" | "rect"
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img


def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

img1 = cv2.resize(load_img(IMAGE1_PATH), SIZE)
img2 = cv2.resize(load_img(IMAGE2_PATH, variant=1), SIZE)
w, h = SIZE

mask = np.zeros((h, w), np.uint8)                                   # black image, single channel
if SHAPE == "circle":
    cv2.circle(mask, (w // 2, h // 2), min(h, w) // 3, 255, -1)     # filled white circle
elif SHAPE == "polygon":
    pts = np.array([[w//2, h//6], [5*w//6, h//2], [w//2, 5*h//6], [w//6, h//2]], np.int32)   # diamond
    cv2.fillPoly(mask, [pts], 255)
else:
    cv2.rectangle(mask, (w//4, h//4), (3*w//4, 3*h//4), 255, -1)

inside  = cv2.bitwise_and(img1, img1, mask=mask)                    # keep image1 where mask = 255
mask_inv = cv2.bitwise_not(mask)                                    # invert mask
outside = cv2.bitwise_and(img2, img2, mask=mask_inv)                # keep image2 where mask = 0
result  = cv2.bitwise_or(inside, outside)                           # merge (they never overlap)

fig, ax = plt.subplots(2, 3, figsize=(15, 9))
items = [(img1, "Image 1"), (img2, "Image 2"), (mask, "Mask"), (inside, "AND: image1 inside mask"),
         (outside, "AND: image2 outside (NOT mask)"), (result, "OR: final composite")]
for a, (im, t) in zip(ax.ravel(), items): show(a, im, t)
plt.tight_layout(); plt.show()
```

# PART 2 — LAB 2 (Photometric / point transformations)

**Key idea:** a *point transformation* changes each pixel's value using only that pixel: `s = T(r)` (r = input intensity, s = output). Histogram equalization is a *data-dependent* version of this (T is computed from the image's own cumulative histogram).

## 2.1 Histogram Analysis + Histogram Equalization

### Source Task / Concept
Original lab task: equalize a dark chest X-ray; compare equalized vs raw; also equalize CT and MRI independently; also equalize every video frame.

**Generalized problem:** Redistribute pixel intensities with the histogram's CDF to increase contrast of a low-contrast image, and visualize the histogram before/after.

### If the exam question says...
- improve / enhance contrast
- image is dark / underexposed / washed-out / low-contrast
- equalize the histogram
- plot / compare histograms before and after
- CLAHE / adaptive equalization
- implement equalization manually

### Values you may need to change
- `IMAGE_PATH`
- `IS_COLOR` (True equalizes brightness only, colors preserved)
- `METHOD` (`global` / `clahe`), `CLIP_LIMIT`, `TILE_GRID`

### Possible exam variations
- Equalize each RGB channel separately (changes colors — usually avoided); equalize a video frame (see 2.8)

### If it does not work immediately
- Over-amplified noise → use `clahe` with lower `CLIP_LIMIT`
- `equalizeHist` error → input must be single-channel `uint8`

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
IS_COLOR   = False            # True -> equalize only the brightness (Y) channel, keep colors
METHOD     = "global"         # "global" = cv2.equalizeHist | "clahe" = local/adaptive
CLIP_LIMIT = 2.0
TILE_GRID  = (8, 8)
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

def equalize(ch):
    if METHOD == "clahe":
        return cv2.createCLAHE(clipLimit=CLIP_LIMIT, tileGridSize=TILE_GRID).apply(ch)
    return cv2.equalizeHist(ch)

def plot_hist(ax, ch, title):
    h = np.bincount(ch.ravel(), minlength=256)
    ax.bar(range(256), h, width=1.0, color="gray"); ax.set_xlim(0, 255); ax.set_title(title)
    ax2 = ax.twinx(); ax2.plot(np.cumsum(h) / h.sum(), color="red"); ax2.set_ylim(0, 1.05)   # red = CDF

img = load_img(IMAGE_PATH, gray=not IS_COLOR)
if IS_COLOR:
    ycc = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
    before = ycc[:, :, 0].copy()
    after = equalize(before)
    ycc[:, :, 0] = after
    result = cv2.cvtColor(ycc, cv2.COLOR_YCrCb2BGR)
else:
    before = img.copy()
    after = equalize(before)
    result = after

# ---- manual equalization (what equalizeHist does): s = round((CDF(r) - CDFmin) / (N - CDFmin) * 255) ----
hist = np.bincount(before.ravel(), minlength=256)
cdf = hist.cumsum(); cdf_min = cdf[cdf > 0].min()
lut = np.round((cdf - cdf_min) / (before.size - cdf_min) * 255).clip(0, 255).astype(np.uint8)
manual = lut[before]
if METHOD == "global":
    print("Max |manual - cv2.equalizeHist| =", int(np.abs(manual.astype(int) - after.astype(int)).max()))

fig, ax = plt.subplots(2, 2, figsize=(12, 9))
show(ax[0, 0], img, "Original"); show(ax[0, 1], result, f"Equalized ({METHOD})")
plot_hist(ax[1, 0], before, "Histogram BEFORE (red = CDF)"); plot_hist(ax[1, 1], after, "Histogram AFTER (red = CDF)")
plt.tight_layout(); plt.show()
```

## 2.2 Logarithmic and Power-Law (Gamma) Transformations

### Source Task / Concept
Original lab task: log transform to expose faint dark detail; gamma < 1 to fine-tune midtones / suppress bright regions; used again on fused images and video frames.

**Generalized problem:** Apply a non-linear intensity mapping to brighten dark regions (log, γ<1) or darken/compress bright regions (γ>1).

### If the exam question says...
- log transformation / logarithmic
- gamma / power-law / gamma correction
- expand dark regions, reveal detail in shadows
- compress bright regions / suppress blown-out whites
- brighten / darken non-linearly

### Values you may need to change
- `GAMMA` (<1 brightens, >1 darkens)
- `AS_GRAY` (False = apply to color image)
- `GAMMAS` list for the comparison sweep

### Possible exam variations
- Compare several γ values; apply log then gamma (pipeline); apply to each frame of a video (2.8); inverse-log

### If it does not work immediately
- Log makes everything too bright → follow with gamma > 1
- Output black/white → check values were scaled back to 0–255 and cast to `uint8`

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
AS_GRAY    = True
GAMMA      = 0.5                       # <1 brighten shadows / midtones, >1 darken
GAMMAS     = [0.3, 0.5, 1.0, 2.0, 3.0] # for the comparison sweep
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

img = load_img(IMAGE_PATH, gray=AS_GRAY)
f = img.astype(np.float32)

# ---- Log:  s = c * log(1 + r),  c = 255 / log(1 + max)  ----
c = 255.0 / np.log1p(f.max())
log_img = (c * np.log1p(f)).clip(0, 255).astype(np.uint8)

# ---- Gamma: s = 255 * (r/255)^gamma  (use a lookup table; fast + works on video) ----
def gamma_lut(g):
    return np.array([255.0 * (i / 255.0) ** g for i in range(256)]).clip(0, 255).astype(np.uint8)
gamma_img = cv2.LUT(img, gamma_lut(GAMMA))

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
show(ax[0], img, "Original"); show(ax[1], log_img, "Log transform"); show(ax[2], gamma_img, f"Gamma = {GAMMA}")
plt.tight_layout(); plt.show()

# ---- gamma sweep + transfer curves ----
fig, ax = plt.subplots(1, len(GAMMAS) + 1, figsize=(4 * (len(GAMMAS) + 1), 4))
for a, g in zip(ax, GAMMAS): show(a, cv2.LUT(img, gamma_lut(g)), f"gamma={g}")
x = np.arange(256)
for g in GAMMAS: ax[-1].plot(x, gamma_lut(g), label=f"γ={g}")
ax[-1].plot(x, c * np.log1p(x), "k--", label="log"); ax[-1].legend(); ax[-1].set_title("Transfer curves s = T(r)")
ax[-1].set_xlabel("input r"); ax[-1].set_ylabel("output s")
plt.tight_layout(); plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH  = "exam_image.png"
# piecewise-linear control points (r_in, s_out): steeper segment = more contrast in that range
BREAKPOINTS = [(0, 0), (70, 30), (140, 200), (255, 255)]
P_LOW, P_HIGH = 2, 98                   # percentiles used for contrast stretching
A, B        = 100, 180                  # gray-level slicing band
SLICE_HIGH  = 255                       # value given to pixels inside [A, B]
KEEP_OUTSIDE = True                     # True: keep other pixels unchanged, False: set them to 0
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

gray = load_img(IMAGE_PATH, gray=True)

xs, ys = zip(*BREAKPOINTS)
lut = np.interp(np.arange(256), xs, ys).astype(np.uint8)          # linear interpolation between breakpoints
piecewise = cv2.LUT(gray, lut)

negative = 255 - gray                                              # s = (L-1) - r

lo, hi = np.percentile(gray, (P_LOW, P_HIGH))                      # robust min/max
stretch = np.clip((gray.astype(np.float32) - lo) * 255.0 / max(hi - lo, 1), 0, 255).astype(np.uint8)

inside = (gray >= A) & (gray <= B)
slicing = np.where(inside, SLICE_HIGH, gray if KEEP_OUTSIDE else 0).astype(np.uint8)

fig, ax = plt.subplots(2, 3, figsize=(15, 9))
show(ax[0, 0], gray, "Original");               show(ax[0, 1], negative, "Negative")
show(ax[0, 2], piecewise, "Piecewise-linear");  show(ax[1, 0], stretch, f"Contrast stretch ({P_LOW}-{P_HIGH} pct)")
show(ax[1, 1], slicing, f"Slicing [{A},{B}]")
ax[1, 2].plot(np.arange(256), lut, label="piecewise"); ax[1, 2].plot(np.arange(256), np.arange(256), "k:", label="identity")
ax[1, 2].plot(xs, ys, "ro"); ax[1, 2].legend(); ax[1, 2].set_title("Piecewise transfer curve"); ax[1, 2].set_xlabel("r"); ax[1, 2].set_ylabel("s")
plt.tight_layout(); plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
USE_HEATMAP_FIRST = False                       # True: gray -> equalize -> JET heatmap, then balance that
B_GAIN, G_GAIN, R_GAIN = 1.20, 1.00, 0.85       # multiply each channel (BGR order)
B_OFF,  G_OFF,  R_OFF  = 0, 0, 0                # add to each channel
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

if USE_HEATMAP_FIRST:
    g = load_img(IMAGE_PATH, gray=True)
    img = cv2.applyColorMap(cv2.equalizeHist(g), cv2.COLORMAP_JET)
else:
    img = load_img(IMAGE_PATH)

gains   = np.array([B_GAIN, G_GAIN, R_GAIN], np.float32)
offsets = np.array([B_OFF, G_OFF, R_OFF], np.float32)
manual = np.clip(img.astype(np.float32) * gains + offsets, 0, 255).astype(np.uint8)

# gray-world automatic balance: force the three channel means to be equal
means = img.reshape(-1, 3).mean(axis=0)                           # [mean_B, mean_G, mean_R]
auto_gain = means.mean() / means
auto = np.clip(img.astype(np.float32) * auto_gain, 0, 255).astype(np.uint8)
print("Channel means before (B,G,R):", means.round(1))
print("Channel means manual  (B,G,R):", manual.reshape(-1, 3).mean(axis=0).round(1))
print("Channel means auto    (B,G,R):", auto.reshape(-1, 3).mean(axis=0).round(1))

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
show(ax[0], img, "Original"); show(ax[1], manual, "Manual balance"); show(ax[2], auto, "Gray-world auto balance")
plt.tight_layout(); plt.show()
```

## 2.6 Threshold / Mask Dense (Bright) Regions or a Color Range

**Generalized problem:** Segment pixels by intensity (or by color range) and keep/mask only the selected region.

### If the exam question says...

### Values you may need to change

### If it does not work immediately

- Hue range in OpenCV is **0–179** (not 360)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
MODE = "binary"           # "binary" | "tozero" | "otsu" | "percentile" | "range"
T    = 180                # threshold for binary / tozero
PCT  = 10                 # percentile mode: keep the brightest PCT percent
LOW, HIGH = 100, 200      # range mode: keep pixels in [LOW, HIGH]
# --- part B: color-range segmentation (HSV, hue 0-179) ---
HSV_LOWER = (35, 50, 50)  # green-ish example
HSV_UPPER = (85, 255, 255)
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

# ----------------------- PART A: intensity thresholding -----------------------
gray = load_img(IMAGE_PATH, gray=True)
if MODE == "binary":      _, mask = cv2.threshold(gray, T, 255, cv2.THRESH_BINARY)
elif MODE == "tozero":    _, mask = cv2.threshold(gray, T, 255, cv2.THRESH_TOZERO)     # keeps original values above T
elif MODE == "otsu":      T, mask = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
elif MODE == "percentile":
    T = np.percentile(gray, 100 - PCT); _, mask = cv2.threshold(gray, T, 255, cv2.THRESH_BINARY)
else:                     mask = cv2.inRange(gray, LOW, HIGH)

binary_mask = (mask > 0).astype(np.uint8) * 255
segmented = cv2.bitwise_and(gray, gray, mask=binary_mask)                   # keep original intensities inside mask
print(f"Threshold used: {T} | pixels kept: {100 * np.count_nonzero(binary_mask) / binary_mask.size:.2f}%")

# ----------------------- PART B: color-range segmentation -----------------------
color = load_img(IMAGE_PATH)
hsv = cv2.cvtColor(color, cv2.COLOR_BGR2HSV)
cmask = cv2.inRange(hsv, np.array(HSV_LOWER), np.array(HSV_UPPER))
cseg = cv2.bitwise_and(color, color, mask=cmask)

fig, ax = plt.subplots(2, 3, figsize=(15, 9))
show(ax[0, 0], gray, "Gray"); show(ax[0, 1], mask, f"Mask ({MODE})"); show(ax[0, 2], segmented, "Segmented (gray)")
show(ax[1, 0], color, "Color image"); show(ax[1, 1], cmask, "HSV range mask"); show(ax[1, 2], cseg, "Segmented (color)")
plt.tight_layout(); plt.show()
```

## 2.8 Per-Frame Enhancement Pipeline on a Video (Raw vs Processed, Side by Side)

### Source Task / Concept
Original lab task: `VideoCapture` on an echocardiogram `.mp4`; per frame apply equalization → heatmap → color balance → log → gamma; show raw and enhanced side by side.

**Generalized problem:** Read a video (file or webcam) frame by frame, apply an image-processing pipeline to each frame, and display / save the input and output together.

### If the exam question says...
- video / frames / real-time processing
- VideoCapture / read from .mp4 / webcam
- apply the enhancement to each frame
- display raw and processed side by side
- save the processed video

### Values you may need to change
- `VIDEO_PATH` (or `0` for webcam)
- `STEPS` flags (turn pipeline stages on/off)
- `GAMMA`, `COLORMAP`, `GAINS`
- `DISPLAY` (`notebook` or `window`), `MAX_FRAMES`, `SAVE_PATH`

### Possible exam variations
- Edge detection per frame; blur / denoise; threshold per frame; grayscale-only; resize frames; different pipeline order

### If it does not work immediately
- `cap.isOpened()` False → wrong path / codec; (code falls back to a demo stream and warns)
- `hstack` error → both frames must have the same height & channels (code converts to 3-channel)
- `cv2.imshow` freezing in Jupyter → use `DISPLAY = 'notebook'`; press **q** to quit in window mode

```python
import cv2
import numpy as np
from IPython.display import display, Image, clear_output

# ==================== USER SETTINGS — CHANGE THESE ====================
VIDEO_PATH = "echo.mp4"          # file path, or 0 for webcam
STEPS = dict(equalize=True, heatmap=True, balance=True, log=True, gamma=True)   # switch stages on/off
COLORMAP = cv2.COLORMAP_JET
GAINS    = (1.1, 1.0, 0.9)       # (B, G, R) color-balance gains
GAMMA    = 1.5                   # >1 suppresses bright backscatter, <1 brightens
DISPLAY  = "notebook"            # "notebook" (inline) | "window" (cv2.imshow, desktop python)
SHOW_EVERY, MAX_FRAMES = 10, 200 # notebook mode: show every Nth frame, stop after MAX_FRAMES
DISPLAY_WIDTH = 900              # shrink the side-by-side image for display
SAVE_PATH = None                 # e.g. "output/result.mp4" to save the side-by-side video
# ======================================================================

LOG_C = 255.0 / np.log1p(255.0)                                          # fixed c -> no flicker between frames
GAMMA_LUT = (255.0 * (np.arange(256) / 255.0) ** GAMMA).clip(0, 255).astype(np.uint8)

def enhance(frame):
    """Pipeline applied to ONE frame (edit/reorder stages here)."""
    out = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    if STEPS["equalize"]: out = cv2.equalizeHist(out)
    out = cv2.applyColorMap(out, COLORMAP) if STEPS["heatmap"] else cv2.cvtColor(out, cv2.COLOR_GRAY2BGR)
    if STEPS["balance"]:  out = np.clip(out.astype(np.float32) * np.array(GAINS, np.float32), 0, 255).astype(np.uint8)
    if STEPS["log"]:      out = (LOG_C * np.log1p(out.astype(np.float32))).clip(0, 255).astype(np.uint8)
    if STEPS["gamma"]:    out = cv2.LUT(out, GAMMA_LUT)
    return out

# ---- video source (falls back to a synthetic stream if the file can't be opened) ----
cap = cv2.VideoCapture(VIDEO_PATH)
use_demo = not cap.isOpened()
n_demo = 0
if use_demo: print(f"[WARN] Could not open '{VIDEO_PATH}' -> using synthetic demo frames. Fix VIDEO_PATH!")

def read_frame():
    global n_demo
    if use_demo:
        if n_demo >= 60: return False, None
        f = np.full((240, 320, 3), 40, np.uint8)
        cv2.circle(f, (30 + 4 * n_demo, 120), 40, (140, 140, 140), -1)
        n_demo += 1
        return True, f
    return cap.read()

writer, count = None, 0
try:
    while True:
        ok, frame = read_frame()
        if not ok: break                                              # end of video
        proc = enhance(frame)
        raw = frame if frame.ndim == 3 else cv2.cvtColor(frame, cv2.COLOR_GRAY2BGR)
        side = np.hstack([raw, proc])                                 # raw LEFT | enhanced RIGHT
        cv2.putText(side, "RAW", (10, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 255), 2)
        cv2.putText(side, "ENHANCED", (raw.shape[1] + 10, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 255), 2)

        if SAVE_PATH:
            if writer is None:
                fps = cap.get(cv2.CAP_PROP_FPS) if not use_demo else 20
                writer = cv2.VideoWriter(SAVE_PATH, cv2.VideoWriter_fourcc(*"mp4v"), fps or 20, (side.shape[1], side.shape[0]))
            writer.write(side)

        if DISPLAY == "window":
            cv2.imshow("Raw | Enhanced  (press q to quit)", side)
            if cv2.waitKey(25) & 0xFF == ord("q"): break
        elif count % SHOW_EVERY == 0:
            scale = DISPLAY_WIDTH / side.shape[1]
            small = cv2.resize(side, None, fx=scale, fy=scale) if scale < 1 else side
            clear_output(wait=True)
            display(Image(data=cv2.imencode(".jpg", small)[1].tobytes()))
        count += 1
        if DISPLAY == "notebook" and count >= MAX_FRAMES: break
finally:
    cap.release()
    if writer is not None: writer.release()
    cv2.destroyAllWindows()
print("Frames processed:", count)
```

# PART 3 — LAB 3 (Geometric transformations)

## The transformation hierarchy (know this for MCQs)

| Level | Matrix form | Degrees of freedom | Preserves | cv2 function |
|---|---|---|---|---|
| Translation | `[I \| t]` | 2 | orientation, size, shape (everything) | `warpAffine` |
| **Rigid / Euclidean** | `[R \| t]` | 3 (θ, tx, ty) | lengths, angles, areas | `warpAffine` |
| **Similarity** | `[sR \| t]` | 4 (s, θ, tx, ty) | angles, ratios of lengths (shape) | `warpAffine` |
| **Affine** | `[A \| t]`, A any 2×2 | 6 | parallel lines, ratios along lines | `warpAffine`, `getAffineTransform` |
| **Projective (Homography)** | full 3×3, last entry = 1 | 8 | straight lines only (parallels may converge) | `warpPerspective`, `getPerspectiveTransform`, `findHomography` |

**Linear (2×2) building blocks:** Scale `diag(sx,sy)` · Rotation `[[cosθ, −sinθ],[sinθ, cosθ]]` (CCW in math axes; with y pointing down use `[[cos, sin],[−sin, cos]]` as cv2 does) · Shear `[[1,kx],[ky,1]]`.
Linear maps fix the **origin**, so translation needs the **3×3 homogeneous** form `[[a,b,tx],[c,d,ty],[0,0,1]]`.

**Point counts:** affine needs **3** point pairs, perspective needs **4**.
**Order matters:** `M = M3 @ M2 @ M1` applies M1 first. Rotating then translating ≠ translating then rotating.
**Rotation/scaling about the center:** `T(c) · R · S · T(−c)`.

## 3.1 Scaling (Enlarge / Shrink) and Keep Centered

### Source Task / Concept
Original lab task: manually build a 2×2 scaling matrix, enlarge a fingerprint by 300% in x and y, and keep it centered after it grows.

**Generalized problem:** Apply a scale matrix about a chosen point (image center) either on the same-size canvas (cropped) or on a canvas that grows.

### If the exam question says...
- scale / enlarge / zoom / shrink by X% (x and y independently)
- build the scaling matrix manually
- keep the image centered after scaling
- non-uniform scaling / aspect-ratio change

### Values you may need to change

- `MODE`: `same_canvas` (centered, cropped if enlarging) or `full_canvas` (output grows)

### Possible exam variations
- Shrink (SX=SY=0.5 and place centered in the same canvas); stretch only horizontally; equivalent `cv2.resize`

### If it does not work immediately
- Image appears shifted after scaling → scaling is about the top-left origin; need `t = c − S·c` (done below)
- Shrunk image has black borders → expected (empty canvas)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
SX, SY = 3.0, 3.0
MODE   = "same_canvas"           # "same_canvas" | "full_canvas"
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

img = load_img(IMAGE_PATH)
h, w = img.shape[:2]

S = np.array([[SX, 0],
              [0, SY]], float)                          # 2x2 scaling matrix (linear part)

if MODE == "same_canvas":
    c = np.array([w / 2, h / 2])                        # image center (x, y)
    t = c - S @ c                                       # translation that keeps the center fixed on screen
    M = np.hstack([S, t.reshape(2, 1)])                 # 2x3 affine matrix [S | t]
    out = cv2.warpAffine(img, M, (w, h))                # dsize = (width, height)
else:
    M = np.hstack([S, np.zeros((2, 1))])
    out = cv2.warpAffine(img, M, (int(round(w * SX)), int(round(h * SY))))

print("Matrix used:\n", M, "\nOutput shape:", out.shape)
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
show(ax[0], img, f"Original {w}x{h}"); show(ax[1], out, f"Scaled ({SX}, {SY}) - {MODE}")
plt.tight_layout(); plt.show()
```

## 3.2 Rotation Without Cropping (Manual Matrix + New Canvas Size)

### Source Task / Concept
Original lab task: build a 2×2 rotation matrix with sin/cos, spin a 45°-rotated satellite image back, and compute new dimensions so corners are not cut off.

**Generalized problem:** Rotate an image about its center by an arbitrary angle and compute the enlarged canvas so the whole rotated image is visible.

### If the exam question says...
- rotate by θ degrees / spin back / correct the tilt
- rotation matrix using sine and cosine
- do not cut off the corners / new dimensions
- rotate about the center

### Values you may need to change
- `ANGLE_DEG` (positive = counter-clockwise; to undo a +45° tilt use −45)
- `EXPAND` (True = enlarged canvas, False = keep original size)

### Possible exam variations
- Rotate about an arbitrary point; rotate by 90°/180° with `np.rot90` or `cv2.rotate`; rotate + scale (`getRotationMatrix2D(c, angle, scale)`)

### If it does not work immediately
- Rotated the wrong way → flip the sign of the angle
- New width/height formula: `W = w|cosθ| + h|sinθ|`, `H = w|sinθ| + h|cosθ|`

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
ANGLE_DEG  = 45              # + = counter-clockwise, - = clockwise
EXPAND     = True            # True: enlarge canvas so no corner is cut
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

img = load_img(IMAGE_PATH)
h, w = img.shape[:2]

th = np.deg2rad(ANGLE_DEG)
cs, sn = np.cos(th), np.sin(th)
R = np.array([[cs, sn],
              [-sn, cs]])                                       # 2x2 rotation (image coords, y down; same as cv2)

if EXPAND:
    new_w = int(np.ceil(w * abs(cs) + h * abs(sn)))             # manual new dimensions
    new_h = int(np.ceil(w * abs(sn) + h * abs(cs)))
else:
    new_w, new_h = w, h

center     = np.array([w / 2, h / 2])
new_center = np.array([new_w / 2, new_h / 2])
t = new_center - R @ center                                     # rotate about the center, then re-center on new canvas
M = np.hstack([R, t.reshape(2, 1)])                             # 2x3
out = cv2.warpAffine(img, M, (new_w, new_h))

# sanity check against OpenCV's own matrix
M_cv = cv2.getRotationMatrix2D((w / 2, h / 2), ANGLE_DEG, 1.0)
print("Linear part equals cv2.getRotationMatrix2D:", np.allclose(M[:, :2], M_cv[:, :2]))
print(f"Original {w}x{h} -> new canvas {new_w}x{new_h}")

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
show(ax[0], img, "Original"); show(ax[1], out, f"Rotated {ANGLE_DEG}° ({'expanded' if EXPAND else 'same'} canvas)")
plt.tight_layout(); plt.show()
```

## 3.3 Shear (Un-Slant an Image)

### Source Task / Concept
Original lab task: 2×2 shear matrix to push pixels horizontally and straighten a slanted barcode.

**Generalized problem:** Apply a shear (horizontal and/or vertical) to correct or create slanting, enlarging the canvas so nothing is lost.

### If the exam question says...
- shear / skew / slant / lean
- straighten slanted text or barcode
- build a shear matrix
- horizontal shear by k or by an angle

### Values you may need to change
- `SHX` (x' = x + SHX·y), `SHY` (y' = y + SHY·x)
- Or set `SHX = np.tan(np.deg2rad(angle))`
- Sign: lines leaning like **/** need **positive** SHX (lower rows pushed right); lines leaning like **\** need **negative**

### Possible exam variations

### If it does not work immediately

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
SHX = 0.3                    # horizontal shear: x' = x + SHX * y   (try negative if wrong direction)
SHY = 0.0                    # vertical shear:   y' = y + SHY * x
# SHX = np.tan(np.deg2rad(20))   # <- alternative: shear by an angle
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

def warp_affine_full(img, M, border=0):
    """Warp with a 2x3 matrix; automatically enlarge + shift the canvas so nothing is cut off."""
    h, w = img.shape[:2]
    corners = np.array([[0, 0, 1], [w, 0, 1], [w, h, 1], [0, h, 1]], float).T
    pts = M @ corners
    mn, mx = pts.min(axis=1), pts.max(axis=1)
    M2 = M.astype(float).copy(); M2[:, 2] += -mn
    new_w, new_h = np.ceil(mx - mn).astype(int)
    return cv2.warpAffine(img, M2, (int(new_w), int(new_h)), borderValue=border), M2

img = load_img(IMAGE_PATH)

A = np.array([[1, SHX],
              [SHY, 1]], float)                                 # 2x2 shear matrix
M = np.hstack([A, np.zeros((2, 1))])
out, M_used = warp_affine_full(img, M)                          # canvas enlarged automatically

print("Shear matrix:\n", A, "\nOutput shape:", out.shape)
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
show(ax[0], img, "Input"); show(ax[1], out, f"Sheared (SHX={SHX:.3f}, SHY={SHY:.3f})")
plt.tight_layout(); plt.show()
```

## 3.5 Combine Transformations into ONE Matrix (Rigid & Similarity)

### Source Task / Concept
Original lab tasks: (Task 5) combine rotation + translation into a single 3×3 rigid matrix to align a chip; (Task 6) combine scale + rotation + translation into a single similarity matrix.

**Generalized problem:** Compose several elementary transforms by matrix multiplication and apply them in one warp.

### If the exam question says...

- similarity transformation (scale + rotation + translation)
- combine into a single 3×3 matrix
- apply simultaneously / in one step
- preserve shape/angles

### Values you may need to change
- `MODE` (`rigid` forces scale = 1)
- `ANGLE`, `SCALE`, `TX`, `TY`
- `ABOUT_CENTER` (rotate/scale about image center vs origin)
- `OUT_SIZE` (None = same as input)

### Possible exam variations

- Wrong result → matrix order: rightmost is applied first (`T @ R @ S` = scale, then rotate, then translate)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH   = "exam_image.png"
MODE         = "similarity"   # "rigid" (no scaling) | "similarity" (scale + rotate + translate)
ANGLE        = 30             # degrees, + = counter-clockwise
SCALE        = 0.8
TX, TY       = 60, -20
ABOUT_CENTER = True
OUT_SIZE     = None           # (width, height) or None
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

# ---------- 3x3 homogeneous building blocks (image coords: x right, y DOWN) ----------
def T(tx, ty):   return np.array([[1, 0, tx], [0, 1, ty], [0, 0, 1]], float)
def S(sx, sy):   return np.array([[sx, 0, 0], [0, sy, 0], [0, 0, 1]], float)
def R(deg):                                    # positive = counter-clockwise (same as cv2.getRotationMatrix2D)
    a = np.deg2rad(deg); c, s = np.cos(a), np.sin(a)
    return np.array([[c, s, 0], [-s, c, 0], [0, 0, 1]], float)
def SH(kx, ky):  return np.array([[1, kx, 0], [ky, 1, 0], [0, 0, 1]], float)

img = load_img(IMAGE_PATH)
h, w = img.shape[:2]
s = 1.0 if MODE == "rigid" else SCALE

core = R(ANGLE) @ S(s, s)                                       # scale first, then rotate
if ABOUT_CENTER:
    cx, cy = w / 2, h / 2
    core = T(cx, cy) @ core @ T(-cx, -cy)                       # do it about the image center
M = T(TX, TY) @ core                                            # ... and finally translate  (ONE 3x3 matrix)
print(f"{MODE.upper()} matrix:\n", M.round(4))

out = cv2.warpAffine(img, M[:2], OUT_SIZE or (w, h))

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
show(ax[0], img, "Input"); show(ax[1], out, f"{MODE}: angle={ANGLE}, scale={s}, t=({TX},{TY})")
plt.tight_layout(); plt.show()
```

## 3.6 General Affine Matrix from 3 Point Correspondences

### Source Task / Concept
Original lab task: given start and end coordinates of 3 landmarks in a squished/slanted/rotated/moved photo, solve a linear system for the 6 unknowns and fix the image.

**Generalized problem:** Estimate the affine transform that maps 3 source points to 3 destination points, then warp the image with it.

### If the exam question says...
- given 3 points and where they should end up
- solve a system of linear equations for the affine matrix
- 6 unknowns
- undo a combination of scale/rotation/shear/translation
- cv2.getAffineTransform

### Values you may need to change
- `IMAGE_PATH`
- `SRC_PTS` = landmark coordinates **in the distorted image**
- `DST_PTS` = where they should be **in the corrected image**
- `OUT_SIZE`

### Possible exam variations
- More than 3 points → `cv2.estimateAffine2D(src, dst)` (least squares/RANSAC); map a point with M·[x,y,1]; invert the matrix to go back

### If it does not work immediately
- Image moves wrong way → you swapped SRC and DST
- `getAffineTransform` error → points must be `float32`, shape (3,2), not collinear

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
SRC_PTS = [[50, 50], [200, 50], [50, 200]]        # (x, y) landmarks in the DISTORTED image
DST_PTS = [[10, 100], [200, 50], [100, 250]]      # (x, y) where they must land (corrected image)
OUT_SIZE = None                                   # (width, height); None = same as input
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

img = load_img(IMAGE_PATH)
h, w = img.shape[:2]
src = np.array(SRC_PTS, np.float32); dst = np.array(DST_PTS, np.float32)

# ---- manual solution: x' = a x + b y + c ;  y' = d x + e y + f  -> 6 equations, 6 unknowns [a b c d e f] ----
A, b = [], []
for (x, y), (u, v) in zip(src, dst):
    A.append([x, y, 1, 0, 0, 0]); b.append(u)
    A.append([0, 0, 0, x, y, 1]); b.append(v)
params = np.linalg.solve(np.array(A, float), np.array(b, float))
M_manual = params.reshape(2, 3)

M_cv = cv2.getAffineTransform(src, dst)                         # same thing in one call
print("Manual solve:\n", M_manual.round(4), "\ncv2.getAffineTransform:\n", M_cv.round(4))
print("Match:", np.allclose(M_manual, M_cv, atol=1e-3))

out = cv2.warpAffine(img, M_manual, OUT_SIZE or (w, h))

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
show(ax[0], img, "Input (SRC points in red)");    ax[0].scatter(src[:, 0], src[:, 1], c="red", s=60)
show(ax[1], out, "Corrected (DST points in lime)"); ax[1].scatter(dst[:, 0], dst[:, 1], c="lime", s=60)
plt.tight_layout(); plt.show()
```

## 3.7 Perspective Correction (Bird's-Eye View / Straighten a Planar Object)

### Source Task / Concept
Original lab task: pick 4 corners of chalk art that should form a perfect square and warp to a top-down view.

**Generalized problem:** Map 4 points that form a quadrilateral in the image to a rectangle (or square) using a projective transform.

### If the exam question says...
- bird's-eye / top-down view
- straighten / rectify / fix perspective / de-skew a photographed document, sign, poster, painting, screen
- 4 points → perspective transform
- far edge looks smaller
- getPerspectiveTransform / warpPerspective

### Values you may need to change
- `SRC_PTS` (4 (x,y) corners; any order, the code orders them)
- `OUT_W, OUT_H` (None = automatic from edge lengths; for a square set both equal)
- `FORCE_SQUARE`

### Possible exam variations
- Solve the homography manually (shown); choose a custom DST rectangle; crop only the region inside the 4 points; inverse warp (`flags=cv2.WARP_INVERSE_MAP`)

### If it does not work immediately
- Result mirrored/twisted → points in wrong order (code orders them TL,TR,BR,BL)
- Need coordinates? → template 4.4
- Result blurry → increase OUT_W/OUT_H

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
SRC_PTS = [[120, 60], [480, 90], [540, 340], [60, 320]]   # 4 corners (x, y) in the input image
OUT_W, OUT_H = None, None                                  # None -> computed automatically
FORCE_SQUARE = False                                       # True -> OUT_W = OUT_H
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

def order_points(pts):
    """Return corners ordered: top-left, top-right, bottom-right, bottom-left."""
    pts = np.array(pts, dtype=np.float32)
    s = pts.sum(axis=1); d = np.diff(pts, axis=1).ravel()      # d = y - x
    return np.array([pts[np.argmin(s)], pts[np.argmin(d)], pts[np.argmax(s)], pts[np.argmax(d)]], np.float32)

def homography_from_points(src, dst):
    """Solve the 8 unknowns of a homography from 4 point pairs (manual DLT, 8x8 linear system)."""
    A, b = [], []
    for (x, y), (u, v) in zip(src, dst):
        A.append([x, y, 1, 0, 0, 0, -x * u, -y * u]); b.append(u)
        A.append([0, 0, 0, x, y, 1, -x * v, -y * v]); b.append(v)
    h = np.linalg.solve(np.array(A, float), np.array(b, float))
    return np.append(h, 1).reshape(3, 3)

img = load_img(IMAGE_PATH)
src = order_points(SRC_PTS)                                      # tl, tr, br, bl
tl, tr, br, bl = src

if OUT_W is None or OUT_H is None:
    OUT_W = int(max(np.linalg.norm(br - bl), np.linalg.norm(tr - tl)))      # longest top/bottom edge
    OUT_H = int(max(np.linalg.norm(tr - br), np.linalg.norm(tl - bl)))      # longest left/right edge
if FORCE_SQUARE: OUT_W = OUT_H = max(OUT_W, OUT_H)

dst = np.array([[0, 0], [OUT_W - 1, 0], [OUT_W - 1, OUT_H - 1], [0, OUT_H - 1]], np.float32)

H_cv = cv2.getPerspectiveTransform(src, dst)                     # 3x3 homography
H_man = homography_from_points(src, dst)                         # manual: 8 equations, 8 unknowns
print("Homography:\n", H_cv.round(5), "\nManual solution matches:", np.allclose(H_cv, H_man, atol=1e-3))

out = cv2.warpPerspective(img, H_cv, (OUT_W, OUT_H))             # dsize = (width, height)

fig, ax = plt.subplots(1, 2, figsize=(14, 6))
show(ax[0], img, "Input with selected corners")
ax[0].plot(*np.vstack([src, src[:1]]).T, "r-o", lw=2)
show(ax[1], out, f"Rectified {OUT_W}x{OUT_H}")
plt.tight_layout(); plt.show()
```

## 3.8 Stitch Two Images into a Panorama (Homography)

### Source Task / Concept
Original lab task: pick 4 matching points visible in both camera images, compute the homography projecting camera 2 into camera 1's frame, and stitch.

**Generalized problem:** Warp image 2 into image 1's coordinate system using point correspondences and merge them on a canvas large enough for both.

### If the exam question says...
- stitch / panorama / merge two overlapping photos
- homography from matching points
- project the second image into the first one's space
- automatic feature matching (ORB + RANSAC)

### Values you may need to change
- `IMG1_PATH` (reference), `IMG2_PATH` (to be warped)
- `PTS1` points in image 1; `PTS2` the SAME physical points in image 2 (same order!)
- `USE_AUTO_POINTS=True` to find matches automatically

### Possible exam variations
- Stitch 3 images (stitch 1+2, then result+3); use `cv2.Stitcher_create()` as an automatic alternative; SIFT instead of ORB

### If it does not work immediately
- Distorted output → matches inaccurate; use ≥ 4 well-spread points (more points + RANSAC is more robust)
- Black canvas → swapped PTS1/PTS2 (H maps image 2 → image 1)
- ORB finds few matches → needs textured images with real overlap

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMG1_PATH = "left.jpg"                 # reference image
IMG2_PATH = "right.jpg"                # image to warp into IMG1's frame
PTS1 = [[350, 50], [550, 50], [550, 350], [350, 350]]    # points in image 1 (x, y)
PTS2 = [[50, 50], [250, 50], [250, 350], [50, 350]]      # SAME points, same order, in image 2
USE_AUTO_POINTS = False                # True -> ORB features + RANSAC (ignores PTS1/PTS2)
BLEND_OVERLAP   = True
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

img1, img2 = cv2.imread(IMG1_PATH), cv2.imread(IMG2_PATH)
if img1 is None or img2 is None:
    print("[WARN] Could not read the images -> using a synthetic overlapping pair (default PTS match it)")
    scene = make_demo(400, 900)
    img1, img2 = scene[:, :600].copy(), scene[:, 300:].copy()
h1, w1 = img1.shape[:2]; h2, w2 = img2.shape[:2]

if USE_AUTO_POINTS:
    orb = cv2.ORB_create(3000)
    k1, d1 = orb.detectAndCompute(cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY), None)
    k2, d2 = orb.detectAndCompute(cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY), None)
    matches = sorted(cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True).match(d2, d1), key=lambda m: m.distance)[:200]
    pts2 = np.float32([k2[m.queryIdx].pt for m in matches]); pts1 = np.float32([k1[m.trainIdx].pt for m in matches])
    print("Auto matches used:", len(matches))
else:
    pts1, pts2 = np.float32(PTS1), np.float32(PTS2)

H, _ = cv2.findHomography(pts2, pts1, cv2.RANSAC if len(pts1) > 4 else 0, 5.0)   # image2 -> image1 frame
print("Homography H (img2 -> img1):\n", H.round(5))

# ---- canvas big enough for both images ----
c2 = cv2.perspectiveTransform(np.float32([[0, 0], [w2, 0], [w2, h2], [0, h2]]).reshape(-1, 1, 2), H).reshape(-1, 2)
c1 = np.float32([[0, 0], [w1, 0], [w1, h1], [0, h1]])
allc = np.vstack([c1, c2])
xmin, ymin = np.floor(allc.min(axis=0)).astype(int); xmax, ymax = np.ceil(allc.max(axis=0)).astype(int)
Tshift = np.array([[1, 0, -xmin], [0, 1, -ymin], [0, 0, 1]], float)       # move everything into positive coordinates
size = (int(xmax - xmin), int(ymax - ymin))                               # (width, height)

warped2 = cv2.warpPerspective(img2, Tshift @ H, size)
canvas1 = np.zeros_like(warped2)
ox, oy = -xmin, -ymin
canvas1[oy:oy + h1, ox:ox + w1] = img1
m1, m2 = canvas1.sum(axis=2) > 0, warped2.sum(axis=2) > 0

result = warped2.copy()
result[m1 & ~m2] = canvas1[m1 & ~m2]
both = m1 & m2
result[both] = (0.5 * canvas1[both] + 0.5 * warped2[both]).astype(np.uint8) if BLEND_OVERLAP else canvas1[both]

fig, ax = plt.subplots(1, 3, figsize=(18, 5))
show(ax[0], img1, "Image 1 (reference)"); show(ax[1], img2, "Image 2")
show(ax[2], result, f"Panorama {size[0]}x{size[1]}")
plt.tight_layout(); plt.show()
```

- Math note: S, R, t and H compose into ONE homography; the printed check shows it equals a direct 4-point homography

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
PAINTING_PATH = "painting.jpg"
SCENE_PATH    = "wall_with_frame.jpg"
SCALE         = 0.4                                  # stage 1: linear scale
ANGLE, TX, TY = 0, 200, 80                           # stage 2: rigid (rotation + translation)
FRAME_PTS = [[200, 80], [430, 60], [460, 300], [180, 330]]   # stage 3: frame corners in SCENE (TL, TR, BR, BL)
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

# ---------- 3x3 homogeneous building blocks (image coords: x right, y DOWN) ----------
def T(tx, ty):   return np.array([[1, 0, tx], [0, 1, ty], [0, 0, 1]], float)
def S(sx, sy):   return np.array([[sx, 0, 0], [0, sy, 0], [0, 0, 1]], float)
def R(deg):                                    # positive = counter-clockwise (same as cv2.getRotationMatrix2D)
    a = np.deg2rad(deg); c, s = np.cos(a), np.sin(a)
    return np.array([[c, s, 0], [-s, c, 0], [0, 0, 1]], float)
def SH(kx, ky):  return np.array([[1, kx, 0], [ky, 1, 0], [0, 0, 1]], float)

painting = load_img(PAINTING_PATH)
scene = cv2.imread(SCENE_PATH)
if scene is None:
    print("[WARN] Scene not found -> synthetic wall with a skewed frame (default FRAME_PTS match it)")
    scene = np.full((400, 600, 3), 190, np.uint8)
    cv2.polylines(scene, [np.int32(FRAME_PTS)], True, (30, 60, 120), 8)
sh, sw = scene.shape[:2]; ph, pw = painting.shape[:2]

# stage 1 (linear scale) + stage 2 (rigid: rotate then translate) as 3x3 matrices
M_stage12 = T(TX, TY) @ R(ANGLE) @ S(SCALE, SCALE)

# where do the painting's 4 corners land after stages 1-2?
corners = np.float32([[0, 0], [pw, 0], [pw, ph], [0, ph]])
c_after = cv2.perspectiveTransform(corners.reshape(-1, 1, 2), M_stage12).reshape(-1, 2).astype(np.float32)

# stage 3 (projective): map those corners onto the frame corners
frame = np.float32(FRAME_PTS)
H3 = cv2.getPerspectiveTransform(c_after, frame)
M_total = H3 @ M_stage12                                        # ONE matrix for all three stages

H_direct = cv2.getPerspectiveTransform(corners, frame)
print("Combined matrix equals direct 4-point homography:", np.allclose(M_total / M_total[2, 2], H_direct / H_direct[2, 2], atol=1e-3))

warped = cv2.warpPerspective(painting, M_total, (sw, sh))
mask = cv2.warpPerspective(np.full((ph, pw), 255, np.uint8), M_total, (sw, sh))
result = scene.copy()
result[mask > 127] = warped[mask > 127]                         # paste only where the painting lands

fig, ax = plt.subplots(1, 3, figsize=(18, 5))
show(ax[0], painting, "Flat painting"); show(ax[1], scene, "Scene + target frame"); ax[1].plot(*np.vstack([frame, frame[:1]]).T, "r-o")
show(ax[2], result, "Painting inserted")
plt.tight_layout(); plt.show()
```

## 3.10 Implement a Warp WITHOUT cv2.warpAffine (Inverse Mapping)

### Source Task / Concept
Concept behind every task in Lab 3: *how* a matrix transforms an image. Useful if the exam says 'do not use built-in warp functions'.

**Generalized problem:** For every output pixel compute where it comes from in the source using the INVERSE matrix, then sample (nearest-neighbour or bilinear).

### If the exam question says...
- implement the transformation manually / from scratch / without cv2.warpAffine
- inverse mapping
- interpolation (nearest vs bilinear)
- why do forward mapping holes appear?

### Values you may need to change
- `M` (any 2×3 affine matrix)
- `OUT_SIZE`
- `BILINEAR` True/False

### Possible exam variations
- Use the same function for rotation/scale/shear/translation; extend to perspective by dividing by the third coordinate

### If it does not work immediately
- Black holes in output → you used FORWARD mapping; always invert the matrix and loop over OUTPUT pixels
- Why inverse? Each output pixel then gets exactly one value, no gaps/overlaps

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
ANGLE, SCALE = 30, 0.8
OUT_SIZE = None               # (width, height); None = same as input
BILINEAR = True               # False = nearest neighbour
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

def warp_affine_manual(img, M, out_size, bilinear=True):
    """Inverse-mapping warp. M is the 2x3 forward matrix (source -> output)."""
    h, w = img.shape[:2]; out_w, out_h = out_size
    Minv = np.linalg.inv(np.vstack([M, [0, 0, 1]]))                    # output -> source
    xs, ys = np.meshgrid(np.arange(out_w), np.arange(out_h))
    coords = np.stack([xs.ravel(), ys.ravel(), np.ones(xs.size)])      # 3 x N homogeneous output pixels
    sx, sy = (Minv @ coords)[:2]                                       # source coordinates (float)
    out = np.zeros((out_h * out_w, 1 if img.ndim == 2 else img.shape[2]), np.float32)
    if bilinear:
        x0, y0 = np.floor(sx).astype(int), np.floor(sy).astype(int)
        valid = (x0 >= 0) & (x0 < w - 1) & (y0 >= 0) & (y0 < h - 1)
        x0, y0, dx, dy = x0[valid], y0[valid], (sx - np.floor(sx))[valid][:, None], (sy - np.floor(sy))[valid][:, None]
        f = img.reshape(h, w, -1).astype(np.float32)
        out[valid] = (f[y0, x0] * (1 - dx) * (1 - dy) + f[y0, x0 + 1] * dx * (1 - dy)
                      + f[y0 + 1, x0] * (1 - dx) * dy + f[y0 + 1, x0 + 1] * dx * dy)
    else:
        xi, yi = np.rint(sx).astype(int), np.rint(sy).astype(int)
        valid = (xi >= 0) & (xi < w) & (yi >= 0) & (yi < h)
        out[valid] = img.reshape(h, w, -1)[yi[valid], xi[valid]]
    return out.reshape((out_h, out_w) + (() if img.ndim == 2 else (img.shape[2],))).clip(0, 255).astype(np.uint8)

img = load_img(IMAGE_PATH)
h, w = img.shape[:2]
M = cv2.getRotationMatrix2D((w / 2, h / 2), ANGLE, SCALE)              # any 2x3 matrix works here
size = OUT_SIZE or (w, h)

mine = warp_affine_manual(img, M, size, BILINEAR)
ref = cv2.warpAffine(img, M, size, flags=cv2.INTER_LINEAR if BILINEAR else cv2.INTER_NEAREST)
print("Mean abs difference vs cv2.warpAffine:", np.abs(mine.astype(float) - ref.astype(float)).mean().round(3))

fig, ax = plt.subplots(1, 3, figsize=(16, 5))
show(ax[0], img, "Input"); show(ax[1], mine, "Manual inverse mapping"); show(ax[2], ref, "cv2.warpAffine")
plt.tight_layout(); plt.show()
```

## 4.1 Document Scanner (Perspective Correction + Adaptive Threshold)

**Values you may need to change:** `SRC_PTS`, `BLOCK_SIZE`, `C`, `PRE_BLUR`, `KEEP_COLOR` (skip thresholding if only straightening is needed).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
SRC_PTS = [[120, 60], [480, 90], [540, 340], [60, 320]]     # 4 corners of the page (x, y)
BLOCK_SIZE, C, PRE_BLUR = 21, 10, 5
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

def order_points(pts):
    """Return corners ordered: top-left, top-right, bottom-right, bottom-left."""
    pts = np.array(pts, dtype=np.float32)
    s = pts.sum(axis=1); d = np.diff(pts, axis=1).ravel()      # d = y - x
    return np.array([pts[np.argmin(s)], pts[np.argmin(d)], pts[np.argmax(s)], pts[np.argmax(d)]], np.float32)

def homography_from_points(src, dst):
    """Solve the 8 unknowns of a homography from 4 point pairs (manual DLT, 8x8 linear system)."""
    A, b = [], []
    for (x, y), (u, v) in zip(src, dst):
        A.append([x, y, 1, 0, 0, 0, -x * u, -y * u]); b.append(u)
        A.append([0, 0, 0, x, y, 1, -x * v, -y * v]); b.append(v)
    h = np.linalg.solve(np.array(A, float), np.array(b, float))
    return np.append(h, 1).reshape(3, 3)

img = load_img(IMAGE_PATH)
src = order_points(SRC_PTS); tl, tr, br, bl = src
W = int(max(np.linalg.norm(br - bl), np.linalg.norm(tr - tl)))
H = int(max(np.linalg.norm(tr - br), np.linalg.norm(tl - bl)))
dst = np.array([[0, 0], [W - 1, 0], [W - 1, H - 1], [0, H - 1]], np.float32)

flat = cv2.warpPerspective(img, cv2.getPerspectiveTransform(src, dst), (W, H))      # step 1: straighten
gray = cv2.cvtColor(flat, cv2.COLOR_BGR2GRAY)
if PRE_BLUR: gray = cv2.GaussianBlur(gray, (PRE_BLUR, PRE_BLUR), 0)
scan = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, BLOCK_SIZE, C)   # step 2

fig, ax = plt.subplots(1, 3, figsize=(17, 5))
show(ax[0], img, "Input"); show(ax[1], flat, "Straightened"); show(ax[2], scan, "Scanned (adaptive threshold)")
plt.tight_layout(); plt.show()
```

## 4.2 Enhance Only a Selected Region (Mask + Point Transform)

**If the exam question says...**

**Values you may need to change:** `REGION` and its coordinates, `ENHANCE` (`equalize`, `gamma`, `log`, `blur`), `GAMMA`, `INVERT_MASK`.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
REGION  = "circle"                  # "circle" | "rect"
CIRCLE  = ((300, 200), 100)         # ((cx, cy), radius)
RECT    = (150, 100, 450, 300)      # (x1, y1, x2, y2)
ENHANCE = "gamma"                   # "equalize" | "gamma" | "log" | "blur"
GAMMA   = 0.5
INVERT_MASK = False                 # True -> enhance everything EXCEPT the region
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

def show(ax, img, title="", fs=12, color="black"):
    """Display a BGR (or gray) OpenCV image on a matplotlib axis."""
    if img.ndim == 2: ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    else:             ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title, fontsize=fs, color=color); ax.axis("off")

img = load_img(IMAGE_PATH)
h, w = img.shape[:2]

if ENHANCE == "equalize":
    ycc = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb); y = ycc[:, :, 0].copy(); ycc[:, :, 0] = cv2.equalizeHist(y)
    enhanced = cv2.cvtColor(ycc, cv2.COLOR_YCrCb2BGR)
elif ENHANCE == "gamma":
    enhanced = cv2.LUT(img, (255.0 * (np.arange(256) / 255.0) ** GAMMA).clip(0, 255).astype(np.uint8))
elif ENHANCE == "log":
    enhanced = (255.0 / np.log1p(255.0) * np.log1p(img.astype(np.float32))).clip(0, 255).astype(np.uint8)
else:
    enhanced = cv2.GaussianBlur(img, (31, 31), 0)

mask = np.zeros((h, w), np.uint8)
if REGION == "circle": cv2.circle(mask, CIRCLE[0], CIRCLE[1], 255, -1)
else:                  cv2.rectangle(mask, RECT[:2], RECT[2:], 255, -1)
if INVERT_MASK: mask = cv2.bitwise_not(mask)

result = cv2.bitwise_or(cv2.bitwise_and(enhanced, enhanced, mask=mask),
                        cv2.bitwise_and(img, img, mask=cv2.bitwise_not(mask)))

fig, ax = plt.subplots(1, 3, figsize=(16, 5))
show(ax[0], img, "Original"); show(ax[1], mask, "Mask"); show(ax[2], result, f"{ENHANCE} applied only in region")
plt.tight_layout(); plt.show()
```

## 4.4 Finding Pixel Coordinates for Source Points (needed by 3.6, 3.7, 3.8, 3.9)

**How to recognize it:** the question says "select/pick 4 points" or "corners of the …" but gives no numbers.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==================== USER SETTINGS — CHANGE THESE ====================
IMAGE_PATH = "exam_image.png"
GRID_STEP  = 50
CROP       = None                # e.g. (0, 0, 300, 300) to zoom into a region (ticks stay in ORIGINAL coordinates)
USE_CV2_CLICK = False            # True (desktop python): click 4 points in a window; 'q' quits
# ======================================================================

# ---------- HELPERS (demo fallback so the cell runs even without your file; harmless in the exam) ----------
def make_demo(h=400, w=600):
    x = np.tile(np.linspace(30, 220, w).astype(np.uint8), (h, 1))
    img = cv2.merge([x, x[:, ::-1].copy(), np.full((h, w), 90, np.uint8)])
    cv2.circle(img, (w // 2, h // 2), 90, (0, 200, 255), -1)
    cv2.rectangle(img, (40, 40), (200, 140), (255, 255, 255), 4)
    cv2.rectangle(img, (w - 200, 40), (w - 40, 140), (0, 200, 0), -1)
    cv2.putText(img, "DEMO", (w // 2 - 110, h - 40), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255, 255, 255), 5)
    return img

def load_img(path, gray=False, variant=0):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE if gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"[WARN] Could not read '{path}' -> using synthetic demo image. Fix the path!")
        img = make_demo()
        if variant: img = 255 - img
        if gray: img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return img

img = load_img(IMAGE_PATH)
h, w = img.shape[:2]
x1, y1, x2, y2 = CROP or (0, 0, w, h)

# ---- Method A: grid + ticks ----
plt.figure(figsize=(12, 8))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB), extent=(0, w, h, 0))
plt.xlim(x1, x2); plt.ylim(y2, y1)
plt.xticks(np.arange(x1, x2 + 1, GRID_STEP)); plt.yticks(np.arange(y1, y2 + 1, GRID_STEP))
plt.grid(color="yellow", alpha=0.5); plt.xlabel("x (columns)"); plt.ylabel("y (rows)")
plt.title("Read (x, y) of the points you need"); plt.show()

# ---- Method B: click in an OpenCV window (desktop only) ----
if USE_CV2_CLICK:
    pts, disp = [], img.copy()
    def on_click(event, x, y, flags, param):
        if event == cv2.EVENT_LBUTTONDOWN:
            pts.append([x, y]); cv2.circle(disp, (x, y), 5, (0, 0, 255), -1); print("point", [x, y])
    cv2.namedWindow("click points"); cv2.setMouseCallback("click points", on_click)
    while True:
        cv2.imshow("click points", disp)
        if (cv2.waitKey(20) & 0xFF) == ord("q") or len(pts) >= 4: break
    cv2.destroyAllWindows(); print("SRC_PTS =", pts)
```

## 5.4 Concept / MCQ notes
- **Point transformation:** `s = T(r)`, depends only on the pixel's own value. **Neighbourhood (spatial) filtering** (blur) uses surrounding pixels.
- **Negative:** `s = 255 − r`. **Log:** `s = c·log(1+r)` — expands dark values, compresses bright ones (use for images with huge dynamic range / dark detail). **Gamma:** `s = c·r^γ` — γ<1 expands darks (brightens), γ>1 compresses darks (darkens); γ=1 is identity.
- **Histogram equalization:** uses the CDF as the mapping, spreads intensities to flatten the histogram → higher global contrast; can amplify noise. **CLAHE:** equalizes in tiles with clip limit → local contrast without over-amplification. Equalizing a color image channel-wise changes hues; equalize luminance only.
- **Contrast stretching** is linear (`(r−min)/(max−min)·255`) and keeps histogram *shape*; equalization changes the shape.
- **Global threshold:** one T for the whole image; fails under uneven illumination. **Adaptive:** T computed per neighbourhood (mean or Gaussian-weighted mean − C). **Otsu:** picks T automatically by minimizing within-class variance (works for bimodal histograms).
- **Bitwise ops** need identical size/type; mask must be `uint8` 0/255.
- **BGR vs RGB:** OpenCV = BGR, Matplotlib = RGB. **HSV hue range in OpenCV = 0–179.**
- **Why 3×3?** A 2×2 matrix cannot represent translation (origin always fixed). Homogeneous coordinates `[x, y, 1]` make translation a matrix multiplication.
- **Matrix order:** `M = M3·M2·M1` applies M1 first; transformations don't commute (except e.g. uniform scale & rotation about the same point).
- **DoF:** translation 2, rigid 3, similarity 4, affine 6, projective 8. Points needed: affine 3, projective 4 (each point gives 2 equations).
- **Preserves:** rigid → lengths & angles; similarity → angles; affine → parallelism; projective → straight lines (not parallelism). Perspective foreshortening needs projective.
- **Inverse mapping** avoids holes: for each output pixel find its source location; interpolation: nearest (fast, blocky), bilinear (smooth), bicubic (smoother, slower).
- `warpAffine(src, M(2×3), dsize=(w, h))` · `warpPerspective(src, H(3×3), dsize=(w, h))` · `getRotationMatrix2D(center, angle, scale)` (positive angle = CCW).
- **Homography vs affine:** affine = 6 params, parallel lines stay parallel; homography = 8 params, used for perspective/stitching planar scenes.

## 5.5 Frequent mistakes (check these first when output looks wrong)
1. `(width, height)` for `dsize`/`resize`, but `shape = (height, width)`.
2. Points are `(x, y)`; NumPy indexing is `[y, x]` / `[row, col]`.
3. `uint8` overflow: `img + 50` wraps around. Use `cv2.add`, or convert to float and `np.clip(..., 0, 255).astype(np.uint8)`.
4. `cv2.imread` returns `None` silently on a bad path.
5. Matplotlib shows grayscale in color unless `cmap="gray"`.
6. Even kernel size for blur/threshold block size.
7. Forgetting to `release()` the video capture / `destroyAllWindows()`.
8. In notebooks use `plt.imshow`, not `cv2.imshow` (it can freeze the kernel).
