### 2.2 Create Practice Images

**What it does:** Generates synthetic test images (image.jpg, image2.jpg, shapes.jpg, left.jpg, right.jpg, tilted.jpg, document.jpg, road.jpg) so every template can be tried without real files.

**When to use:** For practice only. Skip in the exam.

**What to change:** Nothing.

```python
import cv2
import numpy as np

rng = np.random.default_rng(0)
H, W = 480, 640
FONT = cv2.FONT_HERSHEY_SIMPLEX

# --- textured image (many SIFT keypoints) ---
base = np.full((H, W, 3), 235, np.uint8)
for _ in range(250):
    c = tuple(int(v) for v in rng.integers(0, 255, 3))
    p = (int(rng.integers(0, W)), int(rng.integers(0, H)))
    if rng.random() < 0.5:
        cv2.circle(base, p, int(rng.integers(3, 14)), c, -1)
    else:
        q = (p[0] + int(rng.integers(5, 30)), p[1] + int(rng.integers(5, 30)))
        cv2.rectangle(base, p, q, c, -1)
cv2.putText(base, "LAB 6", (230, 250), FONT, 2.0, (0, 0, 0), 4)
cv2.imwrite("image.jpg", base)

# --- rotated + scaled + brighter copy (for matching) ---
M = cv2.getRotationMatrix2D((W / 2, H / 2), 25, 0.8)
img2 = cv2.warpAffine(base, M, (W, H))
img2 = cv2.convertScaleAbs(img2, alpha=1.0, beta=25)
cv2.imwrite("image2.jpg", img2)

# --- left / right overlapping halves (for panorama) ---
cv2.imwrite("left.jpg", base[:, :420])
cv2.imwrite("right.jpg", base[:, 220:])

# --- clean shapes (edges, contours, Hough) ---
sh = np.full((H, W, 3), 255, np.uint8)
cv2.rectangle(sh, (60, 60), (260, 200), (200, 80, 40), -1)
cv2.circle(sh, (450, 130), 60, (30, 30, 200), -1)
cv2.circle(sh, (150, 330), 50, (40, 160, 60), -1)
cv2.circle(sh, (330, 360), 35, (0, 150, 200), -1)
cv2.fillPoly(sh, [np.array([[500, 320], [600, 320], [550, 230]])], (120, 0, 120))
cv2.line(sh, (20, 450), (620, 420), (0, 0, 0), 3)
cv2.line(sh, (300, 20), (380, 200), (0, 0, 0), 3)
cv2.imwrite("shapes.jpg", sh)

# --- tilted text page (deskew) ---
page = np.full((H, W, 3), 255, np.uint8)
for y in range(60, 440, 40):
    cv2.putText(page, "The quick brown fox jumps over the lazy dog", (30, y), FONT, 0.8, (0, 0, 0), 2)
    cv2.line(page, (20, y + 10), (620, y + 10), (0, 0, 0), 1)
Mt = cv2.getRotationMatrix2D((W / 2, H / 2), -7, 1.0)
cv2.imwrite("tilted.jpg", cv2.warpAffine(page, Mt, (W, H), borderValue=(255, 255, 255)))

# --- document photographed at an angle ---
doc = np.full((400, 300, 3), 255, np.uint8)
for y in range(40, 380, 30):
    cv2.putText(doc, "Sample text line", (20, y), FONT, 0.6, (0, 0, 0), 1)
src = np.float32([[0, 0], [299, 0], [299, 399], [0, 399]])
dst = np.float32([[120, 60], [480, 90], [520, 420], [80, 380]])
Mp = cv2.getPerspectiveTransform(src, dst)
bg = np.full((H, W, 3), (60, 90, 120), np.uint8)
warped = cv2.warpPerspective(doc, Mp, (W, H))
m = cv2.warpPerspective(np.full((400, 300), 255, np.uint8), Mp, (W, H))
bg[m > 0] = warped[m > 0]
cv2.imwrite("document.jpg", bg)

# --- road with lane lines ---
road = np.full((H, W, 3), (110, 160, 110), np.uint8)
cv2.fillPoly(road, [np.array([[0, 480], [280, 290], [360, 290], [640, 480]])], (70, 70, 70))
cv2.line(road, (120, 480), (300, 290), (255, 255, 255), 8)
cv2.line(road, (520, 480), (340, 290), (255, 255, 255), 8)
cv2.imwrite("road.jpg", road)

print("Created: image.jpg image2.jpg left.jpg right.jpg shapes.jpg tilted.jpg document.jpg road.jpg")
```

### 2.5 Video or Webcam Frame Processing

**What it does:** Applies an image operation (Canny here) to every frame of a video and writes a new video.

**When to use:** When the question says 'apply to a video' or 'real-time'.

**What to change:** `VIDEO_PATH` (0 = webcam), `OUTPUT_PATH`, and the PROCESS block.

**Exam questions it solves:** Apply edge detection / Hough / SIFT to video.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
VIDEO_PATH = "video.mp4"     # CHANGE THIS: video file, or 0 for webcam
OUTPUT_PATH = "output.avi"    # CHANGE THIS
MAX_FRAMES = 300

cap = cv2.VideoCapture(VIDEO_PATH)
assert cap.isOpened(), "Cannot open video"
fps = cap.get(cv2.CAP_PROP_FPS) or 25
W_, H_ = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)), int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
out = cv2.VideoWriter(OUTPUT_PATH, cv2.VideoWriter_fourcc(*"XVID"), fps, (W_, H_))

first, n = None, 0
while n < MAX_FRAMES:
    ok, frame = cap.read()
    if not ok:
        break
    # ---------------- PROCESS (CHANGE THIS) ----------------
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 1.4), 50, 150)
    result = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)      # writer needs 3 channels
    # --------------------------------------------------------
    out.write(result)
    if first is None:
        first = result
    n += 1
cap.release(); out.release()
print("Frames processed:", n, "->", OUTPUT_PATH)
plt.imshow(first); plt.title("First processed frame"); plt.axis("off"); plt.show()
```

## 3. Wavelet Transform

Needs `pywt` (PyWavelets). Convention: `cA` = approximation (LL, low frequency), `cH` = horizontal detail, `cV` = vertical detail, `cD` = diagonal detail. `pywt.dwt2` returns `(cA, (cH, cV, cD))`; `pywt.wavedec2` returns `[cA_n, (details level n), ..., (details level 1)]` (coarsest first, finest last).

### 3.1 Single-Level 2D DWT

**What it does:** Splits an image into 4 subbands: approximation and horizontal, vertical, diagonal detail. Each subband is half the size.

**When to use:** 'Apply DWT to the image and show the subbands.'

**What to change:** `IMAGE_PATH`, `WAVELET` (haar, db2, db4, sym4, coif1, bior1.3).

**Exam questions it solves:** Apply DWT; show LL LH HL HH; what does each subband contain.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pywt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
WAVELET = "haar"      # CHANGE THIS: haar, db2, db4, sym4, coif1, bior1.3

# ---- load grayscale image ----
gray = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
assert gray is not None, f"Cannot read {IMAGE_PATH}"
h, w = gray.shape
g = gray.astype(np.float32)
norm8 = lambda x: cv2.normalize(np.abs(x), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
to8 = lambda x: np.clip(x, 0, 255).astype(np.uint8)

cA, (cH, cV, cD) = pywt.dwt2(g, WAVELET)          # approx, horizontal, vertical, diagonal
print("Original:", gray.shape, "| each subband:", cA.shape)

items = [(norm8(cA), "Approximation cA (LL): low freq, blurry copy"),
         (norm8(cH), "Horizontal detail cH (LH): horizontal edges"),
         (norm8(cV), "Vertical detail cV (HL): vertical edges"),
         (norm8(cD), "Diagonal detail cD (HH): diagonal / corners")]

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 3.4 Wavelet Image Denoising

**What it does:** Adds Gaussian noise (optional), decomposes, soft-thresholds the detail coefficients (small = noise), reconstructs, and reports PSNR.

**When to use:** 'Denoise using wavelet transform', 'remove noise by thresholding coefficients'.

**What to change:** `IMAGE_PATH`, `ADD_NOISE` / `NOISE_SIGMA` (set ADD_NOISE=False if image is already noisy), `WAVELET`, `LEVEL`, `THRESHOLD` (None = automatic), `THRESH_MODE`.

**Exam questions it solves:** Wavelet denoising pipeline: transform, threshold, inverse.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pywt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
ADD_NOISE = True      # CHANGE THIS: False if the image is already noisy
NOISE_SIGMA = 25      # CHANGE THIS: noise std-dev
WAVELET = "db4"
LEVEL = 3
THRESHOLD = None      # CHANGE THIS: None = automatic (universal threshold), or a number e.g. 30
THRESH_MODE = "soft"  # "soft" or "hard"

# ---- load grayscale image ----
gray = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
assert gray is not None, f"Cannot read {IMAGE_PATH}"
h, w = gray.shape
g = gray.astype(np.float32)
norm8 = lambda x: cv2.normalize(np.abs(x), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
to8 = lambda x: np.clip(x, 0, 255).astype(np.uint8)

rng = np.random.default_rng(0)
noisy = g + rng.normal(0, NOISE_SIGMA, g.shape) if ADD_NOISE else g.copy()

coeffs = pywt.wavedec2(noisy, WAVELET, level=LEVEL)
sigma_est = np.median(np.abs(coeffs[-1][-1])) / 0.6745               # noise estimate from finest diagonal
thr = THRESHOLD if THRESHOLD is not None else sigma_est * np.sqrt(2 * np.log(noisy.size))
den_coeffs = [coeffs[0]] + [tuple(pywt.threshold(c, thr, mode=THRESH_MODE) for c in lvl) for lvl in coeffs[1:]]
den = to8(pywt.waverec2(den_coeffs, WAVELET)[:h, :w])
noisy8 = to8(noisy)

print(f"threshold used: {thr:.2f}")
print(f"PSNR noisy   vs original: {cv2.PSNR(gray, noisy8):.2f} dB")
print(f"PSNR denoised vs original: {cv2.PSNR(gray, den):.2f} dB")
items = [(gray, "Original"), (noisy8, "Noisy"), (den, "Wavelet denoised")]

# ---- display ----
plt.figure(figsize=(15, 5))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 3.5 Wavelet Image Compression

**What it does:** Keeps only the largest X% of wavelet coefficients (sets the rest to 0), reconstructs, and reports PSNR and compression ratio.

**When to use:** 'Compress the image using DWT', 'keep only 10% of coefficients', 'compare quality'.

**What to change:** `IMAGE_PATH`, `WAVELET`, `LEVEL`, `KEEP_LIST` (percentages to keep).

**Exam questions it solves:** Image compression with wavelets; compression vs quality.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pywt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
WAVELET = "db2"
LEVEL = 3
KEEP_LIST = [100, 20, 5, 1]     # CHANGE THIS: % of coefficients to keep

# ---- load grayscale image ----
gray = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
assert gray is not None, f"Cannot read {IMAGE_PATH}"
h, w = gray.shape
g = gray.astype(np.float32)
norm8 = lambda x: cv2.normalize(np.abs(x), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
to8 = lambda x: np.clip(x, 0, 255).astype(np.uint8)

coeffs = pywt.wavedec2(g, WAVELET, level=LEVEL)
arr, slices = pywt.coeffs_to_array(coeffs)

items = []
for keep in KEEP_LIST:
    thr = np.percentile(np.abs(arr), 100 - keep) if keep < 100 else 0
    arr_c = np.where(np.abs(arr) >= thr, arr, 0)                       # zero small coefficients
    c2 = pywt.array_to_coeffs(arr_c, slices, output_format="wavedec2")
    rec = to8(pywt.waverec2(c2, WAVELET)[:h, :w])
    nz = np.count_nonzero(arr_c)
    items.append((rec, f"keep {keep}% | PSNR {cv2.PSNR(gray, rec):.1f} dB | ratio {arr.size / nz:.1f}:1"))
    print(f"keep {keep:>3}% -> nonzero coeffs {nz}, compression ratio {arr.size / nz:.1f}:1")

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 3.8 Audio Signal Denoising with DWT

**What it does:** 1D DWT denoising of an audio/signal array and plots Original, Noisy, Cleaned (like the figure in the manual). Works with a .wav file or a synthetic signal.

**When to use:** 'Denoise the audio / 1D signal using wavelets'.

**What to change:** `AUDIO_PATH` (None = synthetic), `ADD_NOISE`, `NOISE_STD`, `WAVELET`, `LEVEL`, `SAVE`.

**Exam questions it solves:** 1D wavelet denoising.

```python
import numpy as np
import matplotlib.pyplot as plt
import pywt
from scipy.io import wavfile

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
AUDIO_PATH = None          # CHANGE THIS: "audio.wav", or None to use a synthetic signal
ADD_NOISE = True           # CHANGE THIS: False if the audio is already noisy
NOISE_STD = 0.1
WAVELET = "db4"
LEVEL = 5
SAVE = False               # CHANGE THIS: True to write cleaned.wav

if AUDIO_PATH:
    sr, data = wavfile.read(AUDIO_PATH)
    data = data.astype(np.float32)
    if data.ndim > 1:
        data = data.mean(axis=1)               # stereo -> mono
    data = data / (np.abs(data).max() + 1e-9)
else:
    sr = 8000
    t = np.arange(0, 2, 1 / sr)
    data = 0.6 * np.sin(2 * np.pi * 220 * t) * (np.sin(2 * np.pi * 1.5 * t) > 0)

rng = np.random.default_rng(0)
noisy = data + rng.normal(0, NOISE_STD, data.shape) if ADD_NOISE else data.copy()

coeffs = pywt.wavedec(noisy, WAVELET, level=LEVEL)
sigma = np.median(np.abs(coeffs[-1])) / 0.6745
thr = sigma * np.sqrt(2 * np.log(len(noisy)))
coeffs[1:] = [pywt.threshold(c, thr, mode="soft") for c in coeffs[1:]]     # keep coeffs[0] (approximation)
clean = pywt.waverec(coeffs, WAVELET)[:len(noisy)]

snr = lambda ref, x: 10 * np.log10(np.sum(ref ** 2) / np.sum((ref - x) ** 2))
print(f"SNR noisy: {snr(data, noisy):.2f} dB | SNR cleaned: {snr(data, clean):.2f} dB | threshold {thr:.3f}")

plt.figure(figsize=(12, 7))
for i, (sig, name) in enumerate([(data, "Original Audio"), (noisy, "Noisy Audio"), (clean, "Cleaned Audio")], 1):
    plt.subplot(3, 1, i); plt.plot(sig); plt.title(name)
plt.tight_layout(); plt.show()

if SAVE:
    wavfile.write("cleaned.wav", sr, (np.clip(clean, -1, 1) * 32767).astype(np.int16))
```

### 3.10 Wavelet Edge Detection

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pywt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
WAVELET = "haar"

# ---- load grayscale image ----
gray = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
assert gray is not None, f"Cannot read {IMAGE_PATH}"
h, w = gray.shape
g = gray.astype(np.float32)
norm8 = lambda x: cv2.normalize(np.abs(x), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
to8 = lambda x: np.clip(x, 0, 255).astype(np.uint8)

cA, (cH, cV, cD) = pywt.dwt2(g, WAVELET)
edge = np.sqrt(cH ** 2 + cV ** 2 + cD ** 2)                 # high-frequency energy = edges
edge8 = cv2.resize(norm8(edge), (w, h))                     # back to original size
_, binary = cv2.threshold(edge8, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
canny = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 1.4), 50, 150)

items = [(gray, "Original"), (edge8, "Wavelet edge strength"), (binary, "Wavelet edges (Otsu)"), (canny, "Canny (for comparison)")]

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 3.11 Wavelet Feature Extraction

**What it does:** Builds a feature vector from wavelet subband energies (DWT and WPT). Used for texture classification.

**When to use:** 'Extract features using wavelets', 'texture features', 'feature vector for a classifier'.

**What to change:** `IMAGE_PATH`, `WAVELET`, `LEVEL`.

**Exam questions it solves:** Feature extraction with DWT / WPT.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pywt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
WAVELET = "db2"
LEVEL = 2

# ---- load grayscale image ----
gray = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
assert gray is not None, f"Cannot read {IMAGE_PATH}"
h, w = gray.shape
g = gray.astype(np.float32)
norm8 = lambda x: cv2.normalize(np.abs(x), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
to8 = lambda x: np.clip(x, 0, 255).astype(np.uint8)

# ---- DWT energy features ----
coeffs = pywt.wavedec2(g, WAVELET, level=LEVEL)
names, feats = ["cA"], [np.mean(coeffs[0] ** 2)]
for i, (cH, cV, cD) in enumerate(coeffs[1:], start=1):
    lvl = LEVEL - i + 1
    for nm, c in zip(("cH", "cV", "cD"), (cH, cV, cD)):
        names.append(f"{nm}_L{lvl}"); feats.append(np.mean(c ** 2))
feats = np.array(feats)
print("DWT feature vector length:", len(feats))
for n_, f_ in zip(names, feats):
    print(f"{n_:8s} energy = {f_:.2f}")

# ---- WPT energy features ----
wp = pywt.WaveletPacket2D(data=g, wavelet=WAVELET, mode="symmetric", maxlevel=LEVEL)
nodes = wp.get_level(LEVEL)
wpt_feats = np.array([np.mean(n.data ** 2) for n in nodes])
print("WPT feature vector length:", len(wpt_feats))

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1); plt.bar(range(len(feats)), feats); plt.xticks(range(len(feats)), names, rotation=90); plt.title("DWT subband energy")
plt.subplot(1, 2, 2); plt.bar(range(len(wpt_feats)), wpt_feats); plt.xticks(range(len(nodes)), [n.path for n in nodes], rotation=90); plt.title("WPT subband energy")
plt.tight_layout(); plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pywt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
PATH_A = "image.jpg"        # CHANGE THIS
PATH_B = "image2.jpg"       # CHANGE THIS
WAVELET = "db2"

a = cv2.imread(PATH_A, cv2.IMREAD_GRAYSCALE); b = cv2.imread(PATH_B, cv2.IMREAD_GRAYSCALE)
assert a is not None and b is not None, "cannot read images"
b = cv2.resize(b, (a.shape[1], a.shape[0]))                  # same size required
cA1, (cH1, cV1, cD1) = pywt.dwt2(a.astype(np.float32), WAVELET)
cA2, (cH2, cV2, cD2) = pywt.dwt2(b.astype(np.float32), WAVELET)
pick = lambda x, y: np.where(np.abs(x) >= np.abs(y), x, y)   # max-abs rule for details
fused = pywt.idwt2(((cA1 + cA2) / 2, (pick(cH1, cH2), pick(cV1, cV2), pick(cD1, cD2))), WAVELET)
fused = np.clip(fused, 0, 255).astype(np.uint8)[:a.shape[0], :a.shape[1]]
items = [(a, "Image A"), (b, "Image B"), (fused, "Fused")]

# ---- display ----
plt.figure(figsize=(15, 5))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 4.2 Sobel Variations

**What it does:** Other ways to compute Sobel: CV_16S + convertScaleAbs + addWeighted (classic tutorial), cv2.magnitude, blur first, and gradient direction.

**When to use:** When the question asks for edge direction/angle, or specifically for convertScaleAbs/addWeighted.

**What to change:** `IMAGE_PATH`, `KSIZE`, `DIR_THRESH`.

**Exam questions it solves:** Gradient direction; absolute gradients; combine |Gx| and |Gy|.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
KSIZE = 3
DIR_THRESH = 100     # only show direction where magnitude > this

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

gx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=KSIZE)
gy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=KSIZE)
mag = cv2.magnitude(gx, gy)                              # same as sqrt(gx^2 + gy^2)
direction = np.degrees(np.arctan2(gy, gx))               # -180..180 degrees, perpendicular to the edge

# Classic variant: 16-bit signed -> abs -> blend
abs_x = cv2.convertScaleAbs(cv2.Sobel(gray, cv2.CV_16S, 1, 0, ksize=KSIZE))
abs_y = cv2.convertScaleAbs(cv2.Sobel(gray, cv2.CV_16S, 0, 1, ksize=KSIZE))
blend = cv2.addWeighted(abs_x, 0.5, abs_y, 0.5, 0)

# Blur first -> less noise
gray_b = cv2.GaussianBlur(gray, (5, 5), 1.4)
mag_b = cv2.magnitude(cv2.Sobel(gray_b, cv2.CV_64F, 1, 0, ksize=KSIZE), cv2.Sobel(gray_b, cv2.CV_64F, 0, 1, ksize=KSIZE))

items = [(abs_x, "|Gx| (CV_16S + convertScaleAbs)"), (abs_y, "|Gy|"), (blend, "addWeighted 0.5/0.5"),
         (np.uint8(np.clip(mag, 0, 255)), "magnitude (no blur)"), (np.uint8(np.clip(mag_b, 0, 255)), "magnitude (blur first)")]

# ---- display ----
plt.figure(figsize=(15, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()

dir_vis = np.where(mag > DIR_THRESH, direction, np.nan)
plt.figure(figsize=(6, 5)); plt.imshow(dir_vis, cmap="hsv"); plt.colorbar(label="degrees"); plt.title("Gradient direction (strong edges)"); plt.axis("off"); plt.show()
```

### 4.4 Sobel Kernel Calculation by Hand

**What it does:** Reproduces the manual's worked example (vertical edge gives 760, uniform area gives 0) and shows that filter2D with the Sobel kernel equals cv2.Sobel.

**When to use:** 'Calculate the Sobel response of this 3x3 patch', 'apply a custom kernel'.

**What to change:** The `patch` array and kernels.

**Exam questions it solves:** Numerical Sobel question; custom kernel with filter2D.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

Kx = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]], np.float64)     # Gx: responds to vertical edges
Ky = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]], np.float64)     # Gy: responds to horizontal edges

patch = np.array([[10, 10, 200], [10, 10, 200], [10, 10, 200]], np.float64)   # CHANGE THIS: 3x3 patch
uniform = np.full((3, 3), 50.0)

print("Gx on vertical-edge patch:", np.sum(patch * Kx), "(manual: 800 - 40 = 760)")
print("Gy on vertical-edge patch:", np.sum(patch * Ky))
print("Gx on uniform patch      :", np.sum(uniform * Kx), "(zero response)")
print("Magnitude:", np.hypot(np.sum(patch * Kx), np.sum(patch * Ky)))

# Whole-image convolution with a custom kernel
custom_x = cv2.filter2D(gray.astype(np.float64), cv2.CV_64F, Kx)
builtin_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
print("filter2D == Sobel ?", np.allclose(custom_x, builtin_x))

items = [(gray, "Original"), (np.abs(custom_x), "|filter2D(Kx)|")]

# ---- display ----
plt.figure(figsize=(10, 4))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 4.5 Prewitt Roberts Scharr

**What it does:** Other first-derivative edge operators via cv2.filter2D (Prewitt, Roberts) and cv2.Scharr.

**When to use:** 'Apply Prewitt / Roberts / Scharr operator'.

**What to change:** `IMAGE_PATH`.

**Exam questions it solves:** Compare gradient operators.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

g = gray.astype(np.float32)
mag_of = lambda a, b: cv2.normalize(np.hypot(a, b), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

kx_p = np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]], np.float32); ky_p = kx_p.T          # Prewitt
rx = np.array([[1, 0], [0, -1]], np.float32); ry = np.array([[0, 1], [-1, 0]], np.float32)  # Roberts (2x2)

prewitt = mag_of(cv2.filter2D(g, cv2.CV_32F, kx_p), cv2.filter2D(g, cv2.CV_32F, ky_p))
roberts = mag_of(cv2.filter2D(g, cv2.CV_32F, rx), cv2.filter2D(g, cv2.CV_32F, ry))
scharr = mag_of(cv2.Scharr(gray, cv2.CV_32F, 1, 0), cv2.Scharr(gray, cv2.CV_32F, 0, 1))
sobel = mag_of(cv2.Sobel(gray, cv2.CV_32F, 1, 0), cv2.Sobel(gray, cv2.CV_32F, 0, 1))

items = [(sobel, "Sobel"), (prewitt, "Prewitt"), (roberts, "Roberts"), (scharr, "Scharr")]

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 4.10 LoG Zero Crossings

**What it does:** True LoG edge detection: marks pixels where the Laplacian changes sign (strong sign changes only).

**When to use:** 'Edges are at zero crossings of LoG', 'detect zero crossings'.

**What to change:** `IMAGE_PATH`, `SIGMA`, `ZC_FRAC` (higher = fewer edges).

**Exam questions it solves:** Zero-crossing edge map.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
SIGMA = 1.4
ZC_FRAC = 0.05       # CHANGE THIS: fraction of max |Laplacian| a jump must exceed

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

def zero_crossings(L, frac=0.05):
    # marks pixels where the Laplacian changes sign AND the jump is strong enough
    thr = frac * np.abs(L).max()
    zc = np.zeros(L.shape, np.uint8)
    s = np.sign(L)
    right = (s[:, :-1] * s[:, 1:] < 0) & (np.abs(L[:, :-1] - L[:, 1:]) > thr)
    down = (s[:-1, :] * s[1:, :] < 0) & (np.abs(L[:-1, :] - L[1:, :]) > thr)
    zc[:, :-1][right] = 255
    zc[:-1, :][down] = 255
    return zc

blurred = cv2.GaussianBlur(gray, (0, 0), SIGMA)
lap = cv2.Laplacian(blurred, cv2.CV_64F, ksize=3)
zc = zero_crossings(lap, ZC_FRAC)

items = [(gray, "Gray"), (cv2.convertScaleAbs(lap), "|Laplacian| of Gaussian"), (zc, "Zero crossings (edges)")]

# ---- display ----
plt.figure(figsize=(15, 5))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

**What to change:** `SIGMA`.

**Exam questions it solves:** LoG kernel from formula; DoG approximates LoG.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
SIGMA = 2.0

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

def log_kernel(size, sigma):
    ax = np.arange(-(size // 2), size // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    r2 = xx ** 2 + yy ** 2
    k = (r2 - 2 * sigma ** 2) / (sigma ** 4) * np.exp(-r2 / (2 * sigma ** 2))
    return k - k.mean()                                   # zero-sum kernel

SIZE = int(6 * SIGMA) | 1                                 # odd size
k = log_kernel(SIZE, SIGMA)
g = gray.astype(np.float64)
log_resp = cv2.filter2D(g, cv2.CV_64F, k)
dog = cv2.GaussianBlur(g, (0, 0), SIGMA) - cv2.GaussianBlur(g, (0, 0), 1.6 * SIGMA)   # DoG ~ LoG

items = [(gray, "Original"), (k, f"LoG kernel {SIZE}x{SIZE}"), (cv2.convertScaleAbs(log_resp), "|LoG response|"), (cv2.convertScaleAbs(dog * 4), "|DoG| (approx. LoG)")]

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 4.12 Compare Sobel Canny LoG

**What it does:** Runs all three detectors on the same image side by side with edge pixel counts.

**When to use:** 'Compare Sobel, Canny and LoG', 'which detector is best'.

**What to change:** `IMAGE_PATH`, thresholds.

**Exam questions it solves:** Detector comparison.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
CANNY_LOW, CANNY_HIGH = 50, 150
SOBEL_THRESH = 80
ZC_FRAC = 0.05

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

def zero_crossings(L, frac=0.05):
    # marks pixels where the Laplacian changes sign AND the jump is strong enough
    thr = frac * np.abs(L).max()
    zc = np.zeros(L.shape, np.uint8)
    s = np.sign(L)
    right = (s[:, :-1] * s[:, 1:] < 0) & (np.abs(L[:, :-1] - L[:, 1:]) > thr)
    down = (s[:-1, :] * s[1:, :] < 0) & (np.abs(L[:-1, :] - L[1:, :]) > thr)
    zc[:, :-1][right] = 255
    zc[:-1, :][down] = 255
    return zc

blur = cv2.GaussianBlur(gray, (5, 5), 1.4)

gx = cv2.Sobel(blur, cv2.CV_64F, 1, 0); gy = cv2.Sobel(blur, cv2.CV_64F, 0, 1)
sobel = np.uint8(np.clip(np.hypot(gx, gy), 0, 255))
canny = cv2.Canny(blur, CANNY_LOW, CANNY_HIGH)
log_edges = zero_crossings(cv2.Laplacian(blur, cv2.CV_64F), ZC_FRAC)

items = [(rgb, "Original"),
         (sobel, f"Sobel (thick, {np.count_nonzero(sobel > SOBEL_THRESH)} px > {SOBEL_THRESH})"),
         (canny, f"Canny (thin, connected, {np.count_nonzero(canny)} px)"),
         (log_edges, f"LoG zero-crossing ({np.count_nonzero(log_edges)} px)")]

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 4.13.5 Contours and Object Boundaries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
METHOD = "otsu"       # CHANGE THIS: "canny" or "otsu"
MIN_AREA = 500        # ignore tiny contours
OUT_PATH = None       # e.g. "largest.png"

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

blur = cv2.GaussianBlur(gray, (5, 5), 1.4)
if METHOD == "canny":
    binary = cv2.morphologyEx(cv2.Canny(blur, 50, 150), cv2.MORPH_CLOSE, np.ones((5, 5), np.uint8))
else:
    _, binary = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)   # objects white

# OpenCV 4: returns (contours, hierarchy)   [OpenCV 3 returned 3 values]
contours, hierarchy = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contours = [c for c in contours if cv2.contourArea(c) > MIN_AREA]
print("Objects found:", len(contours))

out = img.copy()
cv2.drawContours(out, contours, -1, (0, 255, 0), 2)
for i, c in enumerate(contours):
    x, y, bw, bh = cv2.boundingRect(c)
    cv2.rectangle(out, (x, y), (x + bw, y + bh), (255, 0, 0), 1)
    cv2.putText(out, str(i), (x, y - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)
    print(f"#{i}: area={cv2.contourArea(c):.0f} perimeter={cv2.arcLength(c, True):.0f} box=({x},{y},{bw},{bh})")

items = [(rgb, "Original"), (binary, f"Binary ({METHOD})"), (cv2.cvtColor(out, cv2.COLOR_BGR2RGB), "Contours + boxes")]
if contours:
    big = max(contours, key=cv2.contourArea)
    x, y, bw, bh = cv2.boundingRect(big)
    crop = img[y:y + bh, x:x + bw]
    items.append((cv2.cvtColor(crop, cv2.COLOR_BGR2RGB), "Largest object (cropped)"))
    if OUT_PATH:
        cv2.imwrite(OUT_PATH, crop)

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

**When to use:** 'Extract the boundary of the object using morphology'.

**What to change:** `KERNEL`.

**Exam questions it solves:** Morphological boundary.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
KSIZE = 3

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

_, binary = cv2.threshold(cv2.GaussianBlur(gray, (5, 5), 0), 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
k = np.ones((KSIZE, KSIZE), np.uint8)
inner = cv2.subtract(binary, cv2.erode(binary, k))           # A - erode(A)
outer = cv2.subtract(cv2.dilate(binary, k), binary)          # dilate(A) - A
gradient = cv2.morphologyEx(binary, cv2.MORPH_GRADIENT, k)   # dilate - erode
items = [(binary, "Binary region"), (inner, "Inner boundary"), (outer, "Outer boundary"), (gradient, "Morphological gradient")]

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 5. Hough Transform

Edge pixels VOTE in a parameter space; peaks = shapes. Lines in OpenCV use (rho, theta): `rho = x*cos(theta) + y*sin(theta)` (slope-intercept m,b fails for vertical lines). Circles use (a, b, r) = 3D.

### 5.1 Hough Concept m-b Space

**What it does:** Demonstrates the manual's idea: each point (x, y) becomes a line b = y - m*x in (m, b) space; collinear points intersect at the true (m, b). Votes in an accumulator find it.

**When to use:** Theory/MCQ: parameter space, voting, peak detection.

**What to change:** `POINTS`.

**Exam questions it solves:** Explain / show Hough voting in slope-intercept form.

```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
POINTS = [(1, 3), (2, 5), (3, 7), (4, 9), (5, 4)]     # first four lie on y = 2x + 1, last is an outlier

m_vals = np.round(np.arange(-5, 5.01, 0.1), 2)
b_vals = np.round(np.arange(-10, 10.01, 0.1), 2)
acc = np.zeros((len(b_vals), len(m_vals)), int)

for (x, y) in POINTS:                               # each point votes for all (m, b) lines through it
    b = y - m_vals * x
    bi = np.round((b - b_vals[0]) / 0.1).astype(int)
    ok = (bi >= 0) & (bi < len(b_vals))
    acc[bi[ok], np.arange(len(m_vals))[ok]] += 1

bi, mi = np.unravel_index(np.argmax(acc), acc.shape)
print(f"Peak at m={m_vals[mi]}, b={b_vals[bi]} with {acc[bi, mi]} votes  ->  y = {m_vals[mi]}x + {b_vals[bi]}")

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
xs = np.array([p[0] for p in POINTS]); ys = np.array([p[1] for p in POINTS])
plt.scatter(xs, ys); xx = np.linspace(0, 6, 10); plt.plot(xx, m_vals[mi] * xx + b_vals[bi], "r--"); plt.title("Image space (x, y)")
plt.subplot(1, 3, 2)
for (x, y) in POINTS:
    plt.plot(m_vals, y - m_vals * x)
plt.title("Parameter space (m, b): lines intersect"); plt.xlabel("m"); plt.ylabel("b")
plt.subplot(1, 3, 3)
plt.imshow(acc, extent=[m_vals[0], m_vals[-1], b_vals[-1], b_vals[0]], aspect="auto", cmap="hot"); plt.title("Accumulator"); plt.colorbar()
plt.tight_layout(); plt.show()
```

### 5.2 Hough Lines HoughLines

**What to change:** `IMAGE_PATH`, `THRESHOLD` (min votes; lower = more lines), `RHO`, `THETA`, `MAX_LINES`.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
CANNY_LOW, CANNY_HIGH = 50, 150
RHO = 1                    # distance resolution (pixels)
THETA = np.pi / 180        # angle resolution (radians)
THRESHOLD = 120            # CHANGE THIS: min votes. No lines? lower it. Too many? raise it.
MAX_LINES = 20

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

edges = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 1.4), CANNY_LOW, CANNY_HIGH, apertureSize=3)
lines = cv2.HoughLines(edges, RHO, THETA, THRESHOLD)       # shape (N, 1, 2): [rho, theta]

out = rgb.copy()
if lines is not None:
    print("Lines found:", len(lines))
    for rho, theta in lines.reshape(-1, 2)[:MAX_LINES]:
        a, b = np.cos(theta), np.sin(theta)
        x0, y0 = a * rho, b * rho
        p1 = (int(x0 + 2000 * (-b)), int(y0 + 2000 * a))
        p2 = (int(x0 - 2000 * (-b)), int(y0 - 2000 * a))
        cv2.line(out, p1, p2, (255, 0, 0), 2)
        print(f"rho={rho:.1f}  theta={np.degrees(theta):.1f} deg")
else:
    print("No lines found - lower THRESHOLD")
items = [(rgb, "Original"), (edges, "Canny edges"), (out, "HoughLines")]

# ---- display ----
plt.figure(figsize=(15, 5))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 5.4 Hough Lines From Scratch

**What it does:** Implements the voting yourself: accumulator over (rho, theta), peak picking with suppression, and draws the lines.

**When to use:** 'Implement the Hough transform without cv2.HoughLines', 'build the accumulator'.

**What to change:** `IMAGE_PATH`, `N_LINES`, `MIN_VOTES`, `THETA_STEP`, `NMS`.

**Exam questions it solves:** Hough accumulator implementation.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
N_LINES = 5
MIN_VOTES = 80
THETA_STEP = 1          # degrees
NMS = 10                # suppression window around a peak

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

edges = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 1.4), 50, 150)
ys, xs = np.nonzero(edges)

thetas = np.deg2rad(np.arange(-90, 90, THETA_STEP))
diag = int(np.ceil(np.hypot(h, w)))
rhos = np.arange(-diag, diag + 1)
acc = np.zeros((len(rhos), len(thetas)), np.int32)
cos_t, sin_t = np.cos(thetas), np.sin(thetas)
cols = np.arange(len(thetas))

for x, y in zip(xs, ys):                                   # every edge pixel votes
    r = np.round(x * cos_t + y * sin_t).astype(int) + diag
    acc[r, cols] += 1

work = acc.copy()
peaks = []
for _ in range(N_LINES):                                   # pick the highest peaks
    ri, ti = np.unravel_index(np.argmax(work), work.shape)
    if work[ri, ti] < MIN_VOTES:
        break
    peaks.append((rhos[ri], thetas[ti], work[ri, ti]))
    work[max(0, ri - NMS):ri + NMS + 1, max(0, ti - NMS):ti + NMS + 1] = 0

out = rgb.copy()
for rho, theta, votes in peaks:
    a, b = np.cos(theta), np.sin(theta)
    x0, y0 = a * rho, b * rho
    cv2.line(out, (int(x0 + 2000 * (-b)), int(y0 + 2000 * a)), (int(x0 - 2000 * (-b)), int(y0 - 2000 * a)), (255, 0, 0), 2)
    print(f"rho={rho}, theta={np.degrees(theta):.0f} deg, votes={votes}")

plt.figure(figsize=(14, 5))
plt.subplot(1, 3, 1); plt.imshow(edges, cmap="gray"); plt.title("Edges"); plt.axis("off")
plt.subplot(1, 3, 2); plt.imshow(acc, cmap="hot", aspect="auto", extent=[-90, 90, rhos[-1], rhos[0]]); plt.title("Accumulator (theta, rho)"); plt.xlabel("theta (deg)"); plt.ylabel("rho")
plt.subplot(1, 3, 3); plt.imshow(out); plt.title("Detected lines"); plt.axis("off")
plt.tight_layout(); plt.show()
```

### 5.6 Hough Circles From Scratch

**What it does:** Implements the 3D (a, b, r) voting: each edge pixel votes for centres at distance r in all directions. Shrinks the image for speed.

**When to use:** 'Implement circle Hough transform', 'explain the 3D accumulator'.

**What to change:** `IMAGE_PATH`, `SCALE_W`, `R_MIN`, `R_MAX` (in the SHRUNK image), `N_CIRCLES`, `MIN_FRAC`.

**Exam questions it solves:** Hough circle accumulator.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
SCALE_W = 200            # shrink width for speed
R_MIN, R_MAX = 8, 30    # radius range in the SHRUNK image
N_CIRCLES = 3
MIN_FRAC = 0.4          # fraction of the circumference that must vote
NMS = 8

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

scale = SCALE_W / w
small = cv2.resize(gray, (SCALE_W, int(h * scale)))
sh, sw = small.shape
edges = cv2.Canny(cv2.GaussianBlur(small, (5, 5), 1.0), 50, 150)
ys, xs = np.nonzero(edges)

radii = np.arange(R_MIN, R_MAX + 1)
angles = np.deg2rad(np.arange(0, 360, 1))
cos_a, sin_a = np.cos(angles), np.sin(angles)
acc = np.zeros((len(radii), sh, sw), np.float32)

for ri, r in enumerate(radii):
    a = np.round(xs[:, None] - r * cos_a[None, :]).astype(int)       # candidate centre x
    b = np.round(ys[:, None] - r * sin_a[None, :]).astype(int)       # candidate centre y
    ok = (a >= 0) & (a < sw) & (b >= 0) & (b < sh)
    flat = (b[ok] * sw + a[ok])
    layer = np.bincount(flat, minlength=sh * sw).reshape(sh, sw).astype(np.float32)
    layer = cv2.GaussianBlur(layer, (3, 3), 0)
    acc[ri] = layer / (2 * np.pi * r)                                  # normalise: fraction of circle supporting

found = []
work = acc.copy()
for _ in range(N_CIRCLES):
    ri, cy, cx = np.unravel_index(np.argmax(work), work.shape)
    if work[ri, cy, cx] < MIN_FRAC:
        break
    found.append((int(cx / scale), int(cy / scale), int(radii[ri] / scale), float(work[ri, cy, cx])))
    work[:, max(0, cy - NMS):cy + NMS + 1, max(0, cx - NMS):cx + NMS + 1] = 0

out = rgb.copy()
for x, y, r, v in found:
    cv2.circle(out, (x, y), r, (0, 255, 0), 3); cv2.circle(out, (x, y), 3, (255, 0, 0), -1)
    print(f"centre=({x},{y}) r={r} support={v:.2f}")
if not found:
    print("Nothing found - lower MIN_FRAC or adjust the radius range")
items = [(edges, "Edges (shrunk)"), (out, "Detected circles")]

# ---- display ----
plt.figure(figsize=(14, 6))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 5.7.1 Lane Detection

**What it does:** Canny, then a trapezoid region-of-interest mask (the road), then HoughLinesP; left lanes (negative slope) drawn blue, right lanes red.

**When to use:** 'Detect lane lines on the road', 'detect lines only in the lower region'.

**What to change:** `IMAGE_PATH`, `ROI` polygon fractions, Hough parameters.

**Exam questions it solves:** Lane detection (ROI + Hough).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "road.jpg"      # CHANGE THIS: path to the input image
ROI = [(0.0, 1.0), (0.40, 0.58), (0.60, 0.58), (1.0, 1.0)]   # CHANGE THIS: polygon as (x_frac, y_frac)
THRESHOLD, MIN_LEN, MAX_GAP = 40, 40, 100

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

edges = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 1.4), 50, 150)

mask = np.zeros_like(edges)
poly = np.array([[(int(fx * w), int(fy * h)) for fx, fy in ROI]], np.int32)
cv2.fillPoly(mask, poly, 255)
roi_edges = cv2.bitwise_and(edges, mask)

lines = cv2.HoughLinesP(roi_edges, 1, np.pi / 180, THRESHOLD, minLineLength=MIN_LEN, maxLineGap=MAX_GAP)
out = rgb.copy()
n_left = n_right = 0
if lines is not None:
    for x1, y1, x2, y2 in lines.reshape(-1, 4):
        slope = (y2 - y1) / (x2 - x1 + 1e-9)
        if abs(slope) < 0.3:
            continue                                         # ignore near-horizontal lines
        color = (0, 0, 255) if slope < 0 else (255, 0, 0)    # RGB image: negative slope = left lane (red), positive = right (blue)
        n_left += slope < 0; n_right += slope > 0
        cv2.line(out, (x1, y1), (x2, y2), color, 4)
print("left-lane segments:", n_left, "| right-lane segments:", n_right)
items = [(edges, "Edges"), (roi_edges, "Edges inside ROI"), (out, "Detected lanes")]

# ---- display ----
plt.figure(figsize=(15, 5))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 5.7.2 Deskew Using Hough

**What it does:** Finds long near-horizontal lines with HoughLinesP, takes the median angle, and rotates the image to straighten it.

**When to use:** 'Straighten the tilted document / photograph', 'correct the skew'.

**What to change:** `IMAGE_PATH`, `MIN_LEN_FRAC`, `MAX_TILT`, `OUT_PATH`.

**Exam questions it solves:** Deskew / straighten tilted image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "tilted.jpg"      # CHANGE THIS: path to the input image
MIN_LEN_FRAC = 0.3      # minimum line length as a fraction of image width
MAX_TILT = 45           # only consider lines within +/- this angle of horizontal
OUT_PATH = None         # e.g. 'straight.png'

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

edges = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 1.4), 50, 150)
lines = cv2.HoughLinesP(edges, 1, np.pi / 180, 100, minLineLength=int(MIN_LEN_FRAC * w), maxLineGap=10)

angles = []
if lines is not None:
    for x1, y1, x2, y2 in lines.reshape(-1, 4):
        ang = np.degrees(np.arctan2(y2 - y1, x2 - x1))
        if abs(ang) < MAX_TILT:
            angles.append(ang)
skew = float(np.median(angles)) if angles else 0.0
print(f"Detected skew: {skew:.2f} deg  (from {len(angles)} lines)")

M = cv2.getRotationMatrix2D((w / 2, h / 2), skew, 1.0)     # positive angle = counter-clockwise
straight = cv2.warpAffine(img, M, (w, h), flags=cv2.INTER_CUBIC, borderValue=(255, 255, 255))
if OUT_PATH:
    cv2.imwrite(OUT_PATH, straight)
items = [(rgb, "Tilted"), (cv2.cvtColor(straight, cv2.COLOR_BGR2RGB), f"Straightened (rotated {skew:.1f} deg)")]

# ---- display ----
plt.figure(figsize=(14, 6))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 6.3 Scale-Space and DoG Pyramid

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
SIGMA0 = 1.6
N_LEVELS = 5            # Gaussian levels (gives N_LEVELS-1 DoG layers)
K = 2 ** 0.5

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

g = gray.astype(np.float32) / 255.0
gauss = [cv2.GaussianBlur(g, (0, 0), SIGMA0 * K ** i) for i in range(N_LEVELS)]
dog = [gauss[i + 1] - gauss[i] for i in range(N_LEVELS - 1)]            # D_i = L(sigma_{i+1}) - L(sigma_i)

plt.figure(figsize=(3 * N_LEVELS, 6))
for i, L in enumerate(gauss):
    plt.subplot(2, N_LEVELS, i + 1); plt.imshow(L, cmap="gray"); plt.title(f"L, sigma={SIGMA0 * K ** i:.2f}"); plt.axis("off")
for i, D in enumerate(dog):
    plt.subplot(2, N_LEVELS, N_LEVELS + i + 1)
    plt.imshow(cv2.normalize(np.abs(D), None, 0, 1, cv2.NORM_MINMAX), cmap="gray"); plt.title(f"DoG {i + 1} = L{i + 2} - L{i + 1}"); plt.axis("off")
plt.tight_layout(); plt.show()
```

### 6.4 Scale-Space Extrema 26 Neighbours

**What it does:** Finds candidate keypoints: a pixel in a DoG layer that is larger (or smaller) than all 26 neighbours (8 same layer + 9 above + 9 below).

**When to use:** 'Explain/implement keypoint detection', 'how many neighbours are compared'.

**What to change:** `CONTRAST_THR`, scale parameters.

**Exam questions it solves:** Keypoint localization (extrema detection).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
SIGMA0 = 1.6
N_LEVELS = 5
CONTRAST_THR = 0.02     # ignore weak low-contrast extrema
K = 2 ** 0.5

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

from scipy.ndimage import maximum_filter, minimum_filter

g = gray.astype(np.float32) / 255.0
gauss = [cv2.GaussianBlur(g, (0, 0), SIGMA0 * K ** i) for i in range(N_LEVELS)]
D = np.stack([gauss[i + 1] - gauss[i] for i in range(N_LEVELS - 1)])        # (layers, h, w)

mx = maximum_filter(D, size=(3, 3, 3))          # max over the 3x3x3 cube
mn = minimum_filter(D, size=(3, 3, 3))
ext = ((D == mx) | (D == mn)) & (np.abs(D) > CONTRAST_THR)
ext[0] = False; ext[-1] = False                 # first/last layer have no layer above/below
layer, ys, xs = np.nonzero(ext)
print("Candidate keypoints:", len(xs), "| maxima:", int(np.sum((D == mx) & ext)), "| minima:", int(np.sum((D == mn) & ext)))

# verify one candidate explicitly
for l, y, x in zip(layer, ys, xs):
    if 1 <= y < h - 1 and 1 <= x < w - 1:
        cube = D[l - 1:l + 2, y - 1:y + 2, x - 1:x + 2]
        neighbours = np.delete(cube.ravel(), 13)                 # remove the centre -> 26 values
        print("Neighbours compared:", neighbours.size, "| is extremum:", D[l, y, x] > neighbours.max() or D[l, y, x] < neighbours.min())
        break

plt.figure(figsize=(8, 6)); plt.imshow(rgb)
plt.scatter(xs, ys, c=layer, cmap="coolwarm", s=6); plt.title("DoG extrema (colour = DoG layer)"); plt.axis("off"); plt.show()
```

### 6.5 Orientation Assignment

**What it does:** Builds a 36-bin gradient-orientation histogram (magnitude x Gaussian weight) around the strongest keypoint; the peak is the dominant orientation (rotation invariance).

**When to use:** 'How does SIFT achieve rotation invariance', 'compute the dominant orientation'.

**What to change:** `IMAGE_PATH`, `R` (window radius), `BINS`.

**Exam questions it solves:** Orientation assignment step.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
R = 12                  # window radius around the keypoint
BINS = 36               # 10 degrees per bin

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

sift = cv2.SIFT_create()
kps = sift.detect(gray, None)
kp = max(kps, key=lambda k: k.response)                    # strongest keypoint
x, y = int(kp.pt[0]), int(kp.pt[1])

blur = cv2.GaussianBlur(gray.astype(np.float32), (0, 0), 1.5)
gy, gx = np.gradient(blur)
mag = np.hypot(gx, gy)
ang = np.degrees(np.arctan2(gy, gx)) % 360

y0, y1, x0, x1 = max(0, y - R), min(h, y + R + 1), max(0, x - R), min(w, x + R + 1)
yy, xx = np.mgrid[y0:y1, x0:x1]
weight = np.exp(-((xx - x) ** 2 + (yy - y) ** 2) / (2 * (R / 2) ** 2))              # Gaussian weighting
hist, edges_ = np.histogram(ang[y0:y1, x0:x1], bins=BINS, range=(0, 360), weights=mag[y0:y1, x0:x1] * weight)
dominant = (np.argmax(hist) + 0.5) * (360 / BINS)
print(f"Keypoint at ({x},{y}): dominant orientation = {dominant:.0f} deg | OpenCV kp.angle = {kp.angle:.0f} deg (convention may differ)")

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1); plt.imshow(gray[y0:y1, x0:x1], cmap="gray"); plt.title("Patch around keypoint"); plt.axis("off")
plt.subplot(1, 2, 2); plt.bar(edges_[:-1] + 180 / BINS, hist, width=360 / BINS)
plt.axvline(dominant, color="r"); plt.title("Orientation histogram (peak = dominant)"); plt.xlabel("degrees"); plt.show()
```

### 6.7 Matching BFMatcher

**What it does:** Brute-force matching of SIFT descriptors with Euclidean (L2) distance and cross-check; draws the best matches.

**Exam questions it solves:** Basic keypoint matching.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
PATH1 = "image.jpg"        # CHANGE THIS
PATH2 = "image2.jpg"       # CHANGE THIS
N_DRAW = 50                # number of best matches to draw

img1 = cv2.imread(PATH1); img2 = cv2.imread(PATH2)
assert img1 is not None and img2 is not None, "cannot read images"
rgb1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB); rgb2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY); gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

sift = cv2.SIFT_create()
kp1, d1 = sift.detectAndCompute(gray1, None)
kp2, d2 = sift.detectAndCompute(gray2, None)

bf = cv2.BFMatcher(cv2.NORM_L2, crossCheck=True)           # L2 = Euclidean distance (use NORM_HAMMING for ORB)
matches = sorted(bf.match(d1, d2), key=lambda m: m.distance)
print("keypoints:", len(kp1), len(kp2), "| matches:", len(matches), "| best distance:", matches[0].distance)

result = cv2.drawMatches(rgb1, kp1, rgb2, kp2, matches[:N_DRAW], None, flags=cv2.DRAW_MATCHES_FLAGS_NOT_DRAW_SINGLE_POINTS)
plt.figure(figsize=(16, 7)); plt.imshow(result); plt.title(f"Best {N_DRAW} BF matches"); plt.axis("off"); plt.show()
```

### 6.10 RANSAC Homography and Object Location

**What it does:** Ratio-test matches, then RANSAC (findHomography) rejects outliers; draws only inlier matches and outlines the object in the scene.

**When to use:** 'Remove outliers with RANSAC', 'estimate homography', 'locate the object in the scene'.

**What to change:** `OBJECT_PATH`, `SCENE_PATH`, `RATIO`, `RANSAC_THR` (pixels), `MIN_MATCH`.

**Exam questions it solves:** Outlier rejection + homography + object detection.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
OBJECT_PATH = "image.jpg"      # CHANGE THIS: the object / query image
SCENE_PATH = "image2.jpg"      # CHANGE THIS: the scene / target image
RATIO = 0.75
RANSAC_THR = 5.0               # reprojection error (pixels) allowed for an inlier
MIN_MATCH = 10

obj = cv2.imread(OBJECT_PATH); scene = cv2.imread(SCENE_PATH)
assert obj is not None and scene is not None, "cannot read images"
rgb1 = cv2.cvtColor(obj, cv2.COLOR_BGR2RGB); rgb2 = cv2.cvtColor(scene, cv2.COLOR_BGR2RGB)
gray1 = cv2.cvtColor(obj, cv2.COLOR_BGR2GRAY); gray2 = cv2.cvtColor(scene, cv2.COLOR_BGR2GRAY)

sift = cv2.SIFT_create()
kp1, d1 = sift.detectAndCompute(gray1, None)
kp2, d2 = sift.detectAndCompute(gray2, None)
knn = cv2.BFMatcher(cv2.NORM_L2).knnMatch(d1, d2, k=2)
good = [p[0] for p in knn if len(p) == 2 and p[0].distance < RATIO * p[1].distance]
print("good matches before RANSAC:", len(good))

if len(good) >= MIN_MATCH:
    src = np.float32([kp1[m.queryIdx].pt for m in good]).reshape(-1, 1, 2)
    dst = np.float32([kp2[m.trainIdx].pt for m in good]).reshape(-1, 1, 2)
    H, mask = cv2.findHomography(src, dst, cv2.RANSAC, RANSAC_THR)       # mask: 1 = inlier, 0 = outlier
    inliers = mask.ravel().tolist()
    print("inliers after RANSAC:", sum(inliers), "| outliers removed:", len(good) - sum(inliers))
    print("Homography H =\n", H)

    oh, ow = gray1.shape
    corners = np.float32([[0, 0], [0, oh - 1], [ow - 1, oh - 1], [ow - 1, 0]]).reshape(-1, 1, 2)
    projected = cv2.perspectiveTransform(corners, H)                      # object corners in the scene
    scene_vis = cv2.polylines(rgb2.copy(), [np.int32(projected)], True, (255, 0, 0), 4, cv2.LINE_AA)

    result = cv2.drawMatches(rgb1, kp1, scene_vis, kp2, good, None, matchColor=(0, 255, 0), singlePointColor=None,
                             matchesMask=inliers, flags=cv2.DRAW_MATCHES_FLAGS_NOT_DRAW_SINGLE_POINTS)
    plt.figure(figsize=(16, 7)); plt.imshow(result); plt.title("Inlier matches + object outline"); plt.axis("off"); plt.show()
else:
    print("Not enough matches:", len(good), "<", MIN_MATCH)
```

#### 6.11.5 Invariance Test

**What it does:** Rotates / scales / brightens an image and counts SIFT matches against the original, showing SIFT's invariance (rotation, scale, brightness).

**When to use:** 'Show that SIFT is invariant to rotation and scale', 'how does matching change with transformation'.

**What to change:** `IMAGE_PATH`, `ANGLES`, `SCALES`.

**Exam questions it solves:** Invariance demonstration.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"      # CHANGE THIS: path to the input image
ANGLES = [0, 30, 90, 180]
SCALES = [1.0, 0.5, 1.5]

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

sift = cv2.SIFT_create()
kp0, d0 = sift.detectAndCompute(gray, None)
bf = cv2.BFMatcher(cv2.NORM_L2)

def count_good(im, ratio=0.75):
    kp, d = sift.detectAndCompute(im, None)
    if d is None or len(d) < 2:
        return 0
    return sum(1 for p in bf.knnMatch(d0, d, k=2) if len(p) == 2 and p[0].distance < ratio * p[1].distance)

labels, counts = [], []
for s in SCALES:
    for a in ANGLES:
        M = cv2.getRotationMatrix2D((w / 2, h / 2), a, s)
        t = cv2.warpAffine(gray, M, (w, h))
        labels.append(f"rot {a}, scale {s}"); counts.append(count_good(t))
        print(f"rotation {a:>3} deg, scale {s}: {counts[-1]} good matches")
bright = cv2.convertScaleAbs(gray, alpha=1.0, beta=50)
print("brightness +50:", count_good(bright)); labels.append("brightness +50"); counts.append(count_good(bright))

plt.figure(figsize=(12, 4)); plt.bar(range(len(counts)), counts); plt.xticks(range(len(counts)), labels, rotation=75)
plt.ylabel("good matches"); plt.title("SIFT matches vs transformation"); plt.tight_layout(); plt.show()
```

#### 6.11.6 Same Object Decision

**What it does:** Decides whether two images show the same object using the number of RANSAC inliers against a threshold.

**When to use:** 'Write a function that says whether two images contain the same object'.

**Exam questions it solves:** Image similarity / verification.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
PAIRS = [("image.jpg", "image2.jpg"), ("image.jpg", "shapes.jpg")]   # CHANGE THIS: image pairs to compare
MIN_INLIERS = 15                                                          # CHANGE THIS: decision threshold

def sift_match_homography(gray1, gray2, ratio=0.75, ransac_thr=5.0):
    # SIFT + KNN ratio test + RANSAC homography (maps image1 points -> image2 points)
    sift = cv2.SIFT_create()
    kp1, d1 = sift.detectAndCompute(gray1, None)
    kp2, d2 = sift.detectAndCompute(gray2, None)
    knn = cv2.BFMatcher(cv2.NORM_L2).knnMatch(d1, d2, k=2)
    good = [p[0] for p in knn if len(p) == 2 and p[0].distance < ratio * p[1].distance]
    H, mask = None, None
    if len(good) >= 4:
        src = np.float32([kp1[m.queryIdx].pt for m in good]).reshape(-1, 1, 2)
        dst = np.float32([kp2[m.trainIdx].pt for m in good]).reshape(-1, 1, 2)
        H, mask = cv2.findHomography(src, dst, cv2.RANSAC, ransac_thr)
    return kp1, kp2, good, H, mask

def compare(p1, p2):
    a = cv2.imread(p1, 0); b = cv2.imread(p2, 0)
    assert a is not None and b is not None, "cannot read images"
    kp1, kp2, good, H, mask = sift_match_homography(a, b)
    return len(good), (int(mask.sum()) if mask is not None else 0)

for p1, p2 in PAIRS:
    n_good, n_inl = compare(p1, p2)
    verdict = "SAME OBJECT" if n_inl >= MIN_INLIERS else "DIFFERENT"
    print(f"{p1} vs {p2}: good={n_good}, inliers={n_inl} -> {verdict}")
```

### 7.1 Document Scanner

**What it does:** Detects the page boundary (Canny, contours, 4-corner polygon), then corrects the perspective into a flat rectangle, and makes a black-and-white scan.

**When to use:** 'Scan the document', 'correct the perspective', 'straighten the photograph of the page', 'crop the page'.

**What to change:** `IMAGE_PATH`, `MANUAL_POINTS` (use if automatic detection fails; order any), output size.

**Exam questions it solves:** Boundary detection + perspective transform.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "document.jpg"      # CHANGE THIS: path to the input image
MANUAL_POINTS = None     # CHANGE THIS if needed: [(x1,y1),(x2,y2),(x3,y3),(x4,y4)] four page corners
OUT_PATH = None         # e.g. 'scan.png'

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

def order_points(pts):
    pts = np.array(pts, dtype="float32").reshape(4, 2)
    s = pts.sum(axis=1); d = np.diff(pts, axis=1).ravel()
    return np.array([pts[np.argmin(s)], pts[np.argmin(d)], pts[np.argmax(s)], pts[np.argmax(d)]], dtype="float32")   # TL, TR, BR, BL

if MANUAL_POINTS is not None:
    quad = np.array(MANUAL_POINTS, dtype="float32")
else:
    edges = cv2.Canny(cv2.GaussianBlur(gray, (5, 5), 0), 50, 150)
    edges = cv2.dilate(edges, np.ones((3, 3), np.uint8))
    contours, _ = cv2.findContours(edges, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
    quad = None
    for c in sorted(contours, key=cv2.contourArea, reverse=True)[:5]:
        approx = cv2.approxPolyDP(c, 0.02 * cv2.arcLength(c, True), True)
        if len(approx) == 4:
            quad = approx; break
    assert quad is not None, "no 4-corner contour found - set MANUAL_POINTS"

rect = order_points(quad)
(tl, tr, br, bl) = rect
out_w = int(max(np.linalg.norm(br - bl), np.linalg.norm(tr - tl)))
out_h = int(max(np.linalg.norm(tr - br), np.linalg.norm(tl - bl)))
dst = np.array([[0, 0], [out_w - 1, 0], [out_w - 1, out_h - 1], [0, out_h - 1]], dtype="float32")
M = cv2.getPerspectiveTransform(rect, dst)
warped = cv2.warpPerspective(img, M, (out_w, out_h))
scan = cv2.adaptiveThreshold(cv2.cvtColor(warped, cv2.COLOR_BGR2GRAY), 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 10)
if OUT_PATH:
    cv2.imwrite(OUT_PATH, warped)

marked = rgb.copy(); cv2.polylines(marked, [rect.astype(np.int32)], True, (255, 0, 0), 4)
items = [(marked, "Detected page corners"), (cv2.cvtColor(warped, cv2.COLOR_BGR2RGB), f"Corrected perspective {out_w}x{out_h}"), (scan, "Scanned look")]

# ---- display ----
plt.figure(figsize=(16, 6))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 7.2 Shape Detection and Counting

**What it does:** Finds each shape's contour, approximates the polygon, and names it by the number of vertices (triangle, square, rectangle, pentagon, circle); counts each type.

**When to use:** 'Identify and count triangles, squares, circles', 'label each shape'.

**What to change:** `IMAGE_PATH`, `MIN_AREA`, `EPS` (polygon approximation tightness).

**Exam questions it solves:** Shape classification via approxPolyDP.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
MIN_AREA = 800
EPS = 0.03                # approxPolyDP epsilon as a fraction of the perimeter

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

blur = cv2.GaussianBlur(gray, (5, 5), 0)
_, binary = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

out = rgb.copy(); counts = {}
for c in contours:
    area = cv2.contourArea(c)
    if area < MIN_AREA:
        continue
    peri = cv2.arcLength(c, True)
    approx = cv2.approxPolyDP(c, EPS * peri, True)
    v = len(approx)
    if v == 3:
        name = "Triangle"
    elif v == 4:
        x, y, bw, bh = cv2.boundingRect(approx)
        name = "Square" if 0.9 <= bw / float(bh) <= 1.1 else "Rectangle"
    elif v == 5:
        name = "Pentagon"
    else:
        circularity = 4 * np.pi * area / (peri ** 2)
        name = "Circle" if circularity > 0.8 else f"Polygon({v})"
    counts[name] = counts.get(name, 0) + 1
    M = cv2.moments(c); cx, cy = int(M["m10"] / M["m00"]), int(M["m01"] / M["m00"])
    cv2.drawContours(out, [c], -1, (0, 255, 0), 3)
    cv2.putText(out, name, (cx - 35, cy), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
    cv2.putText(out, name, (cx - 35, cy), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 1)
print("Counts:", counts)
items = [(rgb, "Original"), (binary, "Binary"), (out, f"Detected: {counts}")]

# ---- display ----
plt.figure(figsize=(16, 5))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(1, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 7.3 Extract Object Using Edges

**What it does:** Cuts out the largest object: (A) from its contour mask (Canny + fill) and (B) with GrabCut from a rectangle.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
RECT = (40, 40, 240, 180)       # CHANGE THIS: x, y, w, h box around the object (for GrabCut)
OUT_PATH = None

# ---- load image ----
img = cv2.imread(IMAGE_PATH)                          # BGR (OpenCV order)
assert img is not None, f"Cannot read {IMAGE_PATH}"
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)            # RGB for matplotlib
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # grayscale
h, w = gray.shape

# ---- (A) contour mask ----
blur = cv2.GaussianBlur(gray, (5, 5), 1.4)
edges = cv2.morphologyEx(cv2.Canny(blur, 50, 150), cv2.MORPH_CLOSE, np.ones((7, 7), np.uint8))
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
big = max(contours, key=cv2.contourArea)
mask = np.zeros(gray.shape, np.uint8)
cv2.drawContours(mask, [big], -1, 255, thickness=cv2.FILLED)         # filled silhouette
cut = cv2.bitwise_and(img, img, mask=mask)
x, y, bw, bh = cv2.boundingRect(big)
cropped = cut[y:y + bh, x:x + bw]
if OUT_PATH:
    cv2.imwrite(OUT_PATH, cropped)

# ---- (B) GrabCut (needs a rough rectangle) ----
gc_mask = np.zeros(gray.shape, np.uint8)
bgd, fgd = np.zeros((1, 65), np.float64), np.zeros((1, 65), np.float64)
cv2.grabCut(img, gc_mask, RECT, bgd, fgd, 5, cv2.GC_INIT_WITH_RECT)
fg = np.where((gc_mask == cv2.GC_FGD) | (gc_mask == cv2.GC_PR_FGD), 255, 0).astype(np.uint8)
grab = cv2.bitwise_and(img, img, mask=fg)

items = [(rgb, "Original"), (mask, "Filled contour mask"), (cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB), "Largest object cropped"), (cv2.cvtColor(grab, cv2.COLOR_BGR2RGB), "GrabCut result")]

# ---- display ----
plt.figure(figsize=(12, 8))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 2, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 7.4 Wavelet Denoise then Canny then Hough

**What it does:** Full pipeline: add noise, denoise by wavelet thresholding, detect edges, detect line segments and circles; compares with the non-denoised result.

**When to use:** 'Remove noise using wavelets and then detect lines / shapes', multi-step processing questions.

**What to change:** `IMAGE_PATH`, `NOISE_SIGMA`, `WAVELET`, Hough parameters.

**Exam questions it solves:** End-to-end pipeline.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pywt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "shapes.jpg"      # CHANGE THIS: path to the input image
NOISE_SIGMA = 20
WAVELET = "db4"
LEVEL = 2
HOUGH_THRESH, MIN_LEN, MAX_GAP = 80, 50, 10

# ---- load grayscale image ----
gray = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
assert gray is not None, f"Cannot read {IMAGE_PATH}"
h, w = gray.shape
g = gray.astype(np.float32)
norm8 = lambda x: cv2.normalize(np.abs(x), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
to8 = lambda x: np.clip(x, 0, 255).astype(np.uint8)

rng = np.random.default_rng(0)
noisy = g + rng.normal(0, NOISE_SIGMA, g.shape)

def wavelet_denoise(x):
    c = pywt.wavedec2(x, WAVELET, level=LEVEL)
    thr = np.median(np.abs(c[-1][-1])) / 0.6745 * np.sqrt(2 * np.log(x.size))
    c2 = [c[0]] + [tuple(pywt.threshold(a, thr, mode="soft") for a in lvl) for lvl in c[1:]]
    return to8(pywt.waverec2(c2, WAVELET)[:h, :w])

noisy8 = to8(noisy)
den = wavelet_denoise(noisy)
e_noisy = cv2.Canny(noisy8, 50, 150)
e_den = cv2.Canny(cv2.GaussianBlur(den, (3, 3), 0), 50, 150)

lines = cv2.HoughLinesP(e_den, 1, np.pi / 180, HOUGH_THRESH, minLineLength=MIN_LEN, maxLineGap=MAX_GAP)
circles = cv2.HoughCircles(cv2.medianBlur(den, 5), cv2.HOUGH_GRADIENT, 1.2, 30, param1=100, param2=40, minRadius=10, maxRadius=100)
out = cv2.cvtColor(den, cv2.COLOR_GRAY2RGB)
if lines is not None:
    for x1, y1, x2, y2 in lines.reshape(-1, 4):
        cv2.line(out, (x1, y1), (x2, y2), (255, 0, 0), 2)
if circles is not None:
    for x, y, r in np.round(circles.reshape(-1, 3)).astype(int):
        cv2.circle(out, (x, y), r, (0, 255, 0), 2)
print("PSNR noisy:", round(cv2.PSNR(gray, noisy8), 2), "| PSNR denoised:", round(cv2.PSNR(gray, den), 2))
items = [(noisy8, "Noisy"), (e_noisy, f"Canny on noisy ({np.count_nonzero(e_noisy)} px)"), (den, "Wavelet denoised"),
         (e_den, f"Canny on denoised ({np.count_nonzero(e_den)} px)"), (out, "Lines (blue) and circles (green)")]

# ---- display ----
plt.figure(figsize=(16, 9))
for i, (im, t) in enumerate(items, 1):
    plt.subplot(2, 3, i)
    plt.imshow(im, cmap=None if im.ndim == 3 else "gray")
    plt.title(t, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 8. Theory and Cheat Sheet

### 8.1 Key Concepts

**Wavelets**
- Wavelet transform = frequency content at different scales AND where it occurs (time-frequency). Fourier loses "where".
- CWT: continuous scales and shifts (analysis, Morlet). DWT: discrete (compression, denoising, features; Haar, Daubechies, Symlet).
- WPT: also splits the detail subbands (more flexible, 4^L subbands). Inverse transform rebuilds from coefficients.
- 2D DWT subbands: LL (approximation), LH (horizontal edges), HL (vertical edges), HH (diagonal). Each is half size.
- Denoising = transform, threshold small detail coefficients, inverse. Compression = keep only the largest coefficients.

**Edge detection**
- Edge = sharp change in intensity. Gradient = direction and strength of change.
- Sobel: 3x3 kernels Gx (vertical edges) and Gy (horizontal edges); magnitude = sqrt(Gx^2 + Gy^2); direction = atan2(Gy, Gx). Uniform area gives 0. Thick edges, noise sensitive.
- Canny: 1 Gaussian blur, 2 gradient magnitude and direction, 3 non-maximum suppression (thin to 1 px), 4 hysteresis (strong kept; weak kept only if connected to strong; below low discarded).
- LoG: Gaussian blur then Laplacian (2nd derivative); edge = zero crossing. Sigma controls scale (large sigma = only coarse edges).
- 1st derivative = peak at the edge. 2nd derivative = zero crossing at the edge.

**Hough**
- Edge pixels vote in parameter space; peaks = shapes. Needs an edge map first (Canny).
- Line: y = mx + b gives (m, b) space, but OpenCV uses (rho, theta) because vertical lines have infinite slope. Each point gives a line (m,b) or a sinusoid (rho,theta).
- Circle: (x-a)^2 + (y-b)^2 = r^2 gives a 3D space (a, b, r). OpenCV HoughCircles uses the gradient method (a 2D accumulator for centres, then radius).

**SIFT**
1. Scale-space extrema: Difference of Gaussians D = L(k*sigma) - L(sigma); max/min over 26 neighbours (8 + 9 + 9).
2. Keypoint localization: fit a 3D quadratic to refine position; reject low contrast and edge-like points.
3. Orientation assignment: dominant gradient direction in a local histogram = rotation invariance.
4. Descriptor: 4x4 sub-regions x 8 orientation bins = 128 numbers (robust to illumination).
5. Matching: Euclidean distance (+ ratio test). 6. RANSAC: rejects geometrically inconsistent matches. 7. Homography: 3x3 transform between images (stitching, recognition).
- Invariant to scale, rotation, (mostly) illumination, partially to viewpoint and noise.

### 8.3 Parameter Effects

| Change | Effect |
|---|---|
| Canny thresholds lower | more edges, more noise |
| Canny thresholds higher | fewer edges, only strong ones |
| Larger Gaussian sigma / kernel | smoother, less noise, blurrier and fewer fine edges |
| Sobel `ksize` larger | smoother gradient, thicker response |
| LoG sigma larger | only coarse (large-scale) edges |
| `HoughLines` threshold lower | more lines (more false) |
| `HoughLinesP` `minLineLength` larger | only long segments |
| `HoughLinesP` `maxGap` larger | joins broken segments |
| `HoughCircles` `param2` lower | more circles (more false) |
| `HoughCircles` `dp` larger | coarser accumulator |
| SIFT `contrastThreshold` higher | fewer keypoints |
| SIFT `edgeThreshold` higher | more edge-like keypoints kept |
| Ratio test lower (0.6) | fewer, more reliable matches |
| RANSAC threshold higher | more inliers accepted, less precise |
| Wavelet threshold higher | stronger denoising / more blur |
| Wavelet keep% lower | more compression, lower quality |
| More wavelet levels | coarser approximation, smaller LL |

### 8.4 Exam Traps

- OpenCV loads **BGR**; matplotlib needs **RGB**. Grayscale needs `cmap="gray"`.
- Sobel must use `cv2.CV_64F` (or CV_16S): uint8 clips negative gradients to 0. Use `np.abs` / `convertScaleAbs` before display.
- `cv2.Sobel(..., 1, 0)` = dx=1 = **vertical edges**. `(0, 1)` = **horizontal edges**.
- `cv2.resize` takes **(width, height)**; array shape is **(height, width)**; slicing is `img[y1:y2, x1:x2]`.
- Canny needs uint8 input and returns uint8 0/255. Typical `high` = 2 to 3 x `low`.
- `cv2.Laplacian` output also needs `CV_64F` and abs for display; true LoG edges are zero crossings, not the bright pixels.
- `HoughLines` returns (rho, theta) with theta in **radians**; `HoughLinesP` returns pixel endpoints. Both need an edge map. If the result is `None`, nothing was found (check before looping).
- `HoughCircles` takes a **grayscale** blurred image (not edges); output shape `(1, N, 3)`.
- `findContours` returns 2 values in OpenCV 4 (3 in OpenCV 3). Objects must be **white** on black.
- SIFT needs grayscale uint8; descriptors are **float32 (N, 128)**; use `NORM_L2` (SIFT) vs `NORM_HAMMING` (ORB/AKAZE).
- Homography needs at least **4** matches; RANSAC mask: 1 = inlier. `findHomography(src=query pts, dst=train pts)`.
- `drawMatches` needs both keypoint lists and both images in the same order as the matcher call.
- Wavelet: `waverec2` output can be 1 pixel larger for odd sizes - crop to original. Threshold only the **details**, never `cA`.
- `pywt.dwt2` returns `(cA, (cH, cV, cD))` and `wavedec2` lists coarsest first.
- Image stitching: warp the **right** image into the **left** frame (H maps right to left).
