### 2.1 Feature Extraction Overview

A **feature** is a short list of numbers that describes an image. **Local** features describe small spots
(corners, keypoints). **Global** features describe the whole image (color histogram, texture).

| Technique | What it counts / measures | Output size |
|---|---|---|
| Color histogram | pixels per color bin | bins x channels |
| HOG | edge directions per small cell, block normalised | 3780 for 128x64 with defaults |
| LBP | local neighbour patterns | 10 bins (uniform, 8 points) |
| HED | edge pixel directions, 8 bins, 0-360 | 8 |
| HIG | gradient directions, 9 bins, 0-180 | 9 |
| Texture energy / contrast | window strength / window spread | 256 bins each |

### 2.2 Filtering and Convolution

A filter (kernel) is a small grid of numbers slid over the image. Kernel sums to 1 = blur, sums to 0 = edge detector.

#### 2.2.1 Box Blur

**What it does:** Replaces each pixel by the average of its KxK neighbours (convolution with all 1/(K*K) kernel).

**When to use:** Simple smoothing or noise reduction.

**What to change:** IMAGE_PATH, KSIZE (odd; bigger = blurrier).

**Exam wording:** 'blur the image', 'apply an averaging filter', 'convolve with a 3x3 box kernel'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
KSIZE = 5   # CHANGE THIS: odd kernel size (3, 5, 9 ...)


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH)   # color image, BGR order
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.full((240, 300, 3), 200, np.uint8)
    cv2.rectangle(img, (20, 30), (110, 120), (0, 0, 255), -1)    # red
    cv2.circle(img, (200, 80), 50, (0, 200, 0), -1)              # green
    cv2.rectangle(img, (60, 150), (170, 220), (255, 0, 0), -1)   # blue
    cv2.circle(img, (250, 190), 35, (0, 255, 255), -1)           # yellow

kernel = np.ones((KSIZE, KSIZE), np.float32) / (KSIZE * KSIZE)   # all values 1/(K*K), sums to 1
blurred = cv2.filter2D(img, -1, kernel)                          # -1 = same data type as input
# Same thing with shortcuts: cv2.blur(img, (KSIZE, KSIZE))  or  cv2.boxFilter(img, -1, (KSIZE, KSIZE))

show([img, blurred], ["Original", f"Box blur {KSIZE}x{KSIZE}"])
cv2.imwrite("box_blur_result.jpg", blurred)   # optional: save result
```

#### 2.2.2 Gaussian Blur

**What it does:** Blurs with a bell shaped kernel (centre pixels count more) so it looks smoother than a box blur.

**When to use:** Noise reduction, pre-step before edge detection / thresholding.

**What to change:** IMAGE_PATH, KSIZE (odd), SIGMA (0 = automatic).

**Exam wording:** 'apply Gaussian smoothing', 'build a Gaussian kernel and convolve'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
KSIZE = 5   # CHANGE THIS: odd kernel size
SIGMA = 0   # CHANGE THIS: 0 = auto, larger = stronger blur


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH)   # color image, BGR order
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.full((240, 300, 3), 200, np.uint8)
    cv2.rectangle(img, (20, 30), (110, 120), (0, 0, 255), -1)    # red
    cv2.circle(img, (200, 80), 50, (0, 200, 0), -1)              # green
    cv2.rectangle(img, (60, 150), (170, 220), (255, 0, 0), -1)   # blue
    cv2.circle(img, (250, 190), 35, (0, 255, 255), -1)           # yellow

blur = cv2.GaussianBlur(img, (KSIZE, KSIZE), SIGMA)       # the normal way

# Manual way: getGaussianKernel is 1-D, so multiply by its transpose to get the 2-D kernel
sig = SIGMA if SIGMA > 0 else 0.3 * ((KSIZE - 1) * 0.5 - 1) + 0.8
k1d = cv2.getGaussianKernel(KSIZE, sig)
k2d = k1d @ k1d.T
blur_manual = cv2.filter2D(img, -1, k2d)
print("2-D kernel sums to:", k2d.sum())
print("Max difference between both methods:", np.abs(blur.astype(int) - blur_manual.astype(int)).max())

show([img, blur, blur_manual], ["Original", "cv2.GaussianBlur", "Manual kernel + filter2D"])
```

#### 2.2.3 Kernel Gallery Sharpen Emboss Outline

**What it does:** Applies common 3x3 kernels with filter2D and shows all of them.

**When to use:** Question gives a kernel (or names an effect) and asks you to apply it.

#### 2.2.4 Convolution Types Valid Same Stride

**What it does:** Implements convolution with valid / same (zero padding) / stride and shows output sizes. Also shows the OpenCV equivalent.

**When to use:** Questions about output size, padding, or 'implement convolution yourself'.

**What to change:** KERNEL, STRIDE. Formulas: valid = N-K+1, same = N, valid with stride S = floor((N-K)/S)+1.

**Exam wording:** 'perform valid convolution', 'apply zero padding', 'convolution with stride 2'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
KERNEL = np.array([[1, 0, -1], [2, 0, -2], [1, 0, -1]], np.float32)   # CHANGE THIS: your kernel (odd size)
STRIDE = 2   # CHANGE THIS


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

from numpy.lib.stride_tricks import sliding_window_view

def convolve(image, kernel, padding="valid", stride=1):
    """padding: 'valid' (no padding, smaller output) or 'same' (zero padding, same size)."""
    image = image.astype(np.float32)
    kh, kw = kernel.shape
    if padding == "same":
        image = np.pad(image, ((kh // 2, kh // 2), (kw // 2, kw // 2)), mode="constant")   # zeros
    windows = sliding_window_view(image, (kh, kw))            # every KxK patch
    out = np.einsum("ijkl,kl->ij", windows, kernel)           # multiply patch by kernel and sum
    return out[::stride, ::stride]                            # stride = skip positions

valid  = convolve(img, KERNEL, "valid", 1)
same   = convolve(img, KERNEL, "same", 1)
strided = convolve(img, KERNEL, "valid", STRIDE)
H_, W_ = img.shape
K = KERNEL.shape[0]
print("Input size     :", img.shape)
print("Valid          :", valid.shape, "expected", (H_ - K + 1, W_ - K + 1))
print("Same           :", same.shape)
print(f"Valid stride {STRIDE} :", strided.shape, "expected", ((H_ - K) // STRIDE + 1, (W_ - K) // STRIDE + 1))

# OpenCV equivalents (filter2D ALWAYS returns same size; there is NO 'strides' argument)
cv_same = cv2.filter2D(img.astype(np.float32), -1, KERNEL, borderType=cv2.BORDER_CONSTANT)  # zero padding
p = K // 2
cv_valid = cv_same[p:-p, p:-p]            # crop border -> valid
cv_strided = cv_valid[::STRIDE, ::STRIDE] # slice -> stride
print("Same matches OpenCV:", np.allclose(same, cv_same, atol=1e-3))

show([img, cv2.convertScaleAbs(valid), cv2.convertScaleAbs(same), cv2.convertScaleAbs(strided)],
     ["Original", "Valid", "Same (zero pad)", f"Valid stride {STRIDE}"])
```

#### 2.3.5 LoG Marr-Hildreth

**What it does:** Gaussian blur then Laplacian, edges are where the response crosses zero (zero-crossings).

**When to use:** 'Laplacian of Gaussian', 'Marr-Hildreth', 'zero crossings'.

**What to change:** SIGMA (blur strength), MIN_STRENGTH (ignore weak crossings).

**Exam wording:** 'apply LoG', 'detect edges using zero crossings'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
SIGMA = 2.0          # CHANGE THIS: bigger = smoother, fewer edges
MIN_STRENGTH = 4.0   # CHANGE THIS: bigger = fewer weak edges


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

blur = cv2.GaussianBlur(img, (0, 0), SIGMA)                 # (0,0) -> size computed from sigma
log = cv2.Laplacian(blur, cv2.CV_64F, ksize=3)

pos = log > 0                                               # sign of the response
change_v = pos[:-1, :-1] != pos[1:, :-1]                    # sign flips going down
change_h = pos[:-1, :-1] != pos[:-1, 1:]                    # sign flips going right
strong = (np.abs(log[:-1, :-1] - log[1:, :-1]) > MIN_STRENGTH) | (np.abs(log[:-1, :-1] - log[:-1, 1:]) > MIN_STRENGTH)
zero_cross = np.zeros(log.shape, np.uint8)
zero_cross[:-1, :-1] = ((change_v | change_h) & strong).astype(np.uint8) * 255

show([img, cv2.convertScaleAbs(log), zero_cross], ["Original", "LoG response", "Zero crossings (edges)"])
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
LOW = 100    # CHANGE THIS
HIGH = 200   # CHANGE THIS
BLUR = 5     # CHANGE THIS: odd


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

blur = cv2.GaussianBlur(img, (BLUR, BLUR), 0)      # Canny does NOT blur by itself
edges = cv2.Canny(blur, LOW, HIGH)                 # first number = LOW, second = HIGH
edges_l2 = cv2.Canny(blur, LOW, HIGH, apertureSize=3, L2gradient=True)   # more accurate magnitude

edges_many = cv2.Canny(blur, LOW // 3, HIGH // 3)  # sensitive: many edges
edges_few  = cv2.Canny(blur, LOW * 2, HIGH * 2)    # strict: few edges

show([img, edges, edges_l2, edges_many, edges_few],
     ["Original", f"Canny {LOW},{HIGH}", "L2gradient=True", "Lower thresholds", "Higher thresholds"], cols=3)
cv2.imwrite("canny_edges.png", edges)
```

#### 2.3.7 Auto Canny

**What it does:** Picks Canny thresholds from the image median so no manual tuning is needed.

**When to use:** When no thresholds are given or the image brightness is unknown.

**What to change:** SIGMA (0.33 default).

**Exam wording:** 'detect edges without manually setting thresholds'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
SIGMA = 0.33   # CHANGE THIS: bigger = wider threshold band


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

blur = cv2.GaussianBlur(img, (5, 5), 0)
v = np.median(blur)
lower = int(max(0, (1.0 - SIGMA) * v))
upper = int(min(255, (1.0 + SIGMA) * v))
edges = cv2.Canny(blur, lower, upper)
print("Thresholds used:", lower, upper)
show([img, edges], ["Original", f"Auto Canny ({lower},{upper})"])
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import hog
from skimage import exposure

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
RESIZE_TO = (64, 128)   # CHANGE THIS: (width, height) or None
ORIENT = 9
PPC = (8, 8)
CPB = (2, 2)


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

if RESIZE_TO:
    img = cv2.resize(img, RESIZE_TO)           # cv2.resize takes (width, height)

features, hog_image = hog(img, orientations=ORIENT, pixels_per_cell=PPC,
                          cells_per_block=CPB, block_norm="L2-Hys", visualize=True)
# colour image version: hog(color_img, ..., channel_axis=-1)
hog_rescaled = exposure.rescale_intensity(hog_image, in_range=(0, hog_image.max()))

h, w = img.shape
blocks_y = h // PPC[0] - CPB[0] + 1
blocks_x = w // PPC[1] - CPB[1] + 1
expected = blocks_y * blocks_x * CPB[0] * CPB[1] * ORIENT
print("Image shape (h, w):", img.shape)
print("Feature vector length:", features.shape[0], "| expected:", expected)
print("First 9 values:", np.round(features[:9], 3))

show([img, hog_rescaled], ["Original", "HOG visualisation"])
```

#### 2.4.4 LBP Local Binary Pattern

**What it does:** Compares every pixel with its neighbours (>= centre gives 1), forms a binary code, histogram of codes = texture feature.

**When to use:** Texture classification, face recognition.

**What to change:** RADIUS, N_POINTS (8*radius), METHOD ('uniform' gives n_points+2 bins, 'default' gives 2**n_points).

**Exam wording:** 'extract LBP features', 'LBP histogram'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import local_binary_pattern

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
RADIUS = 1
N_POINTS = 8 * RADIUS
METHOD = "uniform"   # CHANGE THIS: "uniform" or "default"

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

lbp = local_binary_pattern(img, N_POINTS, RADIUS, method=METHOD)
n_bins = N_POINTS + 2 if METHOD == "uniform" else 2 ** N_POINTS
hist, _ = np.histogram(lbp.ravel(), bins=n_bins, range=(0, n_bins))
hist = hist.astype("float")
hist /= (hist.sum() + 1e-6)                    # fractions, sums to 1 = feature vector

print("LBP feature vector length:", len(hist))
print(np.round(hist, 3))

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1); plt.imshow(img, cmap="gray"); plt.title("Original"); plt.axis("off")
plt.subplot(1, 3, 2); plt.imshow(lbp, cmap="gray"); plt.title("LBP image"); plt.axis("off")
plt.subplot(1, 3, 3); plt.bar(range(n_bins), hist); plt.title("LBP histogram"); plt.xlabel("pattern"); plt.ylabel("fraction")
plt.tight_layout(); plt.show()
```

#### 2.4.5 LBP Grid Histogram

**What it does:** LBP histogram per grid cell, joined together, keeps location information (better for faces).

**When to use:** Face recognition style features.

**What to change:** GRID.

**Exam wording:** 'divide the image into regions and compute LBP of each'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import local_binary_pattern

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
GRID = (4, 4)   # CHANGE THIS: (rows, cols)

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

lbp = local_binary_pattern(img, 8, 1, method="uniform")
n_bins = 10
gh, gw = GRID
h, w = lbp.shape
vector = []
for i in range(gh):
    for j in range(gw):
        cell = lbp[i * h // gh:(i + 1) * h // gh, j * w // gw:(j + 1) * w // gw]
        hist, _ = np.histogram(cell.ravel(), bins=n_bins, range=(0, n_bins))
        vector.extend(hist / (hist.sum() + 1e-6))
vector = np.array(vector)
print("Feature vector length:", vector.shape[0], "=", gh * gw, "cells x", n_bins, "bins")
```

#### 2.4.9 Compare Two Images by Histogram

**What it does:** Similarity score between two images using colour histograms (4 methods).

**When to use:** 'which image is more similar', 'compare histograms'.

**What to change:** IMAGE_A, IMAGE_B.

**Exam wording:** 'measure similarity of two images'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

IMAGE_A = "image1.jpg"   # CHANGE THIS
IMAGE_B = "image2.jpg"   # CHANGE THIS
a = cv2.imread(IMAGE_A)
b = cv2.imread(IMAGE_B)
if a is None or b is None:
    print("Images not found -> synthetic pair")
    a = np.full((240, 300, 3), 200, np.uint8)
    cv2.rectangle(a, (20, 30), (110, 120), (0, 0, 255), -1)
    cv2.circle(a, (200, 80), 50, (0, 200, 0), -1)
    b = a[:, :, ::-1].copy()          # channels swapped -> different colours

def color_hist(image):
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    h = cv2.calcHist([hsv], [0, 1], None, [30, 32], [0, 180, 0, 256])
    return cv2.normalize(h, h).flatten()

ha, hb = color_hist(a), color_hist(b)
methods = {"Correlation (1 = identical)": cv2.HISTCMP_CORREL,
           "Chi-Square (0 = identical)": cv2.HISTCMP_CHISQR,
           "Intersection (higher = more similar)": cv2.HISTCMP_INTERSECT,
           "Bhattacharyya (0 = identical)": cv2.HISTCMP_BHATTACHARYYA}
for name, m in methods.items():
    print(f"{name:40s}: {cv2.compareHist(ha, hb, m):.4f}")

plt.figure(figsize=(8, 3))
plt.subplot(1, 2, 1); plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB)); plt.axis("off"); plt.title("A")
plt.subplot(1, 2, 2); plt.imshow(cv2.cvtColor(b, cv2.COLOR_BGR2RGB)); plt.axis("off"); plt.title("B")
plt.show()
```

#### 2.4.12 Texture Energy and Contrast

**What it does:** For each pixel, energy = sum of squared pixels in a window, contrast = standard deviation in the window; histograms of both are the features. (Manual overflow bug fixed.)

**When to use:** Smooth vs rough texture, material classification.

**What to change:** NEIGHBORHOOD (window size).

**Exam wording:** 'compute texture energy and contrast histograms'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
NEIGHBORHOOD = 3   # CHANGE THIS: window size (3, 5, 7)

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

gray = img.astype(np.float32)                         # float, otherwise squaring overflows uint8
k = (NEIGHBORHOOD, NEIGHBORHOOD)
energy = cv2.boxFilter(gray ** 2, -1, k, normalize=False)             # sum of squares in window
mean = cv2.blur(gray, k)
mean_sq = cv2.blur(gray ** 2, k)
contrast = np.sqrt(np.maximum(mean_sq - mean ** 2, 0))                # standard deviation in window

e_hist, e_bins = np.histogram(energy, bins=256, range=(0, energy.max()))
c_hist, c_bins = np.histogram(contrast, bins=256, range=(0, max(contrast.max(), 1e-6)))
e_hist = e_hist / e_hist.sum()
c_hist = c_hist / c_hist.sum()
print("Mean energy:", energy.mean(), "| mean contrast:", contrast.mean())

plt.figure(figsize=(14, 8))
plt.subplot(2, 2, 1); plt.imshow(energy, cmap="gray"); plt.title("Energy map"); plt.axis("off")
plt.subplot(2, 2, 2); plt.imshow(contrast, cmap="gray"); plt.title("Contrast map"); plt.axis("off")
plt.subplot(2, 2, 3); plt.plot(e_bins[:-1], e_hist, "b"); plt.title("Energy histogram")
plt.subplot(2, 2, 4); plt.plot(c_bins[:-1], c_hist, "r"); plt.title("Contrast histogram")
plt.tight_layout(); plt.show()
```

#### 2.4.13 Material Classification Pipeline

**What it does:** Lab 4 final task: extract texture features (LBP + contrast + gradient statistics) from images of wood / metal / fabric and train an SVM. Uses synthetic textures if no dataset folder exists.

**When to use:** 'classify materials/textures', 'build a texture classifier'.

**What to change:** DATASET_DIR (folder with one sub-folder per class), IMG_SIZE.

**Exam wording:** 'extract texture features and classify', 'train a classifier on features'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
DATASET_DIR = "materials"   # CHANGE THIS: folder with wood/ metal/ fabric/ sub-folders
IMG_SIZE = (128, 128)

import os, glob
from skimage.feature import local_binary_pattern
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

def extract_features(gray):
    gray = cv2.resize(gray, IMG_SIZE)
    lbp = local_binary_pattern(gray, 8, 1, method="uniform")
    lbp_hist, _ = np.histogram(lbp.ravel(), bins=10, range=(0, 10))
    lbp_hist = lbp_hist / (lbp_hist.sum() + 1e-6)
    g = gray.astype(np.float32)
    mean, mean_sq = cv2.blur(g, (3, 3)), cv2.blur(g ** 2, (3, 3))
    contrast = np.sqrt(np.maximum(mean_sq - mean ** 2, 0))
    mag = cv2.magnitude(cv2.Sobel(g, cv2.CV_32F, 1, 0), cv2.Sobel(g, cv2.CV_32F, 0, 1))
    return np.concatenate([lbp_hist, [contrast.mean(), contrast.std(), mag.mean(), mag.std(), g.mean() / 255, g.std() / 255]])

X, y = [], []
if os.path.isdir(DATASET_DIR):
    for label in sorted(os.listdir(DATASET_DIR)):
        for path in glob.glob(os.path.join(DATASET_DIR, label, "*")):
            im = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
            if im is not None:
                X.append(extract_features(im)); y.append(label)
else:
    print(f"'{DATASET_DIR}' not found -> synthetic textures (smooth / noisy / stripes)")
    rng = np.random.default_rng(0)
    xs = np.arange(128)
    for _ in range(30):
        smooth = np.tile(np.linspace(80, 160, 128), (128, 1)) + rng.normal(0, 2, (128, 128))
        noisy = rng.normal(128, 45, (128, 128))
        stripes = 128 + 60 * np.sin(2 * np.pi * xs / 8)[None, :] + rng.normal(0, 5, (128, 128))
        for name, tex in [("smooth", smooth), ("noisy", noisy), ("stripes", stripes)]:
            X.append(extract_features(np.clip(tex, 0, 255).astype(np.uint8))); y.append(name)

X, y = np.array(X), np.array(y)
print("Samples:", X.shape[0], "| features per sample:", X.shape[1], "| classes:", sorted(set(y)))
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
model = make_pipeline(StandardScaler(), SVC(kernel="rbf"))
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred))

# Predict ONE new image:
# new = cv2.imread("test.jpg", cv2.IMREAD_GRAYSCALE)
# print(model.predict([extract_features(new)])[0])
```

## 3 Lab 5 Image Segmentation

Segmentation splits an image into regions (objects / areas). Every method answers: which pixels belong together?

### 3.1 Segmentation Overview

| Technique | Groups pixels by | Best for |
|---|---|---|
| Global threshold | brightness above / below one cutoff | even lighting, clear contrast |
| Otsu | automatically chosen cutoff (two-hump histogram) | same, but unknown cutoff |
| Adaptive threshold | local cutoff per neighbourhood | uneven lighting, text |
| Color threshold (HSV) | colour range | find objects of a given colour |
| Edges / contours | boundaries | outlines, shapes |
| Region growing | similar neighbours from a seed | one uniform region |
| Watershed | flooding from markers | touching / overlapping objects |
| K-Means | colour clusters | few dominant colours |
| GrabCut | rectangle + graph cut | foreground / background removal |
| Deep learning | trained network | complex real scenes |

### 3.5 Region Growing

#### 3.5.1 Region Growing Manual

**What it does:** Starts at a seed pixel and adds neighbours whose brightness is within THRESHOLD of the seed (queue based, 4 or 8 neighbours).

**When to use:** One uniform region and you know a point inside it.

**What to change:** SEED = (row, column) NOT (x, y); THRESHOLD (bigger = bigger region); CONNECTIVITY.

**Exam wording:** 'segment the region containing pixel (r, c) using region growing'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
SEED = (100, 100)   # CHANGE THIS: (row, column)
THRESHOLD = 50       # CHANGE THIS: max brightness difference from the seed
CONNECTIVITY = 4     # 4 or 8


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.full((200, 200), 30, np.uint8)
    cv2.circle(img, (100, 100), 50, 200, -1)
    img = np.clip(img + np.random.randint(-10, 10, img.shape), 0, 255).astype(np.uint8)

from collections import deque

def region_growing(image, seed, threshold, connectivity=4):
    h, w = image.shape
    mask = np.zeros((h, w), np.uint8)
    seed_val = int(image[seed])                      # brightness of the seed (int avoids uint8 wrap-around)
    queue = deque([seed]); mask[seed] = 255
    if connectivity == 4:
        nbrs = [(1, 0), (-1, 0), (0, 1), (0, -1)]
    else:
        nbrs = [(a, b) for a in (-1, 0, 1) for b in (-1, 0, 1) if (a, b) != (0, 0)]
    while queue:
        r, c = queue.popleft()
        for dr, dc in nbrs:
            rr, cc = r + dr, c + dc
            if 0 <= rr < h and 0 <= cc < w and mask[rr, cc] == 0:
                if abs(int(image[rr, cc]) - seed_val) < threshold:
                    mask[rr, cc] = 255
                    queue.append((rr, cc))
    return mask

mask = region_growing(img, SEED, THRESHOLD, CONNECTIVITY)
overlay = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
overlay[mask > 0] = (0, 0, 255)
cv2.circle(overlay, (SEED[1], SEED[0]), 4, (0, 255, 0), -1)      # seed shown in green (x=col, y=row)
print("Region size (pixels):", int((mask > 0).sum()))
show([img, mask, overlay], ["Original", "Grown region", "Overlay (green = seed)"])
```

### 3.6 Watershed

#### 3.6.1 Watershed Full Pipeline and Counting

**What it does:** Separates TOUCHING objects: threshold, remove noise, sure background (dilate), sure foreground (distance transform), unknown = difference, label markers, flood. Counts objects. Markers: 0 unknown, 1 background, 2+ objects, -1 boundary.

**When to use:** Touching / overlapping objects, coin counting.

**What to change:** OBJECTS_ARE_DARK, FG_RATIO (raise to 0.7 to split objects that touch a lot, lower to 0.2 if objects get lost), KSIZE.

**Exam wording:** 'separate touching coins', 'count the objects', 'apply watershed'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
OBJECTS_ARE_DARK = True   # CHANGE THIS: False if objects are brighter than background
FG_RATIO = 0.7      # CHANGE THIS: 0.2 - 0.7
KSIZE = 3
OPEN_ITER = 2


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH)   # color image, BGR order
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.full((220, 330, 3), 235, np.uint8)
    for cx, cy, r in [(70, 70, 45), (140, 80, 45), (230, 80, 45), (110, 160, 40), (200, 165, 40), (280, 170, 35)]:
        cv2.circle(img, (cx, cy), r, (70, 70, 70), -1)         # dark touching coins

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
kind = cv2.THRESH_BINARY_INV if OBJECTS_ARE_DARK else cv2.THRESH_BINARY
_, thresh = cv2.threshold(gray, 0, 255, kind + cv2.THRESH_OTSU)           # objects become white

kernel = np.ones((KSIZE, KSIZE), np.uint8)
opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=OPEN_ITER)   # remove specks
sure_bg = cv2.dilate(opening, kernel, iterations=3)                       # certainly background outside this
dist = cv2.distanceTransform(opening, cv2.DIST_L2, 5)                     # distance to nearest black pixel
_, sure_fg = cv2.threshold(dist, FG_RATIO * dist.max(), 255, 0)           # object cores
sure_fg = np.uint8(sure_fg)
unknown = cv2.subtract(sure_bg, sure_fg)                                  # zone to be decided

_, markers = cv2.connectedComponents(sure_fg)                             # label each core 1, 2, 3 ...
markers = markers + 1                                                     # background = 1
markers[unknown == 255] = 0                                               # 0 = unknown

result = img.copy()
markers = cv2.watershed(result, markers)                                  # needs 8-bit 3-channel image
result[markers == -1] = (0, 0, 255)                                       # boundaries in RED (BGR)

labels = [l for l in np.unique(markers) if l > 1]
print("Objects counted:", len(labels))
colored = np.zeros_like(img)
rng = np.random.default_rng(1)
for l in labels:
    colored[markers == l] = rng.integers(60, 255, 3)

show([thresh, sure_fg, unknown, result, colored], ["1 Threshold", "2 Sure foreground", "3 Unknown zone", f"4 Watershed ({len(labels)} objects)", "5 Labelled regions"], cols=3)
```

### 3.9 Morphological Operations

Clean binary masks: erode shrinks white, dilate grows white, opening = erode then dilate (removes specks), closing = dilate then erode (fills holes).

#### 3.9.1 Erode Dilate Open Close

**What it does:** Applies every morphological operation to a binary mask.

**When to use:** Clean noise / fill holes in a mask before counting or contouring.

**What to change:** KSIZE, SHAPE, ITER.

**Exam wording:** 'remove small noise from the mask', 'fill holes'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
KSIZE = 5
SHAPE = "rect"   # CHANGE THIS: "rect", "ellipse" or "cross"
ITER = 1


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.tile(np.linspace(60, 140, 300), (240, 1)).astype(np.uint8)
    cv2.rectangle(img, (30, 40), (130, 140), 220, -1)
    cv2.circle(img, (210, 100), 55, 20, -1)
    cv2.putText(img, "TEST", (80, 215), cv2.FONT_HERSHEY_SIMPLEX, 1.5, 255, 3)
    img = np.clip(img + np.random.normal(0, 6, img.shape), 0, 255).astype(np.uint8)

_, mask = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
shape = {"rect": cv2.MORPH_RECT, "ellipse": cv2.MORPH_ELLIPSE, "cross": cv2.MORPH_CROSS}[SHAPE]
kernel = cv2.getStructuringElement(shape, (KSIZE, KSIZE))
ops = {"Erode": cv2.erode(mask, kernel, iterations=ITER),
       "Dilate": cv2.dilate(mask, kernel, iterations=ITER),
       "Open (remove specks)": cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel, iterations=ITER),
       "Close (fill holes)": cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel, iterations=ITER),
       "Gradient (outline)": cv2.morphologyEx(mask, cv2.MORPH_GRADIENT, kernel),
       "Top-hat": cv2.morphologyEx(mask, cv2.MORPH_TOPHAT, kernel),
       "Black-hat": cv2.morphologyEx(mask, cv2.MORPH_BLACKHAT, kernel)}
show([mask] + list(ops.values()), ["Binary mask"] + list(ops.keys()), cols=4)
```

### 4.1 Counting and Measuring Objects

#### 4.1.1 Count Objects with Connected Components

**What it does:** Blur, Otsu, clean, then connectedComponentsWithStats gives count, area and box of each object.

**When to use:** 'count the objects / cells / coins' when objects do NOT touch (if they touch use watershed).

**What to change:** INVERT, MIN_AREA, KSIZE.

**Exam wording:** 'count the number of objects in the image'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
INVERT = True    # CHANGE THIS
MIN_AREA = 200
KSIZE = 3


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH)   # color image, BGR order
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.full((220, 330, 3), 235, np.uint8)
    for cx, cy, r in [(70, 70, 45), (140, 80, 45), (230, 80, 45), (110, 160, 40), (200, 165, 40), (280, 170, 35)]:
        cv2.circle(img, (cx, cy), r, (70, 70, 70), -1)         # dark touching coins

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 0)
kind = cv2.THRESH_BINARY_INV if INVERT else cv2.THRESH_BINARY
_, mask = cv2.threshold(blur, 0, 255, kind + cv2.THRESH_OTSU)
k = np.ones((KSIZE, KSIZE), np.uint8)
mask = cv2.morphologyEx(cv2.morphologyEx(mask, cv2.MORPH_OPEN, k), cv2.MORPH_CLOSE, k)

n, labels, stats, centroids = cv2.connectedComponentsWithStats(mask, connectivity=8)
out = img.copy()
count = 0
for i in range(1, n):                                    # label 0 is the background
    x, y, w, h, area = stats[i]
    if area < MIN_AREA:
        continue
    count += 1
    cv2.rectangle(out, (x, y), (x + w, y + h), (0, 255, 0), 2)
    cv2.putText(out, str(count), (x, y - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)
    print(f"Object {count}: area={area}, box=({x},{y},{w},{h}), centre=({centroids[i][0]:.0f},{centroids[i][1]:.0f})")
print("Total objects:", count)
show([img, mask, out], ["Original", "Mask", f"{count} objects"])
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# USER INPUT / VALUES TO CHANGE
# =====================================================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to your input image
COLOR_NAME = "green"   # CHANGE THIS
BG_MODE = "white"      # CHANGE THIS: "white", "blur", "gray" or path to a background image


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

img = cv2.imread(IMAGE_PATH)   # color image, BGR order
if img is None:   # file missing -> synthetic test image so the cell still runs
    print(f"'{IMAGE_PATH}' not found -> using a synthetic test image")
    img = np.full((240, 300, 3), 200, np.uint8)
    cv2.rectangle(img, (20, 30), (110, 120), (0, 0, 255), -1)    # red
    cv2.circle(img, (200, 80), 50, (0, 200, 0), -1)              # green
    cv2.rectangle(img, (60, 150), (170, 220), (255, 0, 0), -1)   # blue
    cv2.circle(img, (250, 190), 35, (0, 255, 255), -1)           # yellow

HSV_RANGES = {   # OpenCV HSV: H 0-179, S 0-255, V 0-255 (red needs TWO ranges)
    "red":    [((0, 120, 70), (10, 255, 255)), ((170, 120, 70), (179, 255, 255))],
    "orange": [((10, 100, 100), (20, 255, 255))],
    "yellow": [((20, 100, 100), (35, 255, 255))],
    "green":  [((36, 50, 50), (85, 255, 255))],
    "cyan":   [((85, 50, 50), (100, 255, 255))],
    "blue":   [((100, 100, 50), (130, 255, 255))],
    "purple": [((130, 50, 50), (160, 255, 255))],
    "white":  [((0, 0, 200), (179, 40, 255))],
    "black":  [((0, 0, 0), (179, 255, 50))],
}

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
mask = np.zeros(img.shape[:2], np.uint8)
for lo, hi in HSV_RANGES[COLOR_NAME]:
    mask |= cv2.inRange(hsv, np.array(lo), np.array(hi))
k = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
mask = cv2.morphologyEx(cv2.morphologyEx(mask, cv2.MORPH_OPEN, k), cv2.MORPH_CLOSE, k)   # clean mask

if BG_MODE == "white":
    bg = np.full_like(img, 255)
elif BG_MODE == "blur":
    bg = cv2.GaussianBlur(img, (51, 51), 0)
elif BG_MODE == "gray":
    bg = cv2.cvtColor(cv2.cvtColor(img, cv2.COLOR_BGR2GRAY), cv2.COLOR_GRAY2BGR)
else:                                                                  # path to another image
    bg = cv2.imread(BG_MODE)
    bg = cv2.resize(bg, (img.shape[1], img.shape[0])) if bg is not None else np.zeros_like(img)

mask3 = cv2.merge([mask, mask, mask])
result = np.where(mask3 > 0, img, bg)                                  # object from img, rest from bg
rgba = cv2.merge([*cv2.split(img), mask])                              # transparent background version
cv2.imwrite("object_transparent.png", rgba)
show([img, mask, result], ["Original", "Mask", f"Background = {BG_MODE}"])
```

### 4.4 Comparing and Analysing

#### 4.4.1 Compare Two Images with HOG LBP and Color

**What it does:** Computes HOG, LBP and colour features for two images and prints distances (smaller = more similar).

**When to use:** 'which image is more similar', 'match an image against a template'.

**What to change:** IMAGE_A, IMAGE_B.

**Exam wording:** 'compare two images using extracted features'

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import local_binary_pattern, hog


def show(images, titles, cols=None, figsize=(14, 4), cmap="gray"):
    """Display images side by side. Color images must be BGR (OpenCV order)."""
    n = len(images)
    cols = cols or n
    rows = (n + cols - 1) // cols
    plt.figure(figsize=(figsize[0], figsize[1] * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 3:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        else:
            plt.imshow(im, cmap=cmap)
        plt.title(t)
        plt.axis("off")
    plt.tight_layout()
    plt.show()

IMAGE_A = "image1.jpg"   # CHANGE THIS
IMAGE_B = "image2.jpg"   # CHANGE THIS
a, b = cv2.imread(IMAGE_A), cv2.imread(IMAGE_B)
if a is None or b is None:
    print("Images not found -> synthetic pair")
    a = np.full((200, 200, 3), 200, np.uint8); cv2.circle(a, (100, 100), 60, (0, 0, 255), -1)
    b = cv2.GaussianBlur(a, (9, 9), 0)

def features(image):
    image = cv2.resize(image, (128, 128))
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    f_hog = hog(gray, orientations=9, pixels_per_cell=(16, 16), cells_per_block=(2, 2))
    lbp = local_binary_pattern(gray, 8, 1, method="uniform")
    f_lbp = np.histogram(lbp.ravel(), bins=10, range=(0, 10))[0] / lbp.size
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    f_col = cv2.normalize(cv2.calcHist([hsv], [0, 1, 2], None, [8, 8, 8], [0, 180, 0, 256, 0, 256]), None).flatten()
    return {"HOG": f_hog, "LBP": f_lbp, "Color": f_col}

fa, fb = features(a), features(b)
for name in fa:
    euclid = np.linalg.norm(fa[name] - fb[name])
    cosine = 1 - np.dot(fa[name], fb[name]) / (np.linalg.norm(fa[name]) * np.linalg.norm(fb[name]) + 1e-9)
    print(f"{name:6s} Euclidean distance = {euclid:.4f} | cosine distance = {cosine:.4f}")
show([a, b], ["Image A", "Image B"])
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

IMAGE_PATH = "image.jpg"   # CHANGE THIS
img = cv2.imread(IMAGE_PATH)
if img is None:
    print("Image not found -> synthetic")
    img = np.full((220, 330, 3), 235, np.uint8)
    cv2.circle(img, (100, 100), 50, (70, 70, 70), -1); cv2.rectangle(img, (190, 60), (290, 160), (90, 90, 90), -1)

def preprocess(im):       # CHANGE: grayscale / blur / contrast ...
    return cv2.GaussianBlur(cv2.cvtColor(im, cv2.COLOR_BGR2GRAY), (5, 5), 0)

def segment(gray):        # CHANGE: threshold / Canny / kmeans / hsv ...
    return cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)[1]

def clean(mask):          # CHANGE: morphology
    return cv2.morphologyEx(mask, cv2.MORPH_OPEN, np.ones((3, 3), np.uint8))

def analyse(mask, im):    # CHANGE: count / measure / features / draw
    cnts = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)[-2]
    out = im.copy(); cv2.drawContours(out, cnts, -1, (0, 255, 0), 2)
    print("Objects:", len(cnts))
    return out

gray = preprocess(img); mask = clean(segment(gray)); result = analyse(mask, img)
plt.figure(figsize=(12, 4))
for i, (im, t) in enumerate([(img, "Original"), (mask, "Mask"), (result, "Result")]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB) if im.ndim == 3 else im, cmap="gray"); plt.title(t); plt.axis("off")
plt.tight_layout(); plt.show()
cv2.imwrite("result.png", result)
```

### 5.2 Manual Code Bugs and Fixes

| Where | Problem | Fix |
|---|---|---|
| Valid convolution / strides | `filter2D` always returns same size; `strides=` does not exist | crop border for valid, slice `[::s, ::s]` for stride |
| Same convolution | example used BORDER_REFLECT, not zeros | use `cv2.BORDER_CONSTANT` |
| Gaussian kernel | `getGaussianKernel` is 1-D | `k @ k.T` or `cv2.GaussianBlur` |
| HED / HIG | `arctan2` gives negative angles, dropped by histogram range | add `% 360` or `% 180` |
| HED | edge map never used | `angle[edges > 0]` |
| Texture energy | squaring uint8 overflows | convert to float32 first |
| Texture contrast | box sum, not standard deviation | sqrt(E[x^2] - E[x]^2) |
| HOG visualisation | `in_range=(0,10)` makes picture dark | `(0, hog_image.max())` |
| Adaptive threshold | block 11 / C 2 only suits small text | larger odd block for big objects |
| Region growing | seed (10,10) is a corner; imshow without gray | choose a seed inside the object; `cmap="gray"` |
| Watershed | `[255,0,0]` is BLUE in BGR; 0.2 merges objects | red is `(0,0,255)`; FG_RATIO 0.5-0.7 |
| K-Means / watershed display | plt.imshow on BGR swaps colours | `cv2.cvtColor(img, cv2.COLOR_BGR2RGB)` |

### 5.3 MCQ Facts

- Kernel sum 1 = blur; sum 0 = edge detector. `dx` finds vertical edges, `dy` horizontal.
- Valid output N-K+1, same output N, stride S: floor((N-K)/S)+1.
- Sobel needs `cv2.CV_64F` to keep negative values.
- Canny order: smooth, gradient, non-maximum suppression, hysteresis. Above HIGH = edge, below LOW = not, between = only if connected.
- HOG: 9 bins over 0-180, 8x8 cells, 2x2 blocks; 128x64 image gives 3780 values. HOG keeps position, HIG / HED do not.
- HED: 8 bins over 0-360 on edge pixels. HIG: 9 bins over 0-180 on all pixels.
- LBP uniform with 8 points gives 10 bins (n_points + 2). Not affected by brightness changes.
- Texture energy = sum of squares in window; contrast = standard deviation in window.
- OpenCV loads BGR; matplotlib wants RGB. HSV hue is 0-179.
- `cv2.threshold` returns two values. Otsu: pass 0, best on bimodal histogram.
- Adaptive threshold needs odd block size and 8-bit single channel image.
- Watershed needs 8-bit 3-channel image; markers 0 unknown, 1 background, 2+ objects, -1 boundary.
- K-Means: choose K yourself, float32 input, steps are assign then update until convergence. Mean shift / DBSCAN find the number of groups automatically.
- Opening = erosion then dilation (removes specks). Closing = dilation then erosion (fills holes).
- Semantic segmentation labels classes; instance segmentation labels each object.

### 5.4 Troubleshooting Errors

| Symptom | Cause / fix |
|---|---|
| `AttributeError: NoneType has no attribute shape` or `(-215) !_src.empty()` | wrong file path, `imread` returned None |
| Picture colours look wrong (blue skin) | BGR shown as RGB: use `cv2.cvtColor(img, cv2.COLOR_BGR2RGB)` |

| adaptiveThreshold error `src.type() == CV_8UC1` | image is colour or float: convert to 8-bit gray |
| watershed error about CV_8UC3 | pass the 3-channel colour image, not gray |

| `ModuleNotFoundError: skimage / sklearn` | `pip install scikit-image scikit-learn` |
| `cv2.resize` gives wrong shape | resize takes (width, height); array shape is (height, width) |
