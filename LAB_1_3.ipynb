#### 1.2.1 Display with Matplotlib (manual 11.1)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
TITLE = "My Image"          # CHANGE THIS: title above the image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

# Convert BGR -> RGB (OpenCV loads BGR, Matplotlib expects RGB)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title(TITLE, fontsize=16, color="darkred", pad=15)
plt.axis("off")
plt.show()
```

#### 1.3.2 Split and merge channels, view each channel

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

b, g, r = cv2.split(img)            # each is a 2D array
merged = cv2.merge([b, g, r])        # back to a 3-channel image
print("Merged identical to original:", np.array_equal(merged, img))

# Tinted versions: keep one channel, zero the others
blue_only = img.copy();  blue_only[:, :, 1] = 0;  blue_only[:, :, 2] = 0
green_only = img.copy(); green_only[:, :, 0] = 0; green_only[:, :, 2] = 0
red_only = img.copy();   red_only[:, :, 0] = 0;   red_only[:, :, 1] = 0

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (b, "Blue channel (gray)"), (g, "Green channel (gray)"), (r, "Red channel (gray)"), (blue_only, "Blue only"), (red_only, "Red only")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.4.3 Manual weighted formula (0.299R + 0.587G + 0.114B) and simple average

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

b, g, r = [c.astype(np.float32) for c in cv2.split(img)]

gray_weighted = (0.299 * r + 0.587 * g + 0.114 * b).astype(np.uint8)   # same as OpenCV
gray_average = ((r + g + b) / 3).astype(np.uint8)                       # simple mean
gray_cv = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("Max difference weighted vs cv2:", np.abs(gray_weighted.astype(int) - gray_cv.astype(int)).max())

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(gray_cv, "cv2 grayscale"), (gray_weighted, "Weighted formula"), (gray_average, "Simple average")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.6.1 Scale factor (manual 11.4)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
SCALE_FACTOR = 0.5   # CHANGE THIS: 0.5 = half size, 2 = double
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

resized = cv2.resize(img, None, fx=SCALE_FACTOR, fy=SCALE_FACTOR)   # interpolation=cv2.INTER_LINEAR default

new_h, new_w = resized.shape[:2]
print("Original Resolution: %d x %d" % (width, height))
print("New Resolution     : %d x %d" % (new_w, new_h))

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, f"Original ({width} x {height})"), (resized, f"Resized ({new_w} x {new_h})")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.6.3 Keep aspect ratio (give only target width OR height)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
TARGET_WIDTH = 400   # CHANGE THIS (or set TARGET_HEIGHT instead)
TARGET_HEIGHT = None
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

if TARGET_WIDTH is not None:
    ratio = TARGET_WIDTH / width
    size = (TARGET_WIDTH, int(height * ratio))
else:
    ratio = TARGET_HEIGHT / height
    size = (int(width * ratio), TARGET_HEIGHT)

resized = cv2.resize(img, size, interpolation=cv2.INTER_AREA)
print("New size (W, H):", size)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (resized, f"Aspect ratio kept {size}")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.6.4 Compare interpolation methods

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
SCALE_FACTOR = 3   # CHANGE THIS: >1 enlarge, <1 shrink
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

methods = {"NEAREST": cv2.INTER_NEAREST, "LINEAR": cv2.INTER_LINEAR,
           "CUBIC": cv2.INTER_CUBIC, "AREA": cv2.INTER_AREA}      # AREA best for shrinking, CUBIC/LINEAR for enlarging
results = {n: cv2.resize(img, None, fx=SCALE_FACTOR, fy=SCALE_FACTOR, interpolation=m) for n, m in methods.items()}

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(results['NEAREST'], "NEAREST"), (results['LINEAR'], "LINEAR"), (results['CUBIC'], "CUBIC"), (results['AREA'], "AREA")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.7.1 Gaussian blur (manual 11.5)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
KSIZE = 31   # CHANGE THIS: kernel size, must be ODD (3, 5, 31...)
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

blurred = cv2.GaussianBlur(img, (KSIZE, KSIZE), 0)   # 0 = sigma computed automatically from kernel size

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (blurred, f"Gaussian Blur ({KSIZE}x{KSIZE})")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.7.2 Compare blur types: average, Gaussian, median, bilateral

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
KSIZE = 7   # CHANGE THIS: odd number
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

avg = cv2.blur(img, (KSIZE, KSIZE))                  # box / mean filter
gauss = cv2.GaussianBlur(img, (KSIZE, KSIZE), 0)      # weighted mean
median = cv2.medianBlur(img, KSIZE)                   # best for salt-and-pepper noise
bilateral = cv2.bilateralFilter(img, 9, 75, 75)       # smooths but keeps edges (d, sigmaColor, sigmaSpace)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (avg, "Average"), (gauss, "Gaussian"), (median, "Median"), (bilateral, "Bilateral")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.7.3 Custom kernel with cv2.filter2D (sharpen, edge, emboss)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

sharpen_kernel = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]], dtype=np.float32)
edge_kernel = np.array([[-1, -1, -1],
                        [-1, 8, -1],
                        [-1, -1, -1]], dtype=np.float32)
box_kernel = np.ones((5, 5), np.float32) / 25      # same as cv2.blur 5x5

sharpened = cv2.filter2D(img, -1, sharpen_kernel)   # -1 = same depth as input
edges = cv2.filter2D(img, -1, edge_kernel)
boxed = cv2.filter2D(img, -1, box_kernel)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (sharpened, "Sharpened"), (edges, "Edge kernel"), (boxed, "Box 5x5")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.7.4 Unsharp masking (blur-based sharpening)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
AMOUNT = 1.5   # CHANGE THIS: sharpening strength
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

blur = cv2.GaussianBlur(img, (0, 0), 3)
sharp = cv2.addWeighted(img, 1 + AMOUNT, blur, -AMOUNT, 0)   # img*(1+a) - blur*a

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (sharp, "Unsharp mask")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 1.13.4 Logo/overlay on an image using a mask (bitwise operations)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
IMAGE2_PATH = "image2.jpg"   # CHANGE THIS: logo / overlay image
POS_X, POS_Y = 10, 10        # CHANGE THIS: top-left corner where logo is placed
LOGO_WIDTH = 150             # CHANGE THIS
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

# Load second image
img2 = cv2.imread(IMAGE2_PATH)
if img2 is None:
    raise FileNotFoundError("Could not read " + IMAGE2_PATH)

ratio = LOGO_WIDTH / img2.shape[1]
logo = cv2.resize(img2, (LOGO_WIDTH, int(img2.shape[0] * ratio)))
lh, lw = logo.shape[:2]

roi = img[POS_Y:POS_Y + lh, POS_X:POS_X + lw]
logo_gray = cv2.cvtColor(logo, cv2.COLOR_BGR2GRAY)
_, mask = cv2.threshold(logo_gray, 10, 255, cv2.THRESH_BINARY)       # logo pixels = 255
mask_inv = cv2.bitwise_not(mask)

bg = cv2.bitwise_and(roi, roi, mask=mask_inv)       # keep background where logo is NOT
fg = cv2.bitwise_and(logo, logo, mask=mask)         # keep logo pixels
out = img.copy()
out[POS_Y:POS_Y + lh, POS_X:POS_X + lw] = cv2.add(bg, fg)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (out, "Logo overlaid")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

**Digital image**: x, y and intensity are finite & discrete. Origin (0,0) = top-left; `img[row, col]` = `img[y, x]`; shape = (H, W, C).

| Term | Meaning |
|---|---|
| Feature / Feature extraction | distinctive part / isolating relevant features (filters, algorithms) |
| Edge detection | finding boundaries between regions |
| Object detection / recognition / classification | locate objects / label objects / assign a class to the whole image |
| Segmentation | splitting image into regions (object vs background) |
| CNN / Deep learning / Feature map | grid-data network / multi-layer NN / output of a convolution filter |
| Image registration | aligning images of the same scene to a common coordinate system |
| Optical flow / Tracking | motion between frames / following objects across frames |
| Homography | transform relating two images of the same planar surface |
| Histogram | distribution of pixel intensities |

## 2 LAB 2 - Photometric / Point Transformations (Log, Gamma, Piecewise, Histogram)

Lab 2 manual was unavailable, so this section is built from the syllabus topics: **Log, Gamma, Piecewise, Histogram** (+ negative, contrast stretching, slicing, equalization, CLAHE, matching).

**Point transformation:** `s = T(r)` - the new pixel value depends only on the old value at the same pixel (r = input, s = output). Photometric = changes *what the pixel looks like*, geometric (Lab 3) = changes *where* it is.

| Transform | Formula (L = 256 levels) | Effect |
|---|---|---|
| Negative | s = 255 - r | inverts intensities |
| Log | s = c * log(1 + r), c = 255 / log(1 + max) | expands dark values, compresses bright (brightens dark areas) |
| Inverse log | s = exp(r / c) - 1 | expands bright, compresses dark |
| Gamma (power-law) | s = 255 * (r/255)^gamma | gamma < 1 brightens, gamma > 1 darkens, gamma = 1 identity |
| Contrast stretching | s = (r - rmin) * 255 / (rmax - rmin) | uses the full 0-255 range |
| Piecewise linear | different slope per segment between (r1,s1), (r2,s2) | custom contrast / slicing |
| Histogram equalization | s = round((CDF(r) - CDFmin) / (N - CDFmin) * 255) | flatter histogram, higher global contrast |

Every cell below has `GRAYSCALE = True/False`: True processes a gray image, False processes the color image (applied to each channel).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

f = np.fft.fftshift(np.fft.fft2(img.astype(np.float32)))
mag = np.abs(f)

linear = (mag / mag.max() * 255).astype(np.uint8)                  # almost entirely black
c = 255 / np.log(1 + mag.max())
logged = np.clip(c * np.log(1 + mag), 0, 255).astype(np.uint8)    # details become visible

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Image"), (linear, "Spectrum (linear scale)"), (logged, "Spectrum after log transform")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 2.9.2 Manual equalization from scratch (hist -> CDF -> LUT)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

hist = np.bincount(img.ravel(), minlength=256)        # 1) histogram
cdf = hist.cumsum()                                    # 2) cumulative distribution
cdf_min = cdf[cdf > 0].min()                           # first non-zero CDF value
N = img.size
lut = np.round((cdf - cdf_min) / (N - cdf_min) * 255).clip(0, 255).astype(np.uint8)   # 3) mapping s = T(r)
equalized = lut[img]                                   # 4) apply mapping

builtin = cv2.equalizeHist(img)
print("Max difference vs cv2.equalizeHist:", np.abs(equalized.astype(int) - builtin.astype(int)).max())

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (equalized, "Manual equalization"), (builtin, "cv2.equalizeHist")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 2.9.6 Histogram matching / specification (match to a reference image)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
REF_PATH = "image2.jpg"   # CHANGE THIS: reference image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

ref = cv2.imread(REF_PATH, cv2.IMREAD_GRAYSCALE)
if ref is None:
    raise FileNotFoundError(REF_PATH)

def match_histogram(src, ref):
    s_vals, s_idx, s_counts = np.unique(src.ravel(), return_inverse=True, return_counts=True)
    r_vals, r_counts = np.unique(ref.ravel(), return_counts=True)
    s_cdf = np.cumsum(s_counts) / src.size
    r_cdf = np.cumsum(r_counts) / ref.size
    mapped = np.interp(s_cdf, r_cdf, r_vals)            # for each source level find reference level with same CDF
    return mapped[s_idx].reshape(src.shape).astype(np.uint8)

out = match_histogram(img, ref)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Source"), (ref, "Reference"), (out, "Matched to reference")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 3 LAB 3 - Geometric Transformations

**Geometric transformation** = changes *where* pixels are (x, y) -> (x', y'), not what they look like. Photometric (Lab 2) = changes pixel values.

**Image coordinates:** origin (0,0) = top-left, x -> right, y -> down. `img[y, x]`, point = (x, y). OpenCV `dsize`/`center`/points are **(x, y) = (width, height)**.

**Hierarchy:** Affine = Linear (rotation, scale, shear, reflection) + Translation. Rigid = rotation + translation. Similarity = rotation + uniform scale + translation. Projective (homography/perspective) is NOT affine.

| Type | Matrix | OpenCV | Preserves |
|---|---|---|---|
| Linear | 2x2 `[[a,b],[c,d]]` (origin fixed, A*0 = 0) | - | origin, lines, parallelism |
| Affine | 2x3 `[A | t]`, p' = Ap + b | `warpAffine` | collinearity, parallelism, ratios along a line |
| Projective | 3x3 H, x' = (h11x+h12y+h13)/(h31x+h32y+h33) | `warpPerspective` | only straight lines (parallel lines can converge) |

**2x2 matrix columns:** `[[a,b],[c,d]]` -> first column (a,c) is where the x-axis lands, second column (b,d) is where the y-axis lands. a = horizontal scale, d = vertical scale, b = horizontal shear, c = vertical shear.
**Homogeneous coords:** (x, y) -> (x, y, 1) so translation becomes a matrix multiply; transforms combine by multiplying matrices (order matters: rightmost is applied first). Affine bottom row = [0,0,1] (denominator = 1); projective bottom row varies (denominator creates depth).

**Matrices:**
Translation `[[1,0,tx],[0,1,ty]]` | Scaling `[[sx,0,0],[0,sy,0]]` | Rotation `[[cos,-sin],[sin,cos]]` | Reflection `[[-1,0],[0,1]]` (x) / `[[1,0],[0,-1]]` (y) | Shear X `[[1,k],[0,1]]` (x' = x + ky).

### 3.7 Linear Transformation with ANY 2x2 Matrix (auto-fit canvas)

**What this code does:** Applies a general matrix [[a,b],[c,d]] (+ optional translation) and automatically sizes the canvas so nothing is cut off.

**When to use it:** Question gives a matrix (a, b, c, d) and asks to apply it, or asks what a matrix does.

**What to change:** A (the 2x2 matrix), T (translation, can be 0).

<a id="sec-3-7-1"></a>

#### 3.7.1 Apply a generic 2x2 matrix with automatic output size

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
A = np.float32([[1.0, 0.5],        # CHANGE THIS: [[a, b],
              [0.0, 1.0]])       #               [c, d]]
T_VEC = np.float32([0, 0])       # CHANGE THIS: translation (tx, ty); makes it affine
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

corners = np.float32([[0, 0], [width, 0], [width, height], [0, height]])
new_corners = corners @ A.T + T_VEC                     # transform image corners
min_xy, max_xy = new_corners.min(axis=0), new_corners.max(axis=0)
shift = -min_xy                                         # move so everything is at positive coordinates
M = np.hstack([A, (T_VEC + shift).reshape(2, 1)]).astype(np.float32)       # 2x3 matrix for warpAffine
size = (int(np.ceil(max_xy[0] - min_xy[0])), int(np.ceil(max_xy[1] - min_xy[1])))

out = cv2.warpAffine(img, M, size)
print("Matrix used:\n", M, "\nOutput size (W, H):", size)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (out, "Generic linear/affine transform")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
SX, SY = 0.8, 0.8    # CHANGE THIS
ANGLE = 30           # CHANGE THIS (degrees)
TX, TY = 100, 50     # CHANGE THIS
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

th = np.deg2rad(ANGLE)
S = np.array([[SX, 0, 0], [0, SY, 0], [0, 0, 1]], dtype=np.float32)
R = np.array([[np.cos(th), -np.sin(th), 0], [np.sin(th), np.cos(th), 0], [0, 0, 1]], dtype=np.float32)
Tr = np.array([[1, 0, TX], [0, 1, TY], [0, 0, 1]], dtype=np.float32)

master = Tr @ R @ S              # rightmost is applied FIRST: scale -> rotate -> translate
print("Master matrix:\n", master)
out = cv2.warpAffine(img, master[:2], (width, height))     # warpAffine needs the top 2 rows (2x3)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (out, "Scale -> Rotate -> Translate")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 3.8.2 Rotation about the center = T(c) * R * T(-c)  (and order matters)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
ANGLE = 40   # CHANGE THIS
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

cx, cy = width / 2, height / 2
th = np.deg2rad(ANGLE)
T1 = np.array([[1, 0, cx], [0, 1, cy], [0, 0, 1]], dtype=np.float32)
T2 = np.array([[1, 0, -cx], [0, 1, -cy], [0, 0, 1]], dtype=np.float32)
R = np.array([[np.cos(th), -np.sin(th), 0], [np.sin(th), np.cos(th), 0], [0, 0, 1]], dtype=np.float32)
Tr = np.array([[1, 0, 150], [0, 1, 0], [0, 0, 1]], dtype=np.float32)

about_center = T1 @ R @ T2                                    # move center to origin, rotate, move back
rot_then_trans = Tr @ R                                       # rotate first, then translate
trans_then_rot = R @ Tr                                       # translate first, then rotate (different result!)

a = cv2.warpAffine(img, about_center[:2], (width, height))
b = cv2.warpAffine(img, rot_then_trans[:2], (width, height))
c = cv2.warpAffine(img, trans_then_rot[:2], (width, height))

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(a, "Rotate about center"), (b, "Rotate then translate"), (c, "Translate then rotate")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 3.12.2 Auto-deskew (estimate the tilt of text/objects)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
FLIP_SIGN = 1   # CHANGE THIS to -1 if the result rotates the wrong way
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, bw = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)      # foreground white (dark text on light bg)
coords = np.column_stack(np.where(bw > 0))[:, ::-1].astype(np.float32)            # (row, col) -> (x, y)

angle = cv2.minAreaRect(coords)[-1]
if angle > 45:          # normalize to (-45, 45] (different OpenCV versions report different ranges)
    angle -= 90
elif angle < -45:
    angle += 90
angle *= FLIP_SIGN
print("Estimated skew angle:", angle)

M = cv2.getRotationMatrix2D((width / 2, height / 2), angle, 1.0)
out = cv2.warpAffine(img, M, (width, height), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_REPLICATE)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Original"), (out, f"Deskewed ({angle:.1f} deg)")]
cols = 2
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 3.12.4 Document scanner: automatic corner detection + perspective correction

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
MIN_AREA_FRAC = 0.1   # CHANGE THIS: document must cover at least this fraction of the image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

def order_points(pts):
    pts = pts.reshape(4, 2).astype("float32")
    s = pts.sum(axis=1)
    d = np.diff(pts, axis=1).ravel()
    return np.float32([pts[np.argmin(s)],      # top-left     (smallest x+y)
                       pts[np.argmin(d)],      # top-right    (smallest y-x)
                       pts[np.argmax(s)],      # bottom-right (largest x+y)
                       pts[np.argmax(d)]])     # bottom-left  (largest y-x)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 0)
edges = cv2.Canny(blur, 75, 200)
edges = cv2.dilate(edges, np.ones((3, 3), np.uint8), iterations=1)

contours, _ = cv2.findContours(edges, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
contours = sorted(contours, key=cv2.contourArea, reverse=True)[:10]

doc = None
for c in contours:
    approx = cv2.approxPolyDP(c, 0.02 * cv2.arcLength(c, True), True)
    if len(approx) == 4 and cv2.contourArea(c) > MIN_AREA_FRAC * width * height:
        doc = approx
        break
if doc is None:
    raise RuntimeError("No 4-corner document found - use the manual-corner template instead")

rect = order_points(doc)
tl, tr, br, bl = rect
out_w = int(max(np.linalg.norm(br - bl), np.linalg.norm(tr - tl)))
out_h = int(max(np.linalg.norm(tr - br), np.linalg.norm(tl - bl)))
dst = np.float32([[0, 0], [out_w - 1, 0], [out_w - 1, out_h - 1], [0, out_h - 1]])

M = cv2.getPerspectiveTransform(rect, dst)
scan = cv2.warpPerspective(img, M, (out_w, out_h))

marked = img.copy()
cv2.polylines(marked, [rect.astype(np.int32).reshape(-1, 1, 2)], True, (0, 255, 0), 4)
scan_gray = cv2.cvtColor(scan, cv2.COLOR_BGR2GRAY)
scan_bw = cv2.adaptiveThreshold(scan_gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 21, 10)   # "scanned" look

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(marked, "Detected document"), (scan, "Top-down view"), (scan_bw, "Scanned B/W")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

#### 3.12.10 Image alignment / registration with ORB features + homography (two views of the same scene)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
IMAGE2_PATH = "image2.jpg"   # CHANGE THIS: second view of the same scene (image.jpg = reference)
MAX_FEATURES = 1000
KEEP = 0.2                   # fraction of best matches to keep
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

# Load second image
img2 = cv2.imread(IMAGE2_PATH)
if img2 is None:
    raise FileNotFoundError("Could not read " + IMAGE2_PATH)

g1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
g2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

orb = cv2.ORB_create(MAX_FEATURES)
k1, d1 = orb.detectAndCompute(g1, None)
k2, d2 = orb.detectAndCompute(g2, None)
matches = sorted(cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True).match(d2, d1), key=lambda m: m.distance)
matches = matches[:max(int(len(matches) * KEEP), 4)]

pts2 = np.float32([k2[m.queryIdx].pt for m in matches]).reshape(-1, 1, 2)
pts1 = np.float32([k1[m.trainIdx].pt for m in matches]).reshape(-1, 1, 2)
H, mask = cv2.findHomography(pts2, pts1, cv2.RANSAC, 5.0)       # map image 2 -> image 1
aligned = cv2.warpPerspective(img2, H, (width, height))
print("Homography:\n", H)

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(img, "Reference"), (img2, "Second view"), (aligned, "Image 2 aligned to reference")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# USER INPUT / VALUES TO CHANGE
# ==============================
IMAGE_PATH = "image.jpg"   # CHANGE THIS: path to the input image
# ==============================

# Load image
img = cv2.imread(IMAGE_PATH)
if img is None:
    raise FileNotFoundError("Could not read " + IMAGE_PATH + " - check the path")
height, width = img.shape[:2]

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
rot = cv2.warpAffine(gray, cv2.getRotationMatrix2D((width / 2, height / 2), 30, 1), (width, height))
eq = cv2.equalizeHist(rot)
gam = np.clip(255.0 * (rot / 255.0) ** 0.5, 0, 255).astype(np.uint8)
thr = cv2.threshold(eq, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]

# Display (gray arrays shown as gray, BGR arrays converted to RGB)
def _show(a):
    if a.ndim == 2:
        plt.imshow(a, cmap="gray", vmin=0, vmax=255)
    else:
        plt.imshow(cv2.cvtColor(a, cv2.COLOR_BGR2RGB))

panels = [(gray, "Gray"), (rot, "Rotated"), (eq, "Rotated + equalized"), (gam, "Rotated + gamma 0.5"), (thr, "Rotated + equalized + Otsu")]
cols = 3
rows = (len(panels) + cols - 1) // cols
plt.figure(figsize=(5.5 * cols, 4.5 * rows))
for i, (a, t) in enumerate(panels, 1):
    plt.subplot(rows, cols, i)
    _show(a)
    plt.title(t)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

### 5.2 Pitfalls that cost marks

1. **BGR vs RGB**: `cv2.imread` gives BGR; convert with `cvtColor(..., COLOR_BGR2RGB)` before `plt.imshow`. Drawing colors on a BGR image are BGR: red = `(0, 0, 255)`.
2. **(x, y) vs [row, col]**: points/center/dsize are (x, y) = (width, height); indexing is `img[y, x]`; `shape` = (height, width, channels).
3. **Gray display**: `plt.imshow(gray, cmap="gray", vmin=0, vmax=255)`; without vmin/vmax Matplotlib auto-stretches contrast and hides gamma/contrast effects.
4. **uint8 overflow**: `img + 50` wraps (260 -> 4); `cv2.add` saturates at 255. Use float for math, then `np.clip(..., 0, 255).astype(np.uint8)`.
5. **Kernel sizes odd** (3, 5, 31); `GaussianBlur(img, (k,k), 0)` sigma 0 = auto.
6. **Matrices float32** for `warpAffine` / `getPerspectiveTransform` points.
7. **Rotation sign**: positive angle = counter-clockwise in `getRotationMatrix2D`. A hand-built `[[cos,-sin],[sin,cos]]` looks clockwise on screen (y points down).
8. **warpAffine output size** is `(width, height)`; rotated/sheared/scaled content outside the canvas is cut - enlarge the canvas or use the auto-fit templates.
9. **equalizeHist / threshold need a single-channel uint8** image; for color equalize Y (YCrCb) or V (HSV).
10. **Point order** for perspective: TL, TR, BR, BL in both source and destination.
11. **Affine vs projective**: affine = 2x3, 3 point pairs, parallel lines stay parallel; projective = 3x3, 4 point pairs, parallel lines can converge. Linear transforms fix the origin (A*0 = 0); translation needs the affine +b term / homogeneous coordinates.
12. `cv2.imshow` does not work inline in Jupyter/Colab - use Matplotlib.

### 5.3 Likely MCQ facts (Lab 2 and Lab 3)

- gamma < 1 -> brighter (expands dark levels); gamma > 1 -> darker; gamma = 1 -> identity. Log -> expands dark, compresses bright. Inverse log -> opposite. Negative = L-1-r.
- Histogram equalization uses the CDF to spread intensities; result histogram is approximately flat; increases global contrast; CLAHE = local, with clip limit to avoid noise amplification.
- Dark image -> histogram on the left; bright -> right; low contrast -> narrow histogram.
- Point operations depend only on the pixel's own value; neighbourhood operations (filters) depend on surrounding pixels.
- Piecewise linear: slopes > 1 stretch a range, < 1 compress it; (r1,s1)=(r2,s2) with s=0/L-1 outside = thresholding; (r1,s1)=(rmin,0), (r2,s2)=(rmax,L-1) = contrast stretching.
- Geometry changes *where* pixels are; photometry changes *what* they look like.
- Affine preserves collinearity, parallelism and ratios along lines; may change angles and lengths. Similarity preserves angles; rigid preserves lengths and angles.
- Matrix columns = landing spots of the x-axis and y-axis. a/d scale, b/c shear, negative a or d = reflection. Pure rotation needs all four entries (cos, -sin, sin, cos).
- Homogeneous coordinates add a "1" so translation becomes matrix multiplication and transforms compose into one master matrix. warpAffine = 2x3, warpPerspective = 3x3.
- Projective: denominator h31x + h32y + h33 varies with the point; this creates perspective/depth.
